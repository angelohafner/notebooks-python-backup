```python
import numpy as np
import matplotlib.pyplot as plt

def plot_vectors_and_projection(v1, v2):
    # Calcula a projeção do vetor v1 sobre o vetor v2
    v2_norm = np.dot(v2, v2)
    projection = (np.dot(v1, v2) / v2_norm) * v2

    # Plota os vetores v1 e v2
    plt.figure()
    plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='r', label=r'$\vec u$')
    plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='b', label=r'$\vec v$')

    # Plota a projeção de v1 sobre v2
    plt.quiver(0, 0, 0.99*projection[0], 1.01*projection[1], angles='xy', scale_units='xy', scale=1, color='g', label=r'${{\vec u}_{\vec v}}$')

    # Configura os limites do gráfico
    plt.xlim(0, 5)
    plt.ylim(0, 5)
    # plt.axhline(0, color='black',linewidth=0.5)
    # plt.axvline(0, color='black',linewidth=0.5)
    plt.grid()
    plt.legend()
    plt.show()

# Exemplo de vetores
v2 = np.array([3, 4])
v1 = np.array([1, 4])

plot_vectors_and_projection(v1, v2)
```
