```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
tempo = np.linspace(0, 1/60, 128, endpoint=False)
voltagem = 1.5 * np.sin(2*np.pi*60*tempo)
amperagem = 1.2 * np.sin(2*np.pi*60*tempo-30*np.pi/180)
potencia = voltagem * amperagem
```

```python
plt.plot(tempo, voltagem, label='$v(t)$')
plt.plot(tempo, amperagem, label='$i(t)$')
plt.plot(tempo, potencia, label='$p(t)$')
plt.grid()
plt.legend()
```

```python
import sympy as sp

# Definindo as variáveis
x = sp.symbols('x')

# Definindo a equação quadrática: ax^2 + bx + c = 0
a, b, c = 5, -3, 2
equacao = a*x**2 + b*x + c

# Resolvendo a equação
solucoes = sp.solve(equacao, x)

solucoes[0]
```
