```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
t = sp.symbols("t")
Q = 10000*sp.ln(1+0.02*t)
Q_d = sp.diff(Q, t)
```

```python
Q_d
```

```python
Q_f = sp.lambdify(t, Q)
Q_d_f = sp.lambdify(t, Q_d)
```

```python
tempo = np.linspace(0, 1000, 100)
plt.plot(tempo, Q_f(tempo))
```

```python
plt.plot(tempo, Q_d_f(tempo))
```
