```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
v = np.array([[1],
              [0]])
```

```python
theta = 60*np.pi/180
T = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta), np.cos(theta)]])
T
```

```python
w = T @ v
w
```

```python
plt.figure(figsize=(4,4), dpi=200)
plt.xlim(-2, 2)
plt.ylim(-2, 2)
plt.grid()
plt.quiver(0, 0, v[0], v[1], scale_units='xy', angles='xy', scale=1, color='blue', label='v')
plt.quiver(0, 0, w[0], w[1], scale_units='xy', angles='xy', scale=1, color='red', label='T(v)')
plt.legend()
```
