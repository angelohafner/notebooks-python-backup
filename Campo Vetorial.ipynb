```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5, 5, 10)
y = x
X, Y = np.meshgrid(x,y)
```

```python
u = Y
v = X
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(10,10)
fig.set_dpi = 600
ax.scatter(X, Y, label='dispersão')
ax.quiver(X, Y, u, v)
ax.legend()
fig.show()
```
