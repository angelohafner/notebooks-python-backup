### (a) Tensão no restante do grupo

$${V_{{\rm{GRUPO}}}} = \frac{{6P}}{{6S\left( {P - F} \right) + 5F}}$$

### (b) Corrente de desequilíbrio

$${I_{{\rm{DESEQUILIBRIO}}}} = \frac{{3F}}{{6S\left( {P - F} \right) + 5F}}{I_{N - 1{\rm{PERNA}}}}$$


<blockquote>
$P$ = Número de unidades em paralelo por grupo

$S$ = Número de grupos série por perna

$F$ = Número de fusíveis queimados
</blockquote>

Dado um banco de capacitores de 3600 kVAr, instalado em
13.8 kV, em um local em que o nível de curto-circuito é de 8519
A. A característica do banco é apresentada a seguir. Pede-se para
ajustar as proteções do banco.
Conexão: dupla estrela não aterrada
kVAr de 1 unidade = 200
Número de grupos série por fase (S) = 1
Número de unidades em paralelo por grupo (P) = 3
kVn 1 unidade = 7,967
TC de fase = 400-5 A
TC do neutro da dupla estrela = 15-5 A

```python
import numpy as np
```

```python
S_banco = 3600e3
V_linha = 13.8e3
I_curto = 8519
```

```python
VAr_unidade = 200e3
```

```python
nr_ser = 1
nr_par = 3
volts_unidade = V_linha/np.sqrt(3) / nr_ser
RTC_fase = 400/5
RTC_neutro = 15/5
```

```python
I_nominal_banco = S_banco / (np.sqrt(3)*V_linha)
I_nominal_banco
```

```python
I_n_menos_1perna = I_nominal_banco / (nr_par-1)
I_n_menos_1perna
```

```python
I_tc_fase = 1.2*I_nominal_banco / RTC_fase
I_tc_fase / 5
```

```python
lista_fusiveis = [0, 1, 2, 3]
lista_V_grupo = []
lista_I_deseq = []
for nr_fus in lista_fusiveis:
    V_grupo = 6 * nr_par / (6*nr_ser*(nr_par-nr_fus)+5*nr_fus)
    I_deseq = 3 * nr_fus / (6*nr_ser*(nr_par-nr_fus)+5*nr_fus) * I_n_menos_1perna
    lista_V_grupo.append(V_grupo)
    lista_I_deseq.append(I_deseq)

```

```python
import pandas as pd

# Criando um DataFrame
df = pd.DataFrame({'Fusíveis': lista_fusiveis, 'Tensoes (pu)': lista_V_grupo, 'Desequilíbrio (A)': lista_I_deseq})
df
```

```python
I_ajuste = 0.9**2 * np.array(lista_I_deseq) / RTC_neutro
I_ajuste
```
