```python
import sympy as sp
```

```python
def rotacional(Fx, Fy, Fz):
    x, y, z = sp.symbols("x, y, z")
    rot_x = sp.diff(Fz,y) - sp.diff(Fy,z)
    rot_y = sp.diff(Fx,z) - sp.diff(Fz,x)
    rot_z = sp.diff(Fy,x) - sp.diff(Fx,y)
    return [rot_x, rot_y, rot_z]
```

```python
x, y, z = sp.symbols("x, y, z")
```

```python
Fx = x**2
Fy = -x*z
Fz = -y**2
```

```python
rot_x, rot_y, rot_z = rotacional(Fx, Fy, Fz)
rot_x, rot_y, rot_z
```
