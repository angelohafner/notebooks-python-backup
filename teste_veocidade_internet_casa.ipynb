```python
import speedtest
import time
from datetime import datetime

INTERVAL = 900  # segundos (15 minutos)

while True:
    try:
        st = speedtest.Speedtest()
        st.get_best_server()
        download = st.download() / 1_000_000  # Mbps
        upload = st.upload() / 1_000_000      # Mbps
        ping = st.results.ping
        timestamp = datetime.now().strftime('%Y-%m-%d %H:%M:%S')

        with open("log_internet.csv", "a") as f:
            f.write(f"{timestamp},{ping:.2f},{download:.2f},{upload:.2f}\n")

        print(f"{timestamp} OK: {download:.2f} Mbps ↓ / {upload:.2f} Mbps ↑ / {ping:.2f} ms")
    except Exception as e:
        print(f"{datetime.now()} ERRO: {e}")

    time.sleep(INTERVAL)
```
