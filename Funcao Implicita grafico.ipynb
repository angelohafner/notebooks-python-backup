```python
from matplotlib import pyplot as plt
import numpy as np
```

$$y' =  - \frac{{2x + {y^2}}}{{2xy}}$$


$$2x + {y^2} + 2xy\frac{{dy}}{{dx}} = 0$$

```python
xrange = np.linspace(-10, 10, 1000)
yrange = np.linspace(-10, 10, 1000)

xrange2 = np.linspace(-10, 10, 11)
yrange2 = np.linspace(-10, 10, 11)
```

```python
x, y = np.meshgrid(xrange, yrange)
x2, y2 = np.meshgrid(xrange2, yrange2)
```

```python
z2 = -(2*x2+y2**2) / (2*x2*y2)
ang = np.arctan(z2)
uu = np.cos(ang)
vv = np.sin(ang)
```

```python
z = x*y**2 + x**2
plt.figure(figsize=(4,4), dpi=300)
cs = plt.contour(x, y, z)
plt.clabel(cs, cs.levels, inline=True)
plt.quiver(x2, y2, uu, vv, angles='xy')
```
