```python
import sympy as sp
sp.init_printing()
```

```python
# Matriz das impedâncias das fases
Z_abc = sp.Matrix([['Zp', 'Zm', 'Zm'],
                   ['Zm', 'Zp', 'Zm'],
                   ['Zm', 'Zm', 'Zp']
])
```

```python
angulo = sp.pi*sp.Rational(2,3)
a = sp.cos(angulo) + sp.I*sp.sin(angulo)
# Matriz de transformação de Fortescue
F = sp.Matrix([[1, 1,    1], 
               [1, a**2, a], 
               [1, a,    a**2]])

F_inv = sp.Rational(1,3) * sp.Matrix([[1, 1, 1], 
                                      [1, a, a**2], 
                                      [1, a**2, a]])
```

$$\begin{array}{l}
\left[ {{V_{abc}}} \right] = \left[ {{Z_{abc}}} \right]\left[ {{I_{abc}}} \right]\\
\left\{ {{{\left[ F \right]}^{ - 1}}\left[ {{V_{012}}} \right]} \right\} = \left[ {{Z_{abc}}} \right]\left\{ {{{\left[ F \right]}^{ - 1}}\left[ {{I_{012}}} \right]} \right\}\\
\left[ {{V_{012}}} \right] = \underbrace {\left[ F \right]\left[ {{Z_{abc}}} \right]{{\left[ F \right]}^{ - 1}}}_{\left[ {{Z_{012}}} \right]}\left[ {{I_{012}}} \right]\\
\left[ {{Z_{012}}} \right] = \left[ F \right]\left[ {{Z_{abc}}} \right]{\left[ F \right]^{ - 1}}
\end{array}$$

```python
Z_012 = sp.simplify(F_inv * Z_abc * F)
Z_012
```
