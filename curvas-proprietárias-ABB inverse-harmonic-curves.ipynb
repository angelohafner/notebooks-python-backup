```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider, VBox

# Default curve parameters
A_default = 28.200
B_default = 0.1217
C_default = 2.00
D_default = 29.10
E_default = 1.0
Ip_default = 1.0

def abb_harmonic_curve(I, Ip, A, B, C, D, E):
    x = (I / Ip)**C - 1.0
    t = np.full_like(I, np.inf, dtype=float)
    mask = x > 0.0
    t[mask] = (A / x[mask]) + B + D * E
    return t

# X axis fixed from 1 to 10 pu
I_values = np.linspace(1.001, 100.0, 4000)

# Y axis fixed manually (adjust if needed)
YMIN = 1.
YMAX = 1e4
XMIN = 0
XMAX = 10

def update_plot(A, B, C, D, E, Ip):
    t_values = abb_harmonic_curve(I_values, Ip, A, B, C, D, E)

    plt.figure(figsize=(7, 5))
    plt.plot(I_values, t_values)
    plt.xlabel("Current (pu)")
    plt.ylabel("Operating time (s)")
    plt.title("ABB Harmonic Curve (A–E Parameters + Ip)")

    plt.yscale("log")
    plt.ylim(YMIN, YMAX)   # <--- FIXED SCALE
    plt.xlim(XMIN, XMAX)   # <--- FIXED SCALE

    plt.grid(True, which="major", linewidth=1.0)
    plt.grid(True, which="minor", linestyle="--", linewidth=0.6)
    plt.show()

ui = VBox([
    FloatSlider(min=0.1, max=60.0, step=0.1, value=A_default, description='A'),
    FloatSlider(min=0.0, max=1.0, step=0.01, value=B_default, description='B'),
    FloatSlider(min=0.5, max=4.0, step=0.05, value=C_default, description='C'),
    FloatSlider(min=0.0, max=60.0, step=0.1, value=D_default, description='D'),
    FloatSlider(min=0.0, max=2.0, step=0.1, value=E_default, description='E'),
    FloatSlider(min=0.5, max=5.0, step=0.1, value=Ip_default, description='Ip')
])

interact(update_plot,
         A=ui.children[0],
         B=ui.children[1],
         C=ui.children[2],
         D=ui.children[3],
         E=ui.children[4],
         Ip=ui.children[5])
```
