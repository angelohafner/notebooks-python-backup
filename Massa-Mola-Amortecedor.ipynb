# Massa - Mola - Amortecedor com Força Externa


Definindo a EDO

$$m x{\left(t \right)} + b \frac{d}{d t} x{\left(t \right)} + k \frac{d^{2}}{d t^{2}} x{\left(t \right)} = f(t) $$

```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
```

```python
t = sp.symbols('t')
t
```

```python
x = sp.Function('x')(t)
x
```

```python
#fext = 0
fext = 3*sp.cos(2.4*t)
fext
```

Forma como consegui tornar os números simbólicos.
Se encontrar outra mais fácil me avisem...

```python
m1 = 9.9
b1 = 19.8
k1 = 258.7
```

Condições iniciais

```python
cis= {x.subs(t, 0): 2, x.diff(t).subs(t, 0): 3}
cis
```

```python
edo = sp.Eq(m1*x.diff(t,t) + b1*x.diff(t) + k1*x , fext )
sp.latex(edo)
```

Encontrando a solução da EDO

```python
xSol = sp.dsolve(edo, func=x, ics=cis)
(xSol)
```

```python
xSolDireita = xSol.args[1]
xSolDireita
```

Concertendo a expressão numa forma que se possa calucular para fazer o gráfico.

```python
solNum = sp.lambdify(t, xSolDireita)
```

```python
fig = plt.figure(dpi=300, figsize=(26/2.54,9/2.54))
ax1 = plt.axes()
ax1.set_xlabel("$t$")
ax1.set_ylabel("$x$")
tt = np.linspace(0,6,10000)
ax1.plot(tt,solNum(tt), lw=0.5)
ax1.grid(alpha=0.5)
plt.show()
```
