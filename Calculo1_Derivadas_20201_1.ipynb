1)	Um carro está viajando à noite ao longo de uma rodovia em forma de parábola com seu vértice na origem (veja a figura). O carro parte em um ponto 100 m a oeste e 100 m ao norte da origem e viaja na direção leste. Há uma estátua localizada 100 m a leste e 50 m ao norte da a origem. Em que ponto da rodovia os faróis do carro iluminarão a estátua?

```python
import matplotlib.pyplot as plt
import numpy as np

a = 1/100
```

```python
x = np.linspace(-50,50,100)
parab = a * x**2
plt.plot(x,parab)
```

$$y=mx+n$$
$$y=2x^2+n$$
$$n=y-2x^2$$


```python
plt.plot(x,parab)
m = 50/100
x_interseccao = 100*m/2
n = 100 - 2/100*x_interseccao**2
m_reta = 2/100*x_interseccao
reta = m_reta*x + n
plt.plot(x,reta)
plt.grid(color='0.95')
plt.scatter(50,100)
print(m)
print(n)
```
