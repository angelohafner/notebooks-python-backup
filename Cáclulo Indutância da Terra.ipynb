Sabendo que 
$$L = \frac{\lambda }{I}$$

e considerando a densidade de corrente na terra dada pela expressão:

$${\nabla ^2}\vec J - j\sigma \mu \omega \vec J = 0$$

$$J\left( {r,t} \right) = {J_0}\exp \left( { - \frac{r}{\delta }} \right)\cos \left( {\omega t - \frac{r}{\delta }} \right)$$

pegando somente o módulo 

$$J\left( {r} \right) = {J_0}\exp \left( { - \frac{r}{\delta }} \right)$$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
f, I, L, J0, r, d, mu0 = sp.symbols('lambda, I, L, J_0, r, delta, mu_0')
```

```python
I = J0 * sp.exp(-r/d) * (r**2) / ((10*d)**2)
I
```

```python
num = mu0*I
den = 2*sp.pi*r
integrando = num/den
integrando
```

```python
sp.integrate(integrando, (r, 0, 10*d), d>0)
```
