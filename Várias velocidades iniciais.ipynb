```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

# EDOs de primeria ordem

Fricção do ar ao cair

$$ \frac{d}{dt}v(t) = g -\frac{\alpha}{m} v^2(t)  \hspace{10mm} v(0) = 0$$

Observe que não há dependência do tempo neste exemplo em particular. Precisamos escrever esta equação diferencial em **forma python**

```python
def dvdt(t, v):
    return 10 - 0.1*v**2
```

```python
tt = np.linspace(0, 4, 100000)
fig, ax = plt.subplots()
for v0 in range(-10, 20, 5):
    sol_m1 = odeint(dvdt, y0=v0, t=tt, tfirst=True)
    v_sol_m1 = sol_m1.T[0]
    ax.plot(tt, v_sol_m1, label='$v_0=$'+str(v0))
ax.set_ylabel('$v(t)$', fontsize=22)
ax.set_xlabel('$t$', fontsize=22)
ax.grid()
ax.legend()
fig.show()
```

```python
import sympy as sp
t = sp.Symbol("t")
v = sp.Function("v")
m = 10
b = sp.Rational(1,10)
g = 10
```

```python
edo = sp.Eq(m * v(t).diff(t), g -  b*(v(t))**2 )
edo
```
