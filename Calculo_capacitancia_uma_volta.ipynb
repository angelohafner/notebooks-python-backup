```python
import sympy as sp
import sympy.physics.units as u

def calc_area(r_externo, r_interno):
    if r_interno > r_externo:
        raise ValueError("O raio interno deve ser menor ou igual ao raio externo.")
    return sp.pi * (r_externo**2 - r_interno**2)

def calc_capacitancia(A, d):
    epsilon = 8.85e-12 * u.farad/u.meter
    return epsilon * A / d

def extract_coefficient(expr, unit):
    """Extract coefficient from an expression with unit."""
    simplified_expr = sp.simplify(expr)  # Simplifica a expressão
    coefficient = (simplified_expr / unit).evalf()
    return float(coefficient)


# Entrando com os valores em centímetros usando prefixo
r_externo = 30 * u.centimeter
r_interno = 22 * u.centimeter
d = 30 * u.centimeter

A = calc_area(r_externo, r_interno)
A_cm2 = A  # Já que a área será em centímetros quadrados

print(f"\nÁrea da arruela: {extract_coefficient(A_cm2, u.centimeter**2):.4e} cm²")

C = calc_capacitancia(A, d)
print(f"Capacitância entre as arruelas: {extract_coefficient(C, u.farad):.4e} F")
```
