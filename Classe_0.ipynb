```python
import math

# Definição da classe Circulo
class Circulo:
    def __init__(self, raio, x=0, y=0):
        """
        Inicializa o círculo com raio e centro (x, y).
        :param raio: Raio do círculo (deve ser positivo)
        :param x: Coordenada x do centro (padrão 0)
        :param y: Coordenada y do centro (padrão 0)
        :raises ValueError: Se o raio for menor ou igual a zero
        """
        if raio <= 0:
            raise ValueError("O raio deve ser positivo.")
        self.raio = float(raio)
        self.x = float(x)
        self.y = float(y)

    def area(self):
        """Calcula a área do círculo."""
        return math.pi * self.raio ** 2

    def circunferencia(self):
        """Calcula a circunferência do círculo."""
        return 2 * math.pi * self.raio

    def _distancia(self, px, py):
        """Calcula a distância entre o centro do círculo e um ponto (px, py)."""
        return math.sqrt((px - self.x) ** 2 + (py - self.y) ** 2)

    def contem_ponto(self, px, py):
        """Verifica se um ponto (px, py) está dentro do círculo."""
        return self._distancia(px, py) <= self.raio

    def mover(self, novo_x, novo_y):
        """Move o centro do círculo para as novas coordenadas (novo_x, novo_y)."""
        self.x = float(novo_x)
        self.y = float(novo_y)

    def __str__(self):
        """Retorna uma representação em string do círculo."""
        return f"Círculo com raio {self.raio} e centro ({self.x}, {self.y})"

    def __eq__(self, outro):
        """Verifica se dois círculos são iguais (mesmo raio e mesmo centro)."""
        if isinstance(outro, Circulo):
            return self.raio == outro.raio and self.x == outro.x and self.y == outro.y
        return False

# Exemplo de uso da classe Circulo
def main():
    # Criando dois círculos
    circulo1 = Circulo(5, 0, 0)  # Raio 5, centro (0, 0)
    circulo2 = Circulo(3, 2, 2)  # Raio 3, centro (2, 2)

    # Imprimindo informações dos círculos
    print("Informações iniciais:")
    print(circulo1)
    print(circulo2)

    # Calculando e imprimindo a área e a circunferência
    print("\nCálculos:")
    print(f"Área do círculo 1: {circulo1.area():.2f}")
    print(f"Circunferência do círculo 1: {circulo1.circunferencia():.2f}")
    print(f"Área do círculo 2: {circulo2.area():.2f}")
    print(f"Circunferência do círculo 2: {circulo2.circunferencia():.2f}")

    # Verificando se pontos estão dentro dos círculos
    ponto1 = (1, 1)
    ponto2 = (4, 5)
    print("\nVerificação de pontos:")
    print(f"O ponto {ponto1} está dentro do círculo 1? {'Sim' if circulo1.contem_ponto(*ponto1) else 'Não'}")
    print(f"O ponto {ponto2} está dentro do círculo 1? {'Sim' if circulo1.contem_ponto(*ponto2) else 'Não'}")
    print(f"O ponto {ponto1} está dentro do círculo 2? {'Sim' if circulo2.contem_ponto(*ponto1) else 'Não'}")
    print(f"O ponto {ponto2} está dentro do círculo 2? {'Sim' if circulo2.contem_ponto(*ponto2) else 'Não'}")

    # Movendo o centro do círculo 2
    print("\nMovendo o círculo 2:")
    circulo2.mover(3, 3)
    print(f"Após mover, {circulo2}")

    # Comparando dois círculos
    circulo3 = Circulo(5, 0, 0)  # Mesmo raio e centro que circulo1
    print("\nComparação de círculos:")
    print(f"O círculo 1 é igual ao círculo 3? {'Sim' if circulo1 == circulo3 else 'Não'}")
    print(f"O círculo 1 é igual ao círculo 2? {'Sim' if circulo1 == circulo2 else 'Não'}")

if __name__ == "__main__":
    main()
```
