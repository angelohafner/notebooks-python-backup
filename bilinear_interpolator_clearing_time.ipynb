```python
# Comments in English only

import numpy as np
from scipy.interpolate import RegularGridInterpolator

# x-axis: fault current (kA)
x = np.array([5.0, 10.0, 15.0, 20.0])

# y-axis: system type (number of CALs)
y = np.array([4.0, 8.0, 12.0])

# z matrix: maximum clearing time (cycles)
# Rows -> x (fault current)
# Columns -> y (system)
z = np.array([
    [20.9, 41.7, 62.6],
    [8.8,  17.6, 26.5],
    [5.2,  10.4, 15.7],
    [3.5,   7.1, 10.6]
])

# Create bilinear interpolator
interp = RegularGridInterpolator(
    (x, y),
    z,
    method="linear",
    bounds_error=False,
    fill_value=None
)

# Example: interpolate at I_fault = 12 kA, system = 3-CAL
point = np.array([[12.0, 3.0]])
z_interp = interp(point)

print(f"Interpolated clearing time: {z_interp[0]:.3f} cycles")
```
