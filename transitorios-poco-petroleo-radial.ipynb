# Solução analítica da EDP da difusão aplicada a poços de petróleo (caso radial 1D, reservatório infinito)

Vamos considerar o **modelo clássico de engenharia de reservatórios**: escoamento radial em um meio poroso infinito, com fluido levemente compressível, poço produzindo a **vazão constante** $q$.

---

## 1. Equação da difusão (equação de difusividade radial)

Admitindo:

- meio poroso homogêneo e isotrópico  
- escoamento monofásico, levemente compressível  
- escoamento puramente radial (independente de $\theta$ e $z$)  

obtém-se a **equação de difusividade radial** para a pressão $p(r,t)$:

$$
\frac{\partial p}{\partial t}
=
\eta\,
\frac{1}{r}\,\frac{\partial}{\partial r}
\left(
r\,\frac{\partial p}{\partial r}
\right),
\qquad r>0,\ t>0
$$

onde a **difusividade hidráulica** é

$$
\eta = \frac{k}{\varphi\,\mu\,c_t},
$$

com:

- $k$ = permeabilidade da rocha  
- $\varphi$ = porosidade  
- $\mu$ = viscosidade do fluido  
- $c_t$ = compressibilidade total  

---

## 2. Condições físicas para um poço produtor em reservatório infinito

Assumimos:

- Reservatório infinito, inicialmente na pressão uniforme $p_i$:
  $$
  p(r,0) = p_i, \qquad 0 < r < \infty
  $$

- Poço localizado em $r=0$, produzindo vazão constante $q$ [m³/s].  
  Esse poço é modelado como uma **fonte linear** (na verdade, um sumidouro) concentrada em $r=0$.

- Longe do poço, a pressão permanece na pressão inicial:
  $$
  p(r,t)\to p_i \quad \text{quando} \quad r\to\infty,\ \forall t>0
  $$

Definimos a **depressão de pressão**:

$$
p'(r,t) = p(r,t) - p_i.
$$

Então a EDP em termos de $p'(r,t)$ é a mesma:

$$
\frac{\partial p'}{\partial t}
=
\eta\,
\frac{1}{r}\,\frac{\partial}{\partial r}
\left(
r\,\frac{\partial p'}{\partial r}
\right),
$$

com condição inicial

$$
p'(r,0) = 0,
$$

e condição no infinito

$$
p'(r,t) \to 0 \quad \text{quando} \quad r\to\infty.
$$

O efeito do poço (vazão $q$) aparece como uma **condição de fluxo** em $r\to 0$:

$$
q = -\,\frac{2\pi\,r\,h\,k}{\mu}\,\frac{\partial p}{\partial r}
\quad\Rightarrow\quad
\frac{\partial p'}{\partial r}
\sim
-\,\frac{q\,\mu}{2\pi\,h\,k}\,\frac{1}{r}
\quad \text{quando} \quad r\to 0.
$$

---

## 3. Solução fundamental (Green) da difusão radial

Para a EDP de difusão radial em 2D (radial em torno de uma linha), a solução fundamental de um **pulso instantâneo unitário** em $r=0$ é:

$$
G(r,t) = \frac{1}{4\pi\,\eta\,t}\,
\exp\left(-\,\frac{r^2}{4\,\eta\,t}\right),
\qquad t>0.
$$

Se o poço produz a **vazão constante** $q$ a partir de $t=0$, podemos superpor infinitos pulsos infinitesimais de vazão. A depressão de pressão é dada por uma convolução em tempo:

$$
p'(r,t) = -\,\frac{q\,\mu}{h\,k}
\int_0^t G(r, t-\tau)\,d\tau.
$$

Substituindo $G(r,t-\tau)$:

$$
p'(r,t)
=
-\,\frac{q\,\mu}{h\,k}
\int_0^t
\frac{1}{4\pi\,\eta\,(t-\tau)}\,
\exp\left(-\,\frac{r^2}{4\,\eta\,(t-\tau)}\right)
d\tau.
$$

Fazendo a mudança de variável

$$
u = \frac{r^2}{4\,\eta\,(t-\tau)}
\quad\Rightarrow\quad
d\tau = -\,\frac{r^2}{4\,\eta}\,\frac{du}{u^2},
$$

os limites de integração em $\tau\in[0,t]$ correspondem a $u\in\left[\frac{r^2}{4\eta t},\ \infty\right]$.  
O integral em $\tau$ se transforma em um integral da forma:

$$
\int_{\frac{r^2}{4\eta t}}^{\infty} \frac{e^{-u}}{u}\,du.
$$

Este integral define a **integral exponencial** $E_1(x)$:

$$
E_1(x) = \int_x^{\infty} \frac{e^{-u}}{u}\,du.
$$

Logo, a solução para a depressão de pressão é:

$$
p'(r,t) =
p(r,t) - p_i
=
-\,\frac{q\,\mu}{4\pi\,k\,h}\,
E_1\!\left(
\frac{r^2}{4\,\eta\,t}
\right).
$$

---

## 4. Solução analítica final (forma de Theis)

Usando a expressão de $\eta = \dfrac{k}{\varphi\,\mu\,c_t}$, podemos escrever o argumento da integral exponencial como:

$$
\frac{r^2}{4\,\eta\,t}
=
\frac{r^2\,\varphi\,\mu\,c_t}{4\,k\,t}.
$$

Assim, a **solução analítica da EDP de difusão radial para um poço de petróleo produzindo vazão constante em reservatório infinito** fica:

$$
p(r,t) =
p_i
-
\frac{q\,\mu}{4\pi\,k\,h}\,
E_1\!\left(
\frac{r^2\,\varphi\,\mu\,c_t}{4\,k\,t}
\right),
\qquad r>0,\ t>0.
$$

- $p_i$ = pressão inicial do reservatório  
- $q$ = vazão de produção do poço (m³/s, nas condições do reservatório)  
- $h$ = espessura da zona produtora  
- $E_1(\cdot)$ = integral exponencial de primeira ordem  

Essa expressão é conhecida na engenharia de reservatórios como a **solução de Theis** (ou solução de linha-fonte) e é amplamente usada em **testes de poços**, análise de pressão transiente e diagnóstico de comportamento de reservatórios.

---

## 5. Forma adimensional usual

É conveniente introduzir as variáveis adimensionais:

- **Tempo adimensional**:
  $$
  t_D = \frac{k\,t}{\varphi\,\mu\,c_t\,r_w^2}
  $$

- **Raio adimensional**:
  $$
  r_D = \frac{r}{r_w}
  $$

- **Pressão adimensional (drawdown)**:
  $$
  p_D(r_D,t_D) = \frac{2\pi\,k\,h}{q\,\mu}\,\bigl(p_i - p(r,t)\bigr).
  $$

Nessas variáveis, a solução pode ser escrita de forma compacta:

$$
p_D(r_D,t_D) =
\frac{1}{2}\,
E_1\!\left(\frac{r_D^2}{4\,t_D}\right).
$$

ou, conforme convenções de alguns autores, com fatores numéricos ligeiramente diferentes (dependendo da definição de $p_D$ escolhida). O ponto central é que a **dependência em $r$ e $t$ aparece combinada como $\dfrac{r^2}{4\eta t}$, via a integral exponencial**.

````python
# ================================================
# THEIS SOLUTION ANIMATION
# ================================================

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from mpmath import e1
from IPython.display import HTML, Image

# ---------------------------------------------
# 1) Physical parameters
# ---------------------------------------------
p_i = 30e6          # [Pa] initial reservoir pressure
q = 0.01            # [m^3/s] production rate
k = 1e-13           # [m^2] permeability
phi = 0.20          # porosity
mu = 1e-3           # [Pa.s] viscosity
c_t = 1e-9          # [1/Pa] compressibility
h = 10.0            # [m] reservoir thickness

r_w = 0.1           # [m] well radius
r_max = 1000.0      # [m] plotting limit
n_r = 400           # number of radial points

# Linear radial grid
r = np.linspace(r_w, r_max, n_r)

# ---------------------------------------------
# 2) Time list for animation (more frames)
# ---------------------------------------------
t_hours_list = np.logspace(-1, 7, 1000)  # 120 frames
t_list = t_hours_list * 3600.0

# ---------------------------------------------
# 3) Theis analytical solution
# ---------------------------------------------
def theis_pressure(r_array, t):
    """Compute Theis analytical pressure for time t."""
    factor = (q * mu) / (4.0 * np.pi * k * h)
    x = (r_array**2 * phi * mu * c_t) / (4.0 * k * t)
    E1_vec = np.vectorize(lambda z: float(e1(z)))
    return p_i - factor * E1_vec(x)

# ---------------------------------------------
# 4) Create figure for animation
# ---------------------------------------------
fig, ax = plt.subplots(figsize=(16, 4))

line, = ax.plot([], [], lw=2, color="orange")

ax.set_xlim(r_w, r_max)
ax.set_ylim(0, p_i + 1e6)

ax.set_xlabel("Radius r [m]")
ax.set_ylabel("Pressure p(r,t) [Pa]")
ax.set_title("Pressure Diffusion (Theis Solution)")

# Vertical line marking the well
well_line = ax.axvline(r_w, linestyle="--", color="gray")

# Horizontal line marking initial pressure
p_init_line = ax.axhline(p_i, linestyle=":", color="blue")

# ---------------------------------------------
# 5) Update function
# ---------------------------------------------
def update(frame):
    t = t_list[frame]
    p = theis_pressure(r, t)
    line.set_data(r, p)
    ax.set_title(f"Pressure Diffusion (t = {t_hours_list[frame]:.2f} h)")
    return line, well_line, p_init_line

# ---------------------------------------------
# 6) Build and save animation
# ---------------------------------------------
anim = FuncAnimation(fig, update, frames=len(t_list), interval=120)

gif_path = "theis_animation_linear_wide.gif"
anim.save(gif_path, writer=PillowWriter(fps=15))

print("GIF saved as:", gif_path)

# ---------------------------------------------
# 7) Display GIF in notebook
# ---------------------------------------------
Image(gif_path)
````
