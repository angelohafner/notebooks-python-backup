```python
import numpy as np
import matplotlib.pyplot as plt
```

# Dados

```python
kappa = 1
pb = 1000
pa = np.arange(0, 500, 10)
L = 1
mu = 0.12
A = 10e-4
```

##Equação

$$Q = \frac{{\kappa A}}{\mu }\frac{{\left( {{p_b} - {p_a}} \right)}}{L}$$

## Cálculo

```python
Q = kappa * A * (pb - pa) / ( mu * L )
```

```python
plt.plot(pa, Q)
plt.ylabel('$Q(p_a)$')
plt.xlabel('$p_a$')
plt.grid()
```
