```python
import numpy as np
import matplotlib.pyplot as plt

# Definição de parâmetros
f = 60  # frequência em Hz
w = 2 * np.pi * f  # velocidade angular
T = 1 / f  # período em segundos

# Geração dos pontos de tempo
t = np.linspace(0, 2*T, 1000)  # 1000 pontos de dados em um ciclo

v = 1 * np.sin(w * t)
ia = 8 * np.sin(w * t)
ir = 6 * np.sin(w * t - np.pi/2)
ir2 = 6 * np.sin(w * t + np.pi/2)
i = ia + ir

# Cálculo das funções
P = ia * v
Q = ir * v
P_plus_Q = P + Q

# Plotagem dos gráficos
plt.figure(figsize=(10, 6), facecolor='none')  # Adiciona fundo transparente para a figura
plt.plot(1e3*t, P, label=r'$P \; \rm [MW]$', color="blue")
plt.plot(1e3*t, Q, label=r'$Q \; \rm [MVAr]$', color="red")
plt.plot(1e3*t, P_plus_Q, label=r'$S \; \rm [MVA]$', color="gray")

plt.xlabel('Tempo (ms)')
plt.ylabel('Potência')
plt.legend(framealpha=0)  # Torna a legenda transparente
plt.gca().patch.set_facecolor('none')  # Adiciona fundo transparente para a área de plotagem
plt.show()

# Para salvar a imagem com fundo e legenda transparentes, use:
# plt.savefig('nome_da_imagem.png', transparent=True)
```
