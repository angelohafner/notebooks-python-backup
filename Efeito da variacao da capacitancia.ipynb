```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from engineering_notation import EngNumber
```

```python
freq = np.arange(60, 1200, 1)
ww = 2*np.pi*freq
LL = np.logspace(np.log10(10e-6), np.log10(10e-6 * 1000), num=5)
w0 = 11*2*np.pi*60
CC = 1 / ((w0**2)*LL)
```

```python
XL = np.zeros((len(LL), len(freq)))
XC = np.zeros((len(LL), len(freq)))

for ii in range(len(LL)):
    XL[ii,:] = ww * LL[ii]
    XC[ii,:] = 1 / ( ww * CC[ii] )

ZZ = XL - XC
for ii in range(len(LL)):
    plt.plot(freq/60, np.abs(ZZ[ii,:]), label= str(EngNumber(CC[ii]))+"F")
    plt.legend()
    plt.ylim(0, 1)
    
plt.grid(ls=":", alpha=0.5)
```
