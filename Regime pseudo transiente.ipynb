```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from matplotlib.ticker import LinearLocator
import imageio
from google.colab import files  # Importa a função para download de arquivos
```

```python
def R(raio, rho=1):
    r = np.linspace(0, raio, 100)
    a = np.linspace(0, 2*np.pi, 100)
    rr, aa = np.meshgrid(r, a)
    return np.exp(-rr/rho)
```

```python
def T(t, tau=1):
    return np.exp(-t/tau)
```

```python
rho = 1
r = np.linspace(0, 1, 100)
a = np.linspace(0, 2*np.pi, 100)
rr, aa = np.meshgrid(r, a)

```

```python

pp = ( rr + 0.5 )
levels = np.arange(0.0, np.max(pp)+0.1, 0.1)
filenames = []

for ii in np.arange(0.5, 0.1, -0.01):
    pp = (rr**2 + ii)
    fig, ax = plt.subplots()
    X = rr * np.cos(aa)
    Y = rr * np.sin(aa)
    Z = pp
    CS = ax.contourf(X, Y, Z, levels, cmap='jet')
    ax.axis('equal')
    fig.colorbar(CS, ax=ax)

    filename = f'frame_{ii:.2f}.png'
    fig.savefig(filename)
    filenames.append(filename)
    plt.close(fig)

# Cria um GIF animado
with imageio.get_writer('meu_gif_animado.gif', mode='I') as writer:
    for filename in filenames:
        image = imageio.imread(filename)
        writer.append_data(image)

# Opcional: remover os arquivos de imagem
#for filename in filenames:
#    os.remove(filename)

# Exibe o arquivo GIF (no Colab, isso pode não funcionar diretamente como visualização)
# from IPython.display import Image
# Image(open('meu_gif_animado.gif','rb').read())

# Faz o download do GIF para o seu sistema local
files.download('meu_gif_animado.gif')
```
