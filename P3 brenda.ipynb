```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

Neste primeiro código, estamos importando uma biblioteca que seja capaz de rodar nossos códigos,como gráficos,computação númerica e simbólica.

##$$v1(x)=asen(\frac{x^2}{a})$$

Este Vetor acima, foi o primeiro vetor que escolhi para fazer a combinação linear, abaixo estou exemplificando ele em código python.

```python
a_num = 2
x_num = np.linspace(0,3*np.pi)
v1 = a_num*np.sin((x_num**2)/a_num)
plt.plot(x_num, v1)
```

Neste código acima chamei a letra "a" de "a_num", "a" é a variável que está multiplicando seno, e dividindo a componenente x, dei o valor de 2 ao "a".

O "x" da função, chamei de "x_num", e no valor dado a ele, usei o código linspace para fazer o intervalo de 0 a 3*pi, ou seja, o x possui valores de 0 a aproximadamente 9,5.

E no "v1" especifiquei a função mostrada anteriormente, só que em códigos e na linha de baixo, plotei um grafico com x_num sendo o x e v1 sendo o y.

No gráfico podemos perceber como os valores de y realmente variam de 2 a -2, pois é o valor 2 que demos ao "a" que multiplica o sen, e também podemos perceber o intervalo de x, que vai de o a aproximadamente 9,5 como proposto no "X_num".

##$$v2(x)=acos(\frac{2a}{x})$$

Este vetor acima, foi o segundo vetor que escolhi para fazer a combinação linear,abaixo ele está em código phyton.

```python
v2 = a_num*np.cos(2*a_num/x_num)
plt.plot (x_num, v2)
```

Neste código acima, não dei valor a "a" e nem a "x", pois ja havia dado na função anterior, então nesta primeira linha esta a função acima, e na linha de baixo é plotado um gráfico, com x_num sendo nosso x, e v2 nossa função.



##$$Vt(x) = asen(\frac{x^2}{a}) + acos(\frac{2a}{x})$$

Nesta terceira função, estamos fazendo uma combinação linear de V1 e V2, ou seja estamos somando v1 ao v2 e criando um 3º vetor, por isto combinação.

'Combinação linear é a representação de um vetor a partir de operações com outros vetores', que é o que está representado acima e também no gráfico abaixo.

```python
vt = v1 + v2
fig, ax = plt.subplots(dpi=150)
ax.plot(x_num, v1, label = r'$a\sin(\frac{x^2}{a})$')
ax.plot(x_num, v2, label = r'$a\cos(\frac{2a}{x})$')
ax.plot(x_num, vt, label = r'$a\sin(\frac{x^2}{a})+a\cos(\frac{2a}{x})$')
ax.fill_between(x_num, vt, alpha = 0.25)
plt.legend()
```

Acima foi plotado um gráfico, com a função v1, v2 e a combinação delas.
Percebemos que a função 'vt', realmente é uma combinação das duas pois quando temos no grafico do seno um valor positivo e no grafico do cosseno o valor negativo ou vice-versa o grafico da funçao soma das duas permanece no intervalo entre elas, quando temos a funcao seno e cosseno com os mesmos sinais, ou seja, seno positivo e cosseno positivo, ou seno negativo e cosseno negativo, o grafico da funcao soma das duas tem um valor maior que o maior valor de uma delas.
