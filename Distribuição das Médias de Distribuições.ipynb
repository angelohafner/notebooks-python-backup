```python
import numpy as np
import plotly.express as px
import pandas as pd
import random as rd
```

```python
xx = np.arange(1., 100001., 1)
yy = 10000*np.exp(-5*xx/100000) + 500
```

```python
df = pd.DataFrame(dict(x = xx, 
                       y = yy))

df
```

```python
fig = px.line(df, x="x", y="y", 
              title="Salário do Indivíduo #",
              width=1200, height=400)
fig.update_layout(
    xaxis_title=r'$x$',
    yaxis_title=r'$\exp \left(-5x \right) + 500$')

fig.show()
```

```python
fig = px.histogram(df, x=yy,
              title="Histograma",
              width=1200, height=400)
fig.update_layout(
    xaxis_title=r'Classe do Salário',
    yaxis_title=r'Quantidade de Indivíduos da Classe')
fig.show()
```

```python
nr_pessoas_para_media = 100
nr_total_populacao = len(yy)
nr_medias_a_gerar = 20000
```

```python
media = np.zeros(nr_medias_a_gerar)
for i in range(nr_medias_a_gerar):
    soma = 0
    for k in range(nr_pessoas_para_media):
        pessoa_nr = rd.randint(0, 99999)
        salario_pessoa_escolhida = yy[pessoa_nr]
        soma = soma + salario_pessoa_escolhida
    media[i] = soma / nr_pessoas_para_media
```

```python
xxx = np.arange(1, nr_medias_a_gerar+1, 1)
yyy = media
df3 = pd.DataFrame(dict(media_nr = xxx, salario_medio = yyy))
```

```python
fig3 = px.line(df3, x="media_nr", y="salario_medio", 
               title="Média dos Salários tomados de 40 em 40 pessoas",
               width=1200, height=400)
fig3.show()
```

```python
fig = px.histogram(df3, x=media,
              title="Histograma dos Salários Médios",
              width=1200, height=400)
fig.update_layout(
    xaxis_title=r'Classe',
    yaxis_title=r'Quantidade')
fig.show()
```
