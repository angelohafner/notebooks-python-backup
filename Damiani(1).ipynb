```python
import matplotlib
import numpy as np
import matplotlib.cm as cm
import matplotlib.pyplot as plt
```

```python
delta = 0.001
k = 1e-11
x = np.arange(-3, 3.0, delta)
c = np.arange(-1, 1, delta)
X, C = np.meshgrid(x, c)
Z1 = 4*X**3 + k*X - k*C
cs = plt.contour(X,C,Z1,[0])
```

```python
p = cs.collections[0].get_paths()[0]
XXX = p.vertices[:,0]
CCC = p.vertices[:,1]
```

```python
plt.figure(dpi=150)
plt.plot(XXX,CCC)
plt.xlabel('$x$')
plt.ylabel('$c$')
plt.grid()
```
