```python
import numpy as np
from cmath import *
import math
import matplotlib.pyplot as plt
```

```python
a = complex(math.cos(2*np.pi/3),math.sin(2*np.pi/3))
```

```python
T = np.array([[1,1,1],[1,a**2,a],[1,a,a**2]])
T
```

```python
v = 100 * np.array([[1],[a**2],[a]])
v
```

```python
fig, ax = plt.subplots()
ax.grid(linestyle='--', linewidth=0.5)

x_pos = [0, 0, 0]
y_pos = [0, 0, 0]
x_direct = v.real
y_direct = v.imag

ax.quiver(x_pos,y_pos,x_direct,y_direct, scale=250)
ax.axis([-100,100,-100,100])
ax.set_aspect('equal')

plt.show()
```

```python
u = np.matmul(T,v)
u = np.array(u)
u
```

```python
zero = u[0] * np.array([[1], [1],[1]])
zero
```

```python
fig, ax = plt.subplots()
ax.grid(linestyle='--', linewidth=0.5)

x_pos = [0, 0, 0]
y_pos = [0, 0, 0]
x_direct = zero.real
y_direct = zero.imag

ax.quiver(x_pos,y_pos,x_direct,y_direct, scale=25)
ax.axis([-100,100,-100,100])
ax.set_aspect('equal')

plt.show()
```

```python
pos = u[1] * np.array([[1], [a**2],[a]])
pos
```

```python
fig, ax = plt.subplots()
ax.grid(linestyle='--', linewidth=0.5)

x_pos = [0, 0, 0]
y_pos = [0, 0, 0]
x_direct = pos.real
y_direct = pos.imag

ax.quiver(x_pos,y_pos,x_direct,y_direct, scale=25)
ax.axis([-100,100,-100,100])
ax.set_aspect('equal')

plt.show()
```

```python
neg = u[2] * np.array([[1], [a**2],[a]])
neg
```

```python
fig, ax = plt.subplots()
ax.grid(linestyle='--', linewidth=0.5)

x_pos = [0, 0, 0]
y_pos = [0, 0, 0]
x_direct = neg.real
y_direct = neg.imag

ax.quiver(x_pos,y_pos,x_direct,y_direct, scale=600)
ax.axis([-100,100,-100,100])
ax.set_aspect('equal')


plt.show()
```
