```python
import sympy as sp
x = sp.symbols('x')
y_original = x**2 + x + 1
derivada = sp.diff(y_original, x)
derivada
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.arange(-5, 5, 0.1)
y_original_numpy = x**2 + x + 1
derivada_numpy = 2*x + 1
plt.plot(x, y_original_numpy, label='$x^2+x+1$', color='green')
plt.plot(x, derivada_numpy, label='$2x+1$', color='red')
plt.legend()
plt.grid(True)
plt.show()
```
