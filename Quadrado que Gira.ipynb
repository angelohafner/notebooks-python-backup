```python
import matplotlib.pyplot as plt
import matplotlib.animation
import numpy as np
plt.rcParams["animation.html"] = "jshtml"
plt.ioff()
```

```python
x = np.array([-1, 1, 1, -1, -1])
y = np.array([-1, -1, 1, 1, -1])
v = np.array([x,y])
v.shape
```

```python
plt.scatter(v[0,:], v[1,:])
plt.plot(v[0,:], v[1,:])
plt.show()
```

```python

fig, ax = plt.subplots()
for ang in range(0, 10, 1):
    M = [[np.cos(ang/10), - np.sin(ang/10)],
            [np.sin(ang/10), np.cos(ang/10)]]
    M = np.array(M)
    w = M @ v
    ax.plot(w[0,:], w[1,:])
    
fig.show()
```

```python
fig = plt.figure()

def animate(k):
    plt.cla()
    ang = k /100
    M = [[np.cos(ang), - np.sin(ang)],
            [np.sin(ang), np.cos(ang)]]
    M = np.array(M)
    w = M @ v
    plt.plot(w[0,:], w[1,:])
    plt.xlim(-1.5, 1.5)
    plt.ylim(-1.5, 1.5)
    
matplotlib.animation.FuncAnimation(fig, animate, frames=157, interval=15.7)
```
