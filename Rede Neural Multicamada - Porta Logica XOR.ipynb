```python
import numpy as np
```

```python
def sigmoide(soma):
    return 1 / (1 + np.exp(-soma))
```

### Derivada da função sigmóide
$$\frac{d}{{dx}}\left( {\frac{1}{{1 + \exp \left( { - x} \right)}}} \right) = \frac{1}{{1 + \exp \left( { - x} \right)}}\left( {1 - \frac{1}{{1 + \exp \left( { - x} \right)}}} \right)$$

$$\frac{{ds}}{{dx}} = s\left( {1 - s} \right)$$



```python
def sigmoideDerivada(sig):
    return sig * (1-sig)
```
