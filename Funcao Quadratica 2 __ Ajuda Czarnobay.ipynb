```python
#!pip install tensorflow
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import activations
```

```python
xxx = np.linspace(-50., 50., 100)
yyy = xxx**2
```

```python
plt.scatter(xxx, yyy)
```

```python
tf.random.set_seed(42)
# 1. Criar o modelo
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1]),
    tf.keras.layers.Dense(units=1000, activation="ELU"),
    tf.keras.layers.Dense(units=100, activation="ELU"),
    tf.keras.layers.Dense(units=10, activation="ELU"),
    tf.keras.layers.Dense(units=1, activation="ELU"),
])
```

```python
# 2. Compilar o modelo
model.compile(loss='mae',
              optimizer=tf.keras.optimizers.Adam(0.001),
              metrics=['mae','mse','accuracy']
              ) 
```

```python
# 3. Treinar o modelo
history = model.fit(xxx, yyy, epochs=1000, verbose=False)
print("Finished training the model")
```

```python
plt.xlabel('Época #')
plt.ylabel("Erro")
plt.plot(history.history['loss'])
```

```python
xx = np.linspace(-100, 100, 200)
plt.plot(xx,model.predict([xx]))
plt.plot(xx, xx**2, ls='dashed')
plt.grid()
plt.show()
```
