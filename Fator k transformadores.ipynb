```python
import numpy as np

def calculate_fhl(harmonic_orders, harmonic_currents):
    """
    Calculate the Harmonic Loss Factor (FHL).

    Parameters:
    harmonic_orders (list of int): Orders of the harmonics.
    harmonic_currents (list of float): Proportion of harmonic currents to the fundamental current.

    Returns:
    float: Calculated FHL.
    """
    # Calculate the numerator sum of h^2 * (I_h / I_1)^2
    numerator = sum((h ** 2) * (I_h ** 2) for h, I_h in zip(harmonic_orders, harmonic_currents))

    # Calculate the denominator sum of (I_h / I_1)^2
    denominator = sum(I_h ** 2 for I_h in harmonic_currents)

    # FHL is the ratio of numerator to denominator
    FHL = numerator / denominator
    return FHL

def calculate_pll(I_pu, FHL, PEC_R_pu):
    """
    Calculate the Total Loss Factor under Load (PLL).

    Parameters:
    I_pu (float): Per-unit RMS current.
    FHL (float): Harmonic Loss Factor.
    PEC_R_pu (float): Per-unit Eddy Current Loss under nominal conditions.

    Returns:
    float: Calculated PLL.
    """
    # Calculate PLL using the provided formula
    PLL = I_pu ** 2 + FHL * PEC_R_pu
    return PLL

def calculate_max_capacity_fraction(PLL):
    """
    Calculate the fraction of maximum capacity of the transformer.

    Parameters:
    PLL (float): Total Loss Factor under Load.

    Returns:
    float: Fraction of maximum capacity that can be used safely.
    """
    # Maximum capacity fraction is the inverse of PLL
    max_capacity_fraction = 1 / PLL
    return max_capacity_fraction

def calculate_k_factor(harmonic_orders, harmonic_currents):
    """
    Calculate the k-Factor for the transformer.

    Parameters:
    harmonic_orders (list of int): Orders of the harmonics.
    harmonic_currents (list of float): Proportion of harmonic currents to the fundamental current.

    Returns:
    float: Calculated k-Factor.
    """
    # Calculate k-Factor as the sum of h^2 * (I_h / I_1)^2 without normalization
    k_factor = sum((h ** 2) * (I_h ** 2) for h, I_h in zip(harmonic_orders, harmonic_currents))
    return k_factor

# Data for the practical example
harmonic_orders = [1, 5, 7, 11, 13, 17, 19]  # Harmonic orders
harmonic_currents = [1.00, 0.23, 0.11, 0.042, 0.027, 0.013, 0.008]  # Harmonic current ratios
PEC_R_pu = 0.15  # Per-unit eddy current loss
I_pu = 1  # Per-unit current

# Calculations
FHL = calculate_fhl(harmonic_orders, harmonic_currents)
PLL = calculate_pll(I_pu, FHL, PEC_R_pu)
max_capacity_fraction = calculate_max_capacity_fraction(PLL)
k_factor = calculate_k_factor(harmonic_orders, harmonic_currents)
max_current = max_capacity_fraction * 1200  # Maximum current in amps (based on 1200 A nominal capacity)

# Displaying the results
print(f"Harmonic Loss Factor (FHL): {FHL:.3f}")
print(f"Total Loss Factor under Load (PLL): {PLL:.3f}")
print(f"Maximum Capacity Fraction: {max_capacity_fraction:.2%}")
print(f"Maximum Allowable Current: {max_current:.2f} A")
print(f"k-Factor: {k_factor:.3f}")
```
