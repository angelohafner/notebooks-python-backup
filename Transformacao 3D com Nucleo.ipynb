```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-2, 3, 1)
y = np.arange(-2, 3, 1)
z = np.arange(-2, 3, 1)
```

```python
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')
```

```python
T = [[1, 2, -1],
        [0, 1, 2],
        [1, 3, 1]]
T = np.array(T)
T         
```

```python
v = np.zeros((3, len(x)**3))
```

```python

hh = 0
for ii in range(len(x)):
    for jj in range(len(y)):
        for kk in range(len(z)):
            v[:,hh] = np.array([x[ii],
                                        y[jj],
                                        z[kk]])
            hh = hh + 1
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.scatter(v[0], v[1], v[2])
```

```python
w = T @ v
```

```python
fig2 = plt.figure()
ax2 = plt.axes(projection='3d')
ax2.scatter(v[0], v[1], v[2])
ax2.scatter(w[0], w[1], w[2])
```
