A equação vetorial para o lançamento de projéteis é uma representação que descreve a posição de um projétil em movimento em função do tempo. Essa equação leva em conta tanto a componente horizontal quanto a componente vertical do movimento.

Para um projétil lançado com uma velocidade inicial $\mathbf{v_0}$ em um ângulo $\theta$ em relação à horizontal, a equação vetorial da posição $\mathbf{r}(t)$ do projétil em função do tempo $t$ é dada por:

$$
\mathbf{r}(t) = \mathbf{r_0} + \mathbf{v_0} t + \frac{1}{2} \mathbf{a} t^2
$$

Aqui:

- $\mathbf{r}(t)$ é o vetor posição no tempo $t$.
- $\mathbf{r_0}$ é o vetor posição inicial (ponto de lançamento).
- $\mathbf{v_0}$ é o vetor velocidade inicial, que pode ser decomposto em componentes horizontal e vertical: $\mathbf{v_0} = v_0 \cos(\theta) \, \hat{i} + v_0 \sin(\theta) \, \hat{j}$.
- $\mathbf{a}$ é a aceleração vetorial. No caso de movimento sob a influência da gravidade, $\mathbf{a} = -g \, \hat{j}$, onde $g$ é a aceleração devido à gravidade (aproximadamente $9,81 \, \text{m/s}^2$ para a Terra).

Especificando as componentes, temos:

$$
\mathbf{r}(t) = \mathbf{r_0} + (v_0 \cos(\theta) \, \hat{i} + v_0 \sin(\theta) \, \hat{j}) t + \frac{1}{2} (-g \, \hat{j}) t^2
$$

Isso pode ser separado nas componentes horizontal e vertical:

- **Componente horizontal:**
  $$
  x(t) = x_0 + v_0 \cos(\theta) t
  $$

- **Componente vertical:**
  $$
  y(t) = y_0 + v_0 \sin(\theta) t - \frac{1}{2} g t^2
  $$

Aqui, $x_0$ e $y_0$ são as coordenadas iniciais, geralmente $x_0 = 0$ e $y_0 = 0$ se o projétil for lançado do solo.

Essas equações vetoriais são fundamentais para a análise do movimento de projéteis em diversas áreas da engenharia e física.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0, 10, 0.1)
theta = 30
v0 = 50
x0 = 0
y0 = 0
g = 9.81
for theta in range(0, 91, 5):
    x = x0 + v0 * np.cos(theta* np.pi / 180) * t
    y = y0 + v0 * np.sin(theta* np.pi / 180) * t - 0.5 * g * t**2
    plt.plot(x, y, label=f'{theta}')
plt.legend()
plt.ylim(0, 150)
plt.xlim(0, 300)
plt.show()
```
