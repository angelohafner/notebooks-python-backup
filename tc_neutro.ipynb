```python
class NeutralCTProtectionSelector:
    """
    A class to select the closest standard protection current transformer (CT).
    """

    def __init__(self,
                 i_secondary=5,
                 accuracy_limit_factor=20,
                 default_class="10P20",
                 default_burden="50 VA"):
        """
        Initialize the ProtectionCTSelector with optional secondary current,
        accuracy limit factor, default class, and burden.
        """
        # Standard primary currents for protection CTs
        self.standard_currents = [
            5, 10, 15, 20, 25, 30, 40, 50, 75, 100, 150, 200,
            300, 400, 500, 600, 800, 1000, 1250, 1500, 2000,
            2500, 3000, 4000, 5000
        ]
        # Default class and burden
        self.default_class = default_class
        self.default_burden = default_burden
        self.i_secondary = i_secondary
        self.accuracy_limit_factor = accuracy_limit_factor

    def __str__(self):
        """
        Return a string representation of the ProtectionCTSelector instance.
        """
        return (f"ProtectionCTSelector(\n"
                f"  Default Class: {self.default_class}\n"
                f"  Default Burden: {self.default_burden}\n"
                f"  Secondary Current: {self.i_secondary} A\n"
                f"  Accuracy Limit Factor: {self.accuracy_limit_factor}\n"
                f"  Standard Primary Currents: {self.standard_currents}\n)")

    def select_closest(self, i_rated):
        """
        Select the closest standard primary current for protection CTs.

        Parameters:
            i_rated (float): The input rated current value in Amperes.

        Returns:
            dict: A dictionary with the closest primary current, class, burden, and secondary current.
        """
        closest = min(self.standard_currents, key=lambda x: abs(x - i_rated))
        return closest


# Example usage
selector = NeutralCTProtectionSelector( i_secondary=1,
                                        accuracy_limit_factor=20,
                                        default_class="10P20",
                                        default_burden="50 VA"
)

input_current = 320
result = selector.select_closest(input_current)

```
