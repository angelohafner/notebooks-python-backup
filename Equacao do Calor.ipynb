$${\frac{k}{h}\lambda \cos \left( {\lambda L} \right) = \sin \left( {\lambda L} \right)}$$

```python
def func(lamb, k=401, h=200, L=5):
    zzz = k/h * lamb * np.cos(lamb*L) - np.sin(lamb*L)
    return zzz
```

```python
lamb = np.linspace(0, 10, 100000)
yyy = func(lamb, k=401, h=200, L=5)
plt.figure(figsize=(14,5))
plt.plot(lamb, yyy)
plt.grid()
```
