```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-1, 1, 100000)
y = np.sin(1/x)
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(20,5)
ax.set_xlabel('eixo $x$')
ax.set_ylabel('eixo $y = \\sin \\left( \\dfrac{1}{x} \\right)$')
ax.plot(x, y, lw=0.5)
ax.grid(alpha=0.5, ls='--')
fig.show()
```

```python
import pandas as pd
dados = np.array([x , y])
dados = dados.T
dados
```

```python
df = pd.DataFrame(dados, columns=['x','y'])
df
```
