```python
# Neural network (non-linear) that learns y = x² using original x
# Comments in English, as requested.

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Data
x_dado = np.linspace(0, 2, 11).astype(np.float32).reshape(-1, 1)
y_calc = (x_dado ** 2)

# Model: one hidden layer with ReLU, output layer linear
model = tf.keras.Sequential([
    tf.keras.layers.Dense(8, activation='relu', input_shape=(1,)),  # hidden layer
    tf.keras.layers.Dense(1)                                        # output layer
])

model.compile(optimizer=tf.keras.optimizers.Adam(1e-2), loss='mse')

# Train
history = model.fit(x_dado, y_calc,
                    epochs=1000,
                    verbose=0
                    )

# Predictions
predicted = model.predict(x_dado)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(x_dado, y_calc,  '--', label='True')
plt.plot(x_dado, predicted, '-', label='Predicted')
plt.xlabel('x_dado')
plt.ylabel('y_calc')
plt.legend()
plt.grid(True)
plt.show()
```
