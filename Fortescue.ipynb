```python
from sympy import *
init_printing()
```

```python
a = cos(Rational(2,3)*pi) + Mul((sin(Rational(2,3)*pi)),sqrt(-1))
```

```python
F = Matrix([[1,1,1],[1,Pow(a,2),a],[1,a,Pow(a,2)]])
F
```

```python
Zs, Zm = symbols('Z_s,Z_m')
Zabc = Matrix([[Zs,Zm,Zm],[Zm,Zs,Zm],[Zm,Zm,Zs]])
Zabc
```

```python
Z012 = Inverse(F) * Zabc * F
simplify(Z012)
```

```python
Z0, Z1, Z2 = symbols('Z_0,Z_1,Z_2')
Z012_volta = Matrix([[Z0,0,0],[0,Z1,0],[0,0,Z2]])

Z012_volta
```

```python
Zabc_volta = F * Z012_volta * Inverse(F)
simplify(Zabc_volta)
```
