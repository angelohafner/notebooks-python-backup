```python
import math

# Constants
omega1 = 2 * math.pi * 60  # Angular frequency (rad/s)
Vtrifasica = 12e3  # Trifasica voltage (V)
Qtrifasica = 14.5e6  # Reactive power (VAR)
Vlata = 8e3  # Voltage per can (V)

# Calculations
Vfase = Vtrifasica / math.sqrt(3)  # Phase voltage (V)
Qfase = Qtrifasica / 3  # Phase reactive power (VAR)
Xfase = (Vfase**2) / Qfase  # Reactance (Ohms)
C = 1 / (omega1 * Xfase)  # Capacitance (F)

# Capacitor Bank Calculations
ns = 1  # Number of series capacitors
np = 8  # Number of parallel capacitors
Clata = (C * np) / ns  # Capacitance per can (F)
ncap = ns * np * 3  # Total number of capacitors
Xlata = 1 / (omega1 * Clata)  # Reactance per can (Ohms)
Qlata = (Vlata**2) / Xlata  # Reactive power per can (VAR)

total_Q = ncap * Qlata  # Total reactive power (VAR)

# Input and Results Dictionaries
data = {
    "inputs": {
        "omega1 (rad/s)": omega1,
        "Vtrifasica (V)": Vtrifasica,
        "Qtrifasica (VAR)": Qtrifasica,
        "Vlata (V)": Vlata
    },
    "results": {
        "Vfase (kV)": Vfase / 1e3,
        "Qfase (MVAr)": Qfase / 1e6,
        "Xfase (Ohms)": Xfase,
        "C (μF)": C * 1e6,
        "Clata (μF)": Clata * 1e6,
        "ncap": ncap,
        "Xlata (Ohms)": Xlata,
        "Qlata (kVAr)": Qlata / 1e3,
        "Total Reactive Power (MVAr)": total_Q / 1e6
    }
}

# Output
print(data)
```
