###	Young-Laplace equation 

In what follows, we now evaluate the contact angle θ in eq. (01) through the augmented Young-Laplace equation, given by: 

\begin{equation}
1 - \cos \theta  = \frac{1}{\sigma }\left( {{P_c}{h_0} + \int_{{h_0}}^\infty  {\Pi {\rm{dh}}} } \right)
\tag{4}
\end{equation}

>$\sigma_\ \rightarrow$ oil/brine interfacial tension, \\
>$P_c\ \rightarrow$ capillary pressure $(P_c=P_o-P_w)$, \\
>$h_0\rightarrow$ equilibrium width, \\
>$\mathrm{\Pi}\rightarrow$ interaction pressure. \\

Hence, to evaluate the contact angle, first we need the interaction pressure and from it evaluate the equilibrium width $h_0$.


### Decomposition of the pressure

We have used the interaction pressure $\mathrm{\Pi}$ decomposed as in the DLVO theory, which is called Derjaguin pressure, given by:

\begin{equation}
\mathrm{\Pi}\ \left(h\right)=\mathrm{\Pi}_{EDL}\left(h\right)+\mathrm{\Pi}_{VdW}\left(h\right)+\mathrm{\Pi}_S\left(h\right)
\tag{5}
\end{equation}

The terms on the right-hand side of (5) are algebraically expressed by:


\begin{equation}
\mathrm{\Pi}_{EDL}=\varepsilon_0\varepsilon_3\kappa^2\left[\frac{2\psi_1\psi_2\cosh{\left(\kappa h\right)}-\left(\psi_1^2+\psi_2^2\right)}{2\sinh^2{\left(\kappa h\right)}}\right],
\tag{6}
\end{equation}

\begin{equation}
\mathrm{\Pi}_{VdW}\left(h\right)=-\frac{A}{6\pi h^2}.
\tag{7}
\end{equation}

\begin{equation}
\mathrm{\Pi}_S\left(h\right)=A_se^{-\frac{h}{h_s}}
\tag{8}
\end{equation}


and they represent the double layer electric field, the Van der Waals forces, and the structural forces, respectively. 

NOTE $-$ Electrical double layer forces can act as repulsive or attractive forces depending on the surface potentials of the two interfaces and the separating distance $(h)$.

#### Pressure at the Eletric Double Layer 

In order to obtain the pressure of the electric double layer, we should first evaluate the electric potential from the Gouy-Chapman model at any point of the system given by the following partial differential equation: 

\begin{equation}
{\nabla ^2}\psi  = \sum\limits_i^{} {\left( {\dfrac{{z_i^2{e^2}{n_{io}}}}{{{\varepsilon _0}{\varepsilon _3}{k_B}T}}} \right)} \psi 
\tag{9}
\end{equation}

where:

> $\psi  \to $ electric potential $\rm (V)$ \\
> ${n_{io}} \to $ number of ions of each $i$ species  (dimensionless) \\
> ${e} \to $ electronic charge $(160.2176634 \times 10^{−21}{\rm C})$ \\
> ${z_i} \to $ valency of $i$-type ions (dimensionless) \\
> ${\varepsilon _0} \to $ vacuum permittivity $\left( {8.85418781762039 \times {{10}^{ - 12}}\,\frac{{{{\rm{C}}^{\rm{2}}}}}{{{\rm{Jm}}}}} \right)$ \\
> ${\varepsilon _3} \to $ dielectric constant of brine (dimensionless) \\
>$T \to $ temperature $\rm (K)$


For small electric potentials, $(1)$ can be reduced as follows:
\begin{equation}
{\nabla ^2}\psi  = \kappa \, \psi 
\tag{10}
\end{equation}
where $\kappa$ is given by:

\begin{equation}
\kappa  = \sqrt {\frac{{2 \times {{10}^3}{N_A}}}{{{\varepsilon _0}{\varepsilon _3}{k_B}}}\left( {\frac{I}{T}} \right)}
\tag{11}
\end{equation}

where:
> $N_A \to$ Avogadro's number $\left( {602.214076 \times {{10}^{ - 12}}\,\dfrac{{\rm{1}}}{{{\rm{mol}}}}} \right)$ \\
> $I \to$ ionic strength $\left( {\dfrac{{{\rm{mol}}}}{{\rm{L}}}} \right)$ \\
> $\kappa \to$ constant $\left( {{{\rm{m}}^{ - 1}}} \right)$





## Poisson-Boltzmann Equation

The Poisson-Boltzmann equation is a mathematical equation used to describe the behavior of electric charge in a solution. It predicts how charges are distributed in a solution, and how they interact with each other and with charged surfaces. It is commonly used to study the behavior of ions in solutions, such as in electrolytes, biological systems, and electrochemical systems.

The electric potential is obtained from the Gouy-Chapman model at any point in the system given by:
\begin{equation}
{\nabla ^2}\psi  = \sum\limits_i^{} {\left( {\dfrac{{z_i^2{e^2}{n_{io}}}}{{{\varepsilon _0}{\varepsilon _3}{k_B}T}}} \right)} \psi 
\tag{1}
\end{equation}

where:

> $\psi  \to $ electric potential $\rm (V)$ \\
> ${n_{io}} \to $ number of ions of each species $i$ (dimensionless) \\
> ${e} \to $ electronic charge $(160.2176634 \times 10^{−21}{\rm C})$ \\
> ${z_i} \to $ valency of $i$-type ions (dimensionless) \\
> ${\varepsilon _0} \to $ vacuum permittivity $\left( {8.85418781762039 \times {{10}^{ - 12}}\,\frac{{{{\rm{C}}^{\rm{2}}}}}{{{\rm{Jm}}}}} \right)$ \\
> ${\varepsilon _3} \to $ dielectric constant of brine (dimensionless) \\
>$T \to $ temperature $\rm (K)$


For small electric potentials, $(1)$ can be reduced as follows:
\begin{equation}
{\nabla ^2}\psi  = \kappa \, \psi 
\tag{10}
\end{equation}
where $\kappa$ is given by:

\begin{equation}
\kappa  = \sqrt {\frac{{2 \times {{10}^3}{N_A}}}{{{\varepsilon _0}{\varepsilon _3}{k_B}}}\left( {\frac{I}{T}} \right)}
\tag{11}
\end{equation}

where:
> $N_A \to$ Avogadro's number $\left( {602.214076 \times {{10}^{ - 12}}\,\dfrac{{\rm{1}}}{{{\rm{mol}}}}} \right)$ \\
> $I \to$ ionic strength $\left( {\dfrac{{{\rm{mol}}}}{{\rm{L}}}} \right)$ \\
> $\kappa \to$ constant $\left( {{{\rm{m}}^{ - 1}}} \right)$

The general solution of $(10)$ in cartesian coordinates along the $𝑧$-axis is given by:

\begin{equation}
\psi \left( z \right) = A\cosh \left( {\kappa z} \right) + B\sinh \left( {\kappa z} \right)
\tag{12}
\end{equation}

where $𝐴$ and $𝐵$ are constants given by the boundary conditions. 

For flat surfaces, the boundary conditions consider the potential constant $\psi_1$ on plate 1 and, $\psi_2$ on plate 2, as shown in Figure bellow.

Therefore, from Figure above, the boundary conditions can be expressed by:

\begin{equation}
\psi \left( {z = 0} \right) = {\psi _1}
\tag{13}
\end{equation}

\begin{equation}
\psi \left( {z = h} \right) = {\psi _2}
\tag{14}

After applying the boundary conditions we have: 

\begin{equation}
\psi  = {\psi _1}\cosh \left( {\kappa z} \right) + \frac{{{\psi _2} - {\psi _1}\cosh \left( {\kappa h} \right)}}{{\sinh \left( {\kappa h} \right)}}\sinh \left( {\kappa z} \right)

It is now possible to evaluate the pressure of the electric layer, since it is simply the derivative of the electric potential, and it is given by:
\begin{equation}
\mathrm{\Pi}_{EDL}=\varepsilon_0\varepsilon_3\kappa^2\left[\frac{2\psi_1\psi_2\cosh{\left(\kappa h\right)}-\left(\psi_1^2+\psi_2^2\right)}{2\sinh^2{\left(\kappa h\right)}}\right].
\tag{16}
\end{equation}


#### Van der Waals Pressure ${\Pi}_{{VdW}}$

The Van der Waals pressure between two planes is given by:
\begin{equation}
\mathrm{\Pi}_{VdW}\left(h\right)=-\frac{A}{6\pi h^3}
\tag{17}
\end{equation}

where $h$ is the separation distance between the planes, and $A$ is the Hamaker constant for the oil/brine/rock system.

The Hamaker constant is a function of the dielectric constants and the refractive indices of the materials for the oil and rock phases interacting with the brine, given by:

\begin{equation}
A=A_{\nu=0}+A_{\nu>0}
\tag{18}
\end{equation}

where:

\begin{equation}
A_{\nu=0}=\frac{3}{4}k_BT\left(\frac{\varepsilon_1-\varepsilon_3}{\varepsilon_1+\varepsilon_3}\right)\left(\frac{\varepsilon_2-\varepsilon_3}{\varepsilon_2+\varepsilon_3}\right)
\tag{19}
\end{equation}

and

\begin{equation}
A_{\nu>0}\approx\frac{3h_p\nu_e}{8\sqrt2}\frac{\left(n_1^2-n_3^2\right)\left(n_2^2-n_3^2\right)}{\left(n_1^2+n_3^2\right)^\frac{1}{2}\left(n_2^2+n_3^2\right)^\frac{1}{2}\left\{\left(n_1^2+n_3^2\right)^\frac{1}{2}+\left(n_2^2+n_3^2\right)^\frac{1}{2}\right\}}
\tag{20}
\end{equation}

where:

>$\varepsilon_1\rightarrow$ oil dielectric constant \\
>$\varepsilon_2\rightarrow$ rock dielectric constant \\
>$\varepsilon_3\rightarrow$ brine dielectric constant \\
>$n_1\rightarrow$ refractive oil index \\
>$n_2\rightarrow$ refractive rock index \\
>$n_3\rightarrow$ refractive brine index \\
>$h_P\rightarrow$ Planck's constant $(6.626070\times{10}^{-34}\ \mathrm{m}^\mathrm{2}\mathrm{\ kg\ /\ s})$ \\
>$\nu_e\rightarrow$ main frequency of electrical adsorption in UV $(3.0\times{10}^{15}\ \rm s^{-1})$ \\


For two equal surfaces, the constants $A_{\nu=0}$ and $A_{\nu>0}$  are always positive, which results in an attractive Van der Waals force. However, for different surfaces, the dielectric constants of the materials and refractive indixes dictate the behavior of the Van der Waals force. 

For a brine/rock/oil system where $\varepsilon_3 > \varepsilon_2 > \varepsilon_1$, and $n_3 < n_2 < n_1$, the Hamaker constant will be positive.

##### Brine Dieletric Constant $\varepsilon_{3}$

Describing the electrical behavior of the brine is difficult due to the numerous variables involved. 

To overcome these challenges, an empirical model based on an interpolation of experimental data was used. 

The correlation function relates the brine dielectric constant with the temperature, pressure, and ionic strength, given by the following relationship:

\begin{equation}
\varepsilon_3=6I\left(0.08I-1\right)-0.165\left(T-273\right)+0.019P+82.5
\tag{21}
\end{equation}

where:

>$P\rightarrow$ pressure in ${\rm MPa}$ \\
>$T\rightarrow$ temperature in ${\rm K}$  \\
>$I\rightarrow$ ionic strength or concentration in ${ \rm \frac{mol}{L} }$

```python
I= 100 # = 0.1 mol/L | the lower the ionic strength the better the oil recovery
T = 300 # Kelvin
eps3 = 71.5 # brine
ec = 160.2176634e-21 # Coulomb
kB = 13.80649e-24 # Boltzmann constant
NA = 602.214076e21 # Avogadro's Number
eps0 = 8.85418781762039e-12 # Farad / metro
h = 10e-12 # metro | balance thickness
psi_1 = -50e-3 # Volts
psi_2 = -50e-3 # Volts

num = 2e3*(ec**2)*NA
den = eps0*eps3*kB*T
kappa = np.sqrt(num/den)

print("kappa =", EngNumber(kappa))
```

```python
z = np.linspace(0, h, 10000)
psi_sol = psi_1*np.cosh(kappa*z) + (psi_2-psi_1*np.cosh(kappa*h))*np.sinh(kappa*z)/np.sinh(kappa*h)
```

## Calculation of the change in free energy $\Delta G$

To calculate the free energy per unit area ($𝐺$) the Verwey and Overbeek equation is used, given by:

\begin{equation}
G = \frac{1}{2}\sigma \psi
\tag{10}
\end{equation}

where $G$ is given in ${\rm \frac{J}{m^2} }$ and $\sigma$ is the superficial charge density, in ${\rm \frac{C}{m^2} }$.

$𝐺_ℎ$ is now calculated from the sum of the free energies per unit of area for each plate separated by a distance $ℎ$,

\begin{equation}
{G_h} = \frac{1}{2}\left( {{\sigma _1}{\psi _1} + {\sigma _2}{\psi _2}} \right)
\tag{11}
\end{equation}

where the surface charge density $\sigma$ of the plate is given by:

\begin{equation}
\sigma  =  - {\varepsilon _0}{\varepsilon _3}\frac{{d\psi }}{{dz}}
\tag{12}
\end{equation}

Thus, replacing relation $(12)$ in $(11)$, we obtain:

\begin{equation}
{G_h} =  - \frac{{{\varepsilon _0}{\varepsilon _3}\kappa }}{{2\sinh \left( {\kappa h} \right)}}\left[ {2{\psi _1}{\psi _2} - \left( {\psi _1^2 + \psi _2^2} \right)\cosh \left( {\kappa h} \right)} \right]
\tag{13}
\end{equation}

To calculate the free energy per unit of area for an infinite distance, we apply the limit $h \to \infty$ in relation $(13)$, which results in:

\begin{equation}
{G_\infty } =  - \frac{{{\varepsilon _0}{\varepsilon _3}\kappa }}{2}\left( {\psi _1^2 + \psi _2^2} \right)
\tag{14}
\end{equation}

Finally, the difference between the free energy per unit of area is given by:

\begin{equation}
\Delta G = {G_h} - {G_\infty } = \frac{{{\varepsilon _0}{\varepsilon _3}\kappa }}{2}\left[ {\frac{{2{\psi _1}{\psi _2} - \left( {\psi _1^2 + \psi _2^2} \right)\exp \left( { - \kappa h} \right)}}{{\sinh \left( {\kappa h} \right)}}} \right]
\tag{15}
\end{equation}
