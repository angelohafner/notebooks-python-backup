# Gráfico de linhas

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 1000)
plt.plot(x, np.exp(x), label="exponencial")
plt.plot(x, np.log(x), label="logaritmo")
plt.legend()
plt.xlabel("eixo $x$")
plt.ylabel("eixo $y$")
plt.title("Gráfico de funções")
plt.grid()
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.show()
```

# Gráfico de Superfície

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 9)
y = np.linspace(-2, 2, 9)
# para fazer gráficos 3d precisa da linha abaixo
x, y = np.meshgrid(x, y)
z = x**2 + y**2

# para fazer graficos 3d preciso das duas linhas abaixo
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z)
plt.show()
```
