```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 100, 1000)
sinal = 10*np.sin(2*np.pi*0.1*t) + 2*np.sin(2*np.pi*0.9*t) + 5*np.cos(2*np.pi*0.5*t)
```

```python
plt.figure(figsize=(20,4))
plt.scatter(t, sinal, s=0.2)
```

```python
freqs = np.arange(0.1, 1.1, 0.1)
div_sin = np.zeros(len(freqs), dtype=float)
div_cos = np.zeros(len(freqs), dtype=float)
ii = 0
```

```python
for kk in freqs:
    # para o seno
    qt_sin_num = np.trapz(sinal*np.sin(2*np.pi*kk*t), t)
    qt_sin_den = np.trapz((np.sin(2*np.pi*kk*t))**2, t)
    div_sin[ii] = qt_sin_num/qt_sin_den
    # para o cosseno
    qt_cos_num = np.trapz(sinal*np.cos(2*np.pi*kk*t), t)
    qt_cos_den = np.trapz((np.cos(2*np.pi*kk*t))**2, t)
    div_cos[ii] = qt_cos_num/qt_cos_den
    print(np.round(kk,2) ,"    ",np.round(div_sin[ii], 2))
    print(np.round(kk,2) ,"    ",np.round(div_cos[ii], 2))
    print("")
    ii = ii + 1
```

```python
plt.figure(figsize=(10,5))
plt.plot(freqs, div_sin, label='seno', ls='--')
plt.plot(freqs, div_cos, label='cosseno', ls='--')
plt.plot(freqs, np.sqrt(div_cos**2 + div_sin**2), alpha=0.3, lw=4, label='TOTAL')
plt.legend()
```
