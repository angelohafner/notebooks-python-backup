```python
import sympy as sy
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = sy.Symbol("x")
T = sy.Symbol("T")
n = sy.Symbol("n", integer=True, positive=True)
x
```

```python
f = x**3
f
```

```python
den_cos = sy.integrate((sy.cos(n*x))**2, (x, -sy.pi, sy.pi))
den_cos
```

```python
den_sin = sy.integrate((sy.sin(n*x))**2, (x, -sy.pi, sy.pi))
den_sin
```

```python
num_cos = sy.integrate((f * sy.cos(n*x)), (x, -sy.pi, sy.pi))
num_cos
```

```python
num_sin = sy.integrate((f * sy.sin(n*x)), (x, -sy.pi, sy.pi))
num_sin
```

```python
alpha_n = num_cos / den_cos
alpha_n
```

```python
beta_n = num_sin / den_sin
beta_n
```

```python
num = sy.integrate((f * T), (x, -sy.pi, sy.pi))
den = sy.integrate((T), (x, -sy.pi, sy.pi))
alpha_0 = num / den
alpha_0
```

```python
x = np.linspace(-np.pi, np.pi, 100)
fy = x**3
```

```python
soma = 0
for n in range(1, 1000, 1):
    alpha = 0
    alpha_cos = alpha * np.cos(n*x)
    beta = ( -(2*(-1)**n)*(np.pi**3)/n + (12*(-1)**n)*(np.pi)/(n**3) ) / np.pi
    beta_sin = beta * np.sin(n*x)
    soma = soma + alpha_cos + beta_sin
```

```python
plt.plot(x, fy)
plt.scatter(x, soma)
```
