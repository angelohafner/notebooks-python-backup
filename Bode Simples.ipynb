```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Definindo a função de transferência H = 1/((s+2)(s+3))
numerator = [1]
denominator = [1, 5, 6]
system = signal.TransferFunction(numerator, denominator)
system_p1 = signal.TransferFunction(numerator, [1, 2])
system_p2 = signal.TransferFunction(numerator, [1, 3])

# Frequências para o diagrama de Bode
frequencies = np.logspace(-1, 2, 500)

# Gerando o diagrama de Bode
w, mag, phase = signal.bode(system, frequencies)
w, mag1, phase1 = signal.bode(system_p1, frequencies)
w, mag2, phase2 = signal.bode(system_p2, frequencies)

# Plotando o diagrama de Bode
plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.semilogx(w, mag)  # Diagrama de magnitude
plt.semilogx(w, mag1)  # Diagrama de magnitude
plt.semilogx(w, mag2)  # Diagrama de magnitude
plt.title('Diagrama de Bode para H = 1/((s+2)(s+3))')
plt.xlabel('Frequência [rad/s]')
plt.ylabel('Magnitude [dB]')
plt.plot([2,2],[-80, -20])
plt.plot([3,3],[-80, -20])

plt.subplot(2, 1, 2)
plt.semilogx(w, phase)  # Diagrama de fase
plt.xlabel('Frequência [rad/s]')
plt.ylabel('Fase [graus]')
plt.plot([2,2],[-180, 0])
plt.plot([3,3],[-180, 0])

plt.tight_layout()
plt.show()
```
