```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go

a = np.linspace(-1.5, 1.5, 200)
b = np.linspace(-1.5, 1.5, 200)
A, B = np.meshgrid(a ,b)

plt.scatter(A, B)
plt.scatter(a, b)
```

```python
X = A + 1j*B
Z = X**2 + 1

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(2*X.real, 2*X.imag, Z.real*0, alpha=0.5)
ax.plot_surface(X.real, X.imag, Z.real, cmap='viridis')
ax.scatter3D(0, -1, 0)
ax.scatter3D(0,  1, 0)


ax.set_xlabel('X.real')
ax.set_ylabel('X.imag')
ax.set_zlabel('Z.real')
```

```python
import plotly.graph_objects as go

# Superfície do plano (2*X.real, 2*X.imag, Z=0)
plane = go.Surface(
    x=2*X.real,
    y=2*X.imag,
    z=0*Z.real,
    opacity=0.5,
    showscale=False,
    colorscale='Greys'
)

# Superfície principal (X.real, X.imag, Z.real)
surf_real = go.Surface(
    x=X.real,
    y=X.imag,
    z=Z.real,
    surfacecolor=np.zeros_like(Z.real),  # cor uniforme
    colorscale=[[0, 'blue'], [1, 'blue']],  # tudo azul
    showscale=False,  # remove barra de cores
    opacity=0.6       # transparência (0=totalmente transparente, 1=opaco)
)

# Pontos (0,-1,0) e (0,1,0)
p1 = go.Scatter3d(x=[0], y=[-1], z=[0], mode='markers',
                  marker=dict(size=6, color='red'), name='Ponto (0,-1,0)')
p2 = go.Scatter3d(x=[0], y=[1], z=[0], mode='markers',
                  marker=dict(size=6, color='blue'), name='Ponto (0,1,0)')

# Criar figura
fig = go.Figure(data=[plane, surf_real, p1, p2])

# Layout e rótulos
fig.update_layout(
    scene=dict(
        xaxis_title='X.real',
        yaxis_title='X.imag',
        zaxis_title='Z.real'
    ),
    width=900, height=600
)

fig.show()
```

```python
# Superfície principal (X.real, X.imag, Z.imag)
surf_imag = go.Surface(
    x=X.real,
    y=X.imag,
    z=Z.imag,
    surfacecolor=np.zeros_like(Z.imag),  # cor uniforme
    colorscale=[[0, 'green'], [1, 'green']],  # tudo azul
    showscale=False,  # remove barra de cores
    opacity=0.6       # transparência (0=totalmente transparente, 1=opaco)
)


# Criar figura
fig = go.Figure(data=[plane, surf_imag, p1, p2])

# Layout e rótulos
fig.update_layout(
    scene=dict(
        xaxis_title='X.real',
        yaxis_title='X.imag',
        zaxis_title='Z.imag'
    ),
    width=900, height=600
)

fig.show()
```

```python
# Superfície principal (X.real, X.imag, Z.imag)
surf_abs = go.Surface(
    x=X.real,
    y=X.imag,
    z=abs(Z),
    surfacecolor=np.zeros_like(Z.imag),  # cor uniforme
    colorscale=[[0, 'red'], [1, 'red']],  # tudo azul
    showscale=False,  # remove barra de cores
    opacity=0.6       # transparência (0=totalmente transparente, 1=opaco)
)


# Criar figura
fig = go.Figure(data=[plane, surf_abs, p1, p2])

# Layout e rótulos
fig.update_layout(
    scene=dict(
        xaxis_title='X.real',
        yaxis_title='X.imag',
        zaxis_title='Z.modulo'
    ),
    width=900, height=600
)

fig.show()
```
