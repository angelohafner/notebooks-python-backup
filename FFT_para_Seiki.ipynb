```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros do sinal
fs = 1000  # Taxa de amostragem (Hz)
t_final = 1.0  # Duração do sinal (segundos)
f1, f2 = 50, 120  # Frequências das senóides (Hz)
A1, A2 = 1.0, 0.5  # Amplitudes

# Vetor tempo
t = np.arange(0, t_final, 1/fs)

# Sinal composto
sinal = A1 + A1 * np.sin(2 * np.pi * f1 * t) + A2 * np.sin(2 * np.pi * f2 * t)

# Salvar o sinal em um arquivo .txt
np.savetxt('sinal.txt', sinal)

# Número de amostras
N = len(sinal)

# FFT
fft_resultado = np.fft.fft(sinal)
frequencias = np.fft.fftfreq(N, d=1/fs)

# Apenas parte positiva
frequencias_positivas = frequencias[:N//2]
amplitudes = np.abs(fft_resultado[:N//2]) * 2 / N
amplitudes[0] = np.abs(fft_resultado[0]) / N  # Correção do componente DC


# Gráfico
plt.figure(figsize=(10, 5))
plt.plot(frequencias_positivas, amplitudes)
plt.xlabel('Frequência (Hz)')
plt.ylabel('Amplitude')
plt.title('FFT do Sinal Gerado')
plt.grid(True)
plt.tight_layout()
plt.show()
```
