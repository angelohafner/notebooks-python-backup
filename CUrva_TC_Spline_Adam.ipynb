```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

# Seus dados
H = np.array([0, 0.17378, 0.25119, 0.40738, 0.57544, 0.91201, 1.58489, 2.63027, 4.16869, 8.20352, 14.32188, 16.2181, 21.87762, 25.23481, 29.24152, 34.67368, 39.81072, 199.52623])
B = np.array([0, 0.00096, 0.00191, 0.0048, 0.00957, 0.01909, 0.04796, 0.09569, 0.19093, 0.47961, 0.95694, 1.15049, 1.44838, 1.58812, 1.65533, 1.73334, 1.74134, 1.90934])

# Definindo a função de Langevin
def langevin(H, Ms, H0):
    return Ms * (1 / np.tanh(H/H0) - H0/H)

# Preparando para o ajuste de curva, ignorando o ponto H = 0 para evitar divisão por zero
popt, _ = curve_fit(langevin, H[1:], B[1:])

# Avaliando a função de Langevin com os parâmetros otimizados
H_new = np.linspace(0.17378, 199.52623, 100)
B_new = langevin(H_new, *popt)

# Plotando os dados originais e a curva ajustada em um gráfico log-log
plt.loglog(H[1:], B[1:], 'o', label='Dados Originais')  # Ignorando o ponto H = 0
plt.loglog(H_new, B_new, '-', label=f'Langevin Fit: Ms={popt[0]:.3f}, H0={popt[1]:.3f}')
plt.legend()
plt.xlabel('H')
plt.ylabel('B')
plt.grid(True, which="both", ls="--")
plt.show()
```
