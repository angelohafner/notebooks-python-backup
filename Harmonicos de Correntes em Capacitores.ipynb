```python
# Importando bibliotecas necessárias
import matplotlib.pyplot as plt

def calcular_corrente_harmonica(percentual_tensao_harmonica, ordem_harmonica):
    """
    Calcula o percentual da corrente harmônica dado o percentual da tensão harmônica e a ordem harmônica.

    Parâmetros:
    - percentual_tensao_harmonica: Percentual da tensão harmônica (em %).
    - ordem_harmonica: Ordem da harmônica (inteiro).

    Retorna:
    - percentual_corrente_harmonica: Percentual da corrente harmônica (em %).
    """
    percentual_corrente_harmonica = ordem_harmonica * (percentual_tensao_harmonica / 100) * 100
    return percentual_corrente_harmonica

def plotar_corrente_vs_tensao(faixa_percentual_tensao, ordens_harmonicas):
    """
    Plota o percentual da corrente harmônica em função do percentual da tensão harmônica para diversas ordens harmônicas.

    Parâmetros:
    - faixa_percentual_tensao: Lista de percentuais de tensão harmônica.
    - ordens_harmonicas: Lista das ordens harmônicas (inteiros).
    """
    plt.figure(figsize=(12, 8))

    for ordem in ordens_harmonicas:
        percentuais_corrente = [calcular_corrente_harmonica(pt, ordem) for pt in faixa_percentual_tensao]
        plt.plot(faixa_percentual_tensao, percentuais_corrente, marker='none', label=f'Ordem {ordem}')

    plt.title('Percentual da Corrente Harmônica vs Percentual da Tensão Harmônica em um Banco de Capacitores')
    plt.xlabel('Percentual da Tensão Harmônica (%)')
    plt.ylabel('Percentual da Corrente Harmônica (%)')
    plt.grid(True)
    plt.legend()
    plt.show()

# Exemplo de uso
ordens_harmonicas_impares = list(range(1, 16, 2))  # Harmônicos ímpares de ordem 3 até 15
faixa_percentual_tensao = list(range(0, 5))  # Percentuais de tensão harmônica de 0% a 10%


```

```python
# Plotando o gráfico
plotar_corrente_vs_tensao(faixa_percentual_tensao, ordens_harmonicas_impares)
```

# Impacto de Harmônicos de Tensão na Corrente de um Capacitor

Para entender o impacto de um harmônico de tensão de 5ª ordem na corrente de um capacitor, precisamos considerar a relação entre a tensão aplicada e a corrente através do capacitor em diferentes frequências harmônicas.

## Relação Básica

Para um capacitor $C$, a corrente $I$ é dada pela relação:

$$ I = C \cdot \frac{dV}{dt} $$

No domínio da frequência, isso se traduz para a relação:

$$ I_h = j \cdot \omega_h \cdot C \cdot V_h $$

Onde:
- $I_h$ é a corrente na frequência harmônica $h$.
- $\omega_h = 2 \pi h f_1$ é a frequência angular da $h$-ésima harmônica.
- $C$ é a capacitância.
- $V_h$ é a tensão na $h$-ésima harmônica.
- $j$ é a unidade imaginária ($j^2 = -1$).

## Análise para a 5ª Harmônica

Para um harmônico de 5ª ordem ($h = 5$):

### Frequência Angular

$$ \omega_5 = 2 \pi \cdot 5 \cdot f_1 $$

### Corrente Harmônica

$$ I_5 = j \cdot (2 \pi \cdot 5 \cdot f_1) \cdot C \cdot V_5 $$

Se a tensão da 5ª harmônica é 3% da tensão fundamental ($V_1$):

$$ V_5 = 0.03 \cdot V_1 $$

Então, a corrente na 5ª harmônica é:

$$ I_5 = j \cdot (2 \pi \cdot 5 \cdot f_1) \cdot C \cdot (0.03 \cdot V_1) $$

## Comparação com a Corrente Fundamental

A corrente fundamental ($I_1$) para a frequência fundamental ($h = 1$) é:

$$ I_1 = j \cdot (2 \pi \cdot f_1) \cdot C \cdot V_1 $$

## Razão das Correntes

A razão entre a corrente da 5ª harmônica e a corrente fundamental é:

$$ \frac{I_5}{I_1} = \frac{j \cdot (2 \pi \cdot 5 \cdot f_1) \cdot C \cdot (0.03 \cdot V_1)}{j \cdot (2 \pi \cdot f_1) \cdot C \cdot V_1} $$

Simplificando:

$$ \frac{I_5}{I_1} = \frac{5 \cdot 0.03}{1} = 0.15 $$

Portanto, a corrente de 5ª ordem será 15% da corrente fundamental do capacitor. Ou seja, um harmônico de tensão de 5ª ordem com 3% da tensão fundamental resulta em uma corrente harmônica de 5ª ordem que é 15% da corrente fundamental do capacitor.
