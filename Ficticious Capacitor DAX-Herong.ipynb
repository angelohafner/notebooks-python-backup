```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
t, w1, v1, v3, v5 = sp.symbols("t omega_1 v_1 v_3 v_5")
sp.init_printing()
```

Imagine a ficticious capacitor $C$, with rated data:

```python
C = sp.Rational(80,1e6) #F
Q_rated = 12e6 #VAr
w_rated = 2*np.pi*60.0 #rad/s
```

Now, imagine that a voltage $v = v_1 + v_3 + v_5$ is applied to its therminals, where:

```python
f1 = 60
w1 = 2*sp.pi*f1
v1 = 20 * sp.sin(1*w1*t) * 1000
v3 =  5 * sp.sin(3*w1*t) * 1000
v5 = 10 * sp.sin(5*w1*t) * 1000
v = v1 + v3 + v5
```

So, the current that will flow will be:

```python
i1 = C * sp.diff(v1, t)
i1
```

```python
i3 = C * sp.diff(v3, t)
i3
```

```python
i5 = C * sp.diff(v5, t)
i5
```

```python
i = C * sp.diff(v, t)
```

In *rms* terms, we have:

```python
T1 = sp.Rational(1,f1)
V1_rms = sp.sqrt((1/T1) * sp.integrate(v1**2, (t,0,T1)))
I1_rms = sp.sqrt((1/T1) * sp.integrate(i1**2, (t,0,T1)))
display(V1_rms, I1_rms)
```

```python
T3 = sp.Rational(1,3*f1)
V3_rms = sp.sqrt((1/T3) * sp.integrate(v3**2, (t,0,T3)))
I3_rms = sp.sqrt((1/T3) * sp.integrate(i3**2, (t,0,T3)))
display(V3_rms, I3_rms)
```

```python
T5 = sp.Rational(1,5*f1)
V5_rms = sp.sqrt((1/T5) * sp.integrate(v5**2, (t,0,T5)))
I5_rms = sp.sqrt((1/T5) * sp.integrate(i5**2, (t,0,T5)))
display(V5_rms, I5_rms)
```

```python
T1 = sp.Rational(1,f1)
V_rms = sp.sqrt((1/T1) * sp.integrate(v**2, (t,0,T1)))
I_rms = sp.sqrt((1/T1) * sp.integrate(i**2, (t,0,T1)))
display(V_rms, I_rms)
```

For calculating the capacitors power:
$$Q_1=\left( V_1∠0 \right) \left(I_1∠\frac{\pi}{2} \right)$$
$$Q_3=\left( V_3∠0 \right) \left(I_3∠\frac{\pi}{2} \right)$$
$$Q_5=\left( V_3∠0 \right) \left(I_5∠\frac{\pi}{2} \right)$$

```python
Q1 = V1_rms * I1_rms
Q3 = V3_rms * I3_rms
Q5 = V5_rms * I5_rms
QT = Q1 + Q3 + Q5
SQ  = V_rms * I_rms
display(Q1, Q3, Q5, QT, SQ)
```

```python
Q1_num = Q1.evalf(4)
Q3_num = Q3.evalf(4)
Q5_num = Q5.evalf(4)
QT_num = QT.evalf(4)
SQ_num = SQ.evalf(4)
print(Q1_num)
print(Q3_num)
print(Q5_num)
print(QT_num)
print(SQ_num)
```

Our question is: *how to size the capacitors in equivalent terms of plate data?*

From now we imagine the solution that we like to confirm with your team…

We take the ratio  $\frac{Q_T}{Q_1}$


```python
ratio = QT_num / Q1_num
ratio
```

and we size the capacitor voltage as $V_{rated} = \sqrt{ratio} \times V_1 \; \cdots$

```python
sqrt_ratio = sp.sqrt(ratio)
sqrt_ratio
```

```python
V_rated = sqrt_ratio * V1_rms
V_rated.evalf(5)
```
