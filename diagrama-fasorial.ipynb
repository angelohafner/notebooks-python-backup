```python
import matplotlib.pyplot as plt
import numpy as np

def plot_phasor(c):
    # Extract magnitude and angle
    magnitude = abs(c)
    angle_rad = np.angle(c)
    angle_deg = np.degrees(angle_rad)

    # Create figure and axis objects
    fig, ax = plt.subplots(figsize=(6,6))
    ax.quiver(0, 0, c.real, c.imag, angles='xy', scale_units='xy', scale=1, color='b')
    ax.set_xlim(-magnitude*1.2, magnitude*1.2)
    ax.set_ylim(-magnitude*1.2, magnitude*1.2)
    ax.grid(True, linestyle=':')
    ax.set_aspect('equal')

    # Annotations
    ax.set_title(f'Phasor Diagram\nMagnitude = {magnitude:.2f}, Angle = {angle_deg:.2f}°')
    ax.set_xlabel('Real Axis')
    ax.set_ylabel('Imaginary Axis')
    # ax.text(c.real, c.imag, f'  {c.real:.2f} + j{c.imag:.2f}', fontsize=10)

    return fig
```
