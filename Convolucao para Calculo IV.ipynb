```python
import matplotlib.pyplot as plt
import matplotlib.animation
plt.rcParams["animation.html"] = "jshtml"
plt.rcParams['figure.dpi'] = 150  
# plt.rcParams['font.family'] = 'Palatino Linotype'
plt.ioff()
```

```python
import numpy as np
import sympy as sp
```

```python
def gg(t_num, g_func):
    temp = np.zeros(len(t_num))
    for kk in range (len(t_num)):
        if t_num[kk]>=0:
            temp[kk] = g_func(t_num[kk])
    return temp
```

```python
s, t = sp.symbols('s t')
m = 1
b = 2
k = 5
G  = 1 / (m*s**2+b*s+k)
g = sp.inverse_laplace_transform(G, s, t)
g = g / sp.Heaviside(t)
g_func = sp.lambdify(t, g)
```

```python
t_ini = -6
t_fin = 20    
```

```python
tau_num = np.linspace(t_ini, t_fin, 500)
tempo_num = tau_num
conv_num = np.zeros(len(tau_num))
```

```python
fig = plt.figure()

def animate(kk):
    plt.cla()
    g_num = gg(tempo_num[kk]-tau_num, g_func)
    plt.plot(tau_num, g_num)
    f_num = np.heaviside(tau_num, 0.5)
    plt.plot(tau_num, f_num)
    conv_num[kk] = np.trapz(g_num * f_num, tau_num, tau_num)
    plt.fill_between(tempo_num[0:kk], conv_num[0:kk], alpha=0.5)
    plt.xlim(t_ini, t_fin)
    plt.ylim(-0.2, 1.2)
```

```python
anim = matplotlib.animation.FuncAnimation(fig, animate, frames=len(tempo_num), interval=10)
anim
```

```python
writervideo = matplotlib.animation.FFMpegWriter(fps=60)
anim.save('convolucao_mbk.mp4', writer=writervideo)
```
