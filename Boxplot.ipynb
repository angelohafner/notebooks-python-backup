```python
import matplotlib.pyplot as plt
import numpy as np

# Inicializando a semente para reprodutibilidade
np.random.seed(0)

# Gerando dados para harmônicos h1 a h50
dados_harmonicos = []
for i in range(1, 51):
    # Média aumenta com i, desvio padrão é proporcional a i
    dados = np.random.normal(loc=1/i, scale=0.02, size=5)
    dados_harmonicos.append(dados)

# Criando o boxplot
plt.figure(figsize=(12, 8))
plt.boxplot(dados_harmonicos)
plt.title('Boxplot de Harmônicos H1 a H50')
plt.xlabel('Ordem Harmônica')
plt.ylabel('Valores')
plt.xticks(np.arange(1, 51, 5), np.arange(1, 51, 5))  # Mostrando rótulos de 5 em 5 harmônicos
plt.grid(True)
plt.show()
```
