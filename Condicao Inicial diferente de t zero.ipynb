```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

```python
def ode(t,S):
    y,u = S
    return [u,
            1/(t**2-3*t) * ( t*u - (t+3*y) )]

y0 = 2
u0 = 1
S0 = (y0,u0)    
```

```python
t = np.linspace(1,2.99,10)
sol = odeint(ode,y0=S0,t=t,tfirst=True)
```

```python
plt.figure(dpi=150,figsize=(6,3))
plt.plot(t,sol[:,0],label="$y(t)$")
plt.plot(t,sol[:,1],label="$y'(t)$",ls='dashed')
plt.grid('on')
plt.legend()
plt.show()
```
