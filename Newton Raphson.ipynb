```python
import numpy as np
import matplotlib.pyplot as plt

def funcao_aula(t):
    y = (t-1)*(t-2)*(t-3)
    return y

def derivada(t):
    delta_t = 0.001
    delta_y = funcao_aula(t+delta_t) - funcao_aula(t)
    return  delta_y / delta_t
```

```python
x = np.zeros(100)
raizes = 99999*np.ones(30)
indice = 0
for x0 in range(-10, 11, 1):
    x[0] = x0

    # esse loop encontra a raiz
    for k in range(len(x)-1):
        x[k+1] = x[k] -  funcao_aula(x[k]) / derivada(x[k])

    raizes[indice] = x[-1]
    indice = indice + 1

print(raizes)
```

```python
xx = np.linspace(0, 4, 1000)
yy = funcao_aula(xx)
plt.plot(xx, yy)
plt.scatter(raizes[0:20], funcao_aula(raizes[0:20]), color='red')
plt.grid(ls='--')
```
