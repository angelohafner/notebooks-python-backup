```python
import matplotlib.pyplot as plt
import matplotlib.animation
plt.rcParams["animation.html"] = "jshtml"
plt.rcParams['figure.dpi'] = 150  
plt.ioff()
import numpy as np
```

```python
L = 5
c = 0.5
t = np.linspace(0, 5, 100)
x = np.linspace(0, L, 1000)
```

```python
for k in range(len(t)):
    p1 = 0.5 * np.sin(2*np.pi*x/L - c*t[k])
    p2 = 0.5 * np.sin(2*np.pi*x/L + c*t[k])
    u = p1 + p2
    plt.plot(x, u, label=str(np.round(t[k],1)))
```

\end{gathered} $$

```python
for k in range(len(t)):
    p1 = 2e-3 * np.sin(2*np.pi*x/L - c*t[k])
    p2 = 2e-3 * np.sin(2*np.pi*x/L + c*t[k])
    G1 = 3e-3 * L/(4*np.pi)*np.cos(4*np.pi*x/L + c*t[k])
    G2 = 3e-3 * L/(4*np.pi)*np.cos(4*np.pi*x/L - c*t[k])
    u = 1/2*(p1 + p2) + 1/(2*0.5)*(G1 - G2)
    plt.plot(x, u, label=str(np.round(t[k],1)))

```

```python
fig = plt.figure()
def animate(k):
    plt.cla()
    p1 = 2e-3 * np.sin(2*np.pi*x/L - c*t[k])
    p2 = 2e-3 * np.sin(2*np.pi*x/L + c*t[k])
    G1 = 3e-3 * L/(4*np.pi)*np.cos(4*np.pi*x/L + c*t[k])
    G2 = 3e-3 * L/(4*np.pi)*np.cos(4*np.pi*x/L - c*t[k])
    u = 1/2*(p1 + p2) + 1/(2*0.5)*(G1 - G2)
    plt.plot(x, u, label='sin')
    p1 = 2e-3 * np.sin(2*np.pi*x/L - c*t[k])
    p2 = 2e-3 * np.sin(2*np.pi*x/L + c*t[k])
    u = 1/2*(p1 + p2)
    plt.plot(x, u, label='0')

# matplotlib.animation.FuncAnimation(fig, animate, frames=len(t), interval=250)
```

```python
matplotlib.animation.FuncAnimation(fig, animate, frames=len(t), interval=250)
```
