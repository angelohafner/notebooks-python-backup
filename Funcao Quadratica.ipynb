```python
#!pip install tensorflow
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import activations
```

```python
xxx = np.linspace(-10., 10., 20)
yyy = xxx**2
```

```python
plt.scatter(xxx, yyy)
```

```python
# l0 é a camada (layer) que recebe as entradas
l0 = tf.keras.layers.Dense(units=1, input_shape=[1])
# l1 é a camada (layer) que recebe as saídas de l0
l1 = tf.keras.layers.Dense(units=100)
l2 = tf.keras.layers.Dense(units=100)
l3 = tf.keras.layers.Dense(units=100)
l4 = tf.keras.layers.Dense(units=100)
l9 = tf.keras.layers.Dense(units=1)
```

```python
# Monta as camadas no modelo
model = tf.keras.Sequential([l0, l1, l2, l3, l4, l9])
```

```python
# Compila o modelo. Dica: Adam(learning rate) between 0.1 and 0.001
model.compile(loss='mean_squared_error',
              optimizer=tf.keras.optimizers.Adam(0.01)
              ) 
```

```python
history = model.fit(xxx, yyy, epochs=1000, verbose=False) #Train the model
print("Finished training the model")
```

```python
plt.xlabel('Época #')
plt.ylabel("Erro")
plt.plot(history.history['loss'])
plt.grid()
```

```python
xx = np.linspace(-50, 50, 200)
plt.plot(xx,model.predict([xx]))
plt.scatter(xxx, yyy)
```
