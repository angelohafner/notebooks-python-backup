```python
import pandas as pd
import plotly.graph_objects as go
from google.colab import files

def plot_oscilloscope_csv_upload():
    """
    Requests CSV upload (Rigol format) and plots voltage vs time using Plotly.
    """
    # Request file upload
    uploaded = files.upload()
    file_path = list(uploaded.keys())[0]

    # Read CSV
    df = pd.read_csv(file_path)

    # Extract start time and increment from the first row
    start_time = float(df.loc[0, 'Start'])
    time_increment = float(df.loc[0, 'Increment'])

    # Get numeric data starting from the second row
    df_clean = df.iloc[1:].copy()
    df_clean = df_clean[['X', 'CH1']].astype(float)

    # Calculate time axis
    time_axis = start_time + df_clean['X'] * time_increment

    # Create interactive plot
    fig = go.Figure()
    fig.add_trace(go.Scatter(
        x=time_axis,
        y=df_clean['CH1'],
        mode='lines',
        name='CH1 Voltage'
    ))

    # Layout settings
    fig.update_layout(
        title='Oscilloscope Data - CH1',
        xaxis_title='Time (s)',
        yaxis_title='Voltage (V)',
        template='plotly_white'
    )

    fig.show()

# Run in Colab:
plot_oscilloscope_csv_upload()
```
