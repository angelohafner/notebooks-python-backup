```python
import numpy as np
f1 = 60
w1 = 2*np.pi*f1


```

```python
def impedancia_filtro(w, R, C, L, r=1e-9):
    w[w == 0] = 1e-9
    impedancia_indutor = r+1j*w*L
    admitancia_ramo_RL = 1/R + 1/impedancia_indutor
    impedancia_ramo_RL = 1 / admitancia_ramo_RL
    admitancia_capacitor = 1j*w*C
    impedancia_capacitor = 1 / admitancia_capacitor
    impedancia_filtro = impedancia_capacitor + impedancia_ramo_RL
    return [impedancia_indutor, impedancia_ramo_RL, impedancia_capacitor, impedancia_filtro]
```

```python
kv_trifasico = 34.5
v_trifasico = 1e3 * kv_trifasico
v_monofasico = v_trifasico / np.sqrt(3)
```

```python
mH_filtro = 34.303
uF_filtro = 8.543

H_filtro = 1e-3 * mH_filtro
F_filtro = 1e-6 * uF_filtro
ohms_paralelo = 100

rad_sec_sintonia = 1/(np.sqrt(H_filtro*F_filtro))
Hz_sintonia = rad_sec_sintonia / 2*np.pi
h_sintonia = rad_sec_sintonia / w1

omegas = np.arange(0, 12, 1)*w1
impedancias_freq = impedancia_filtro(omegas, R=ohms_paralelo, C=F_filtro, L=H_filtro)
impedancia_indutor, impedancia_ramo_RL, impedancia_capacitor, impedancia_filtro = impedancias_freq

corrente_fund_filtro = v_monofasico / impedancia_filtro[1]
tensao_fund_banco = impedancia_capacitor[1] * corrente_fund_filtro
potencia_fund_banco_trifasica  = 3 * tensao_fund_banco * np.conj(corrente_fund_filtro)
potencia_fund_filtro_trifasica = 3 * v_monofasico      * np.conj(corrente_fund_filtro)
tensao_ramo_RL_fund = corrente_fund_filtro * impedancia_ramo_RL[1]
corrente_indutor_fund = tensao_ramo_RL_fund / impedancia_indutor[1]
corrente_resistor_fund = tensao_ramo_RL_fund / ohms_paralelo
potencia_fund_indutor_trifasica = 3 * tensao_ramo_RL_fund * np.conj(corrente_indutor_fund)
potencia_fund_resistor_trifasica = 3 * tensao_ramo_RL_fund * np.conj(corrente_resistor_fund)

potencia_fund_banco_trifasica
```
