```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib

# Configuração de fonte para imitar a fonte do LaTeX
matplotlib.rcParams['font.size'] = 22

# Define o intervalo de x de 0 a 2*pi com 1000 pontos
x = np.linspace(0, 1, 1000)
sin_x = np.sin(2*np.pi*x)
cos_x = np.cos(3*2*np.pi*x)
sin_cos_x = sin_x * cos_x

# Cria a figura
plt.figure(figsize=(16, 7), dpi=300)

# Plota sin(x)
plt.plot(x, sin_x, label=r'${\sin \left( {2\pi x} \right)}$', color='blue')

# Plota cos(x)
plt.plot(x, cos_x, label=r'${\sin \left( {3 \times 2\pi x} \right)}$', color='red')

# Preenche a área sob sin(x)*cos(x)
plt.fill_between(x, sin_cos_x, color='purple', alpha=0.3, label=r'$\int {\sin \left( {2\pi x} \right)\sin \left( {3 \times 2\pi x} \right)\,dx}$')
plt.plot(x, sin_cos_x, color='purple', linestyle='--')  # Linha para melhor visualização

# Adiciona títulos e rótulos
plt.xlabel('$x$')
plt.ylabel('$y$')
plt.grid(True)
plt.legend()

# Mostra o gráfico
plt.show()
```
