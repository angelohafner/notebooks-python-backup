```python
import numpy  as np
import matplotlib.pyplot as plt
```

```python
V0 = 100
n = 200
```

```python
b = 5
x = np.linspace(-b,b,1000)
a = 7
y = np.linspace(0,a,1000)

X,Y = np.meshgrid(x,y)
```

$$V\left( {x,y} \right) = {{4{V_0}} \over \pi }\sum\limits_{n = 1,3,7}^{ + \infty } {{1 \over n}} {{\cosh \left( {{{n\pi } \over a}x} \right)} \over {\cosh \left( {{{n\pi } \over a}b} \right)}}\sin \left( {{{n\pi } \over a}y} \right)$$

```python
fora = 4*V0/np.pi
VV = 0
for ii in range(1,n,2):
    num = np.cosh(ii*np.pi/a*X) * np.sin(ii*np.pi/a*Y)
    den = ii * np.cosh(ii*np.pi/a*b)
    VV = VV + num/den

VV = fora * VV 

```

```python
plt.figure(dpi=300)
CS = plt.contourf(X,Y,VV, cmap='jet', levels=np.linspace(0, 100, 21))
plt.clabel(CS, inline=3, fontsize=6)
plt.xlim(-b*1.1,b*1.1)
plt.ylim(-0.1*a,a*1.1)
plt.colorbar()
plt.show()
```
