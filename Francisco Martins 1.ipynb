```python
# Função para ajustar o modelo ARIMA e fazer previsões

def fit_arima_and_forecast(data):

    forecast_results = {}

    for name, values in data.items():

        # Ajustar o modelo ARIMA

        model = ARIMA(values, order=(1, 0, 0))  # Ordem (p, d, q) = (1, 0, 0)

        model_fit = model.fit()

        # Fazer previsões para os próximos 6 números

        forecast = model_fit.forecast(steps=6)

        forecast_results[name] = forecast

    return forecast_results
```
