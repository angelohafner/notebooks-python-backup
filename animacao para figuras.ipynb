```python
from PIL import Image
import os
import zipfile
from google.colab import files
import shutil



# Fazer upload do arquivo GIF
uploaded = files.upload()

# Obter o nome do arquivo GIF carregado
gif_path = list(uploaded.keys())[0]

# Definir a pasta de saída e o nome do arquivo ZIP
output_folder = '/content/frames/'  # Pasta para salvar os frames
zip_filename = '/content/saida.zip'  # Nome do arquivo ZIP final

# Criar a pasta de saída se não existir
os.makedirs(output_folder, exist_ok=True)

# Abrir o GIF
with Image.open(gif_path) as gif:
    # Contador para os frames
    frame_number = 0
    while True:
        # Salvar cada frame como um arquivo de imagem
        frame_path = os.path.join(output_folder, f'frame-{frame_number:02d}.png')
        gif.save(frame_path)

        try:
            # Tentar mover para o próximo frame
            gif.seek(gif.tell() + 1)
        except EOFError:
            # Se atingir o fim do GIF, sair do loop
            break

        frame_number += 1

print(f'Frames salvos na pasta: {output_folder}')

# Criar um arquivo ZIP com todos os frames
with zipfile.ZipFile(zip_filename, 'w') as zipf:
    for root, dirs, files in os.walk(output_folder):
        for file in files:
            zipf.write(os.path.join(root, file), os.path.relpath(os.path.join(root, file), output_folder))

print(f'Arquivo ZIP criado: {zip_filename}')

# Apagar a pasta e todo o seu conteúdo
shutil.rmtree(output_folder)
```
