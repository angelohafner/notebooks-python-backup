# Gráfico de uma função racional em python

$$f\left( x \right) = \frac{{{x^2} - 4x + 4}}{{{x^2} + x - 6}}$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xLim = 5
x = np.linspace(-xLim,+xLim,100)
num = x**2 - 4*x + 4
den = x**2 + x -6
y = num / den
```

```python
plt.plot(x,y)
plt.grid()
plt.xlabel("eixo $x$")
plt.ylabel("eixo $y$")
plt.title("Gráfico de $f(x)$")
plt.ylim(-100,100)
plt.show()
```
