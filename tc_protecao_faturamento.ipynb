```python
# Comments in English only
import numpy as np
import plotly.graph_objects as go

def lambda_tanh(i_m: np.ndarray, L0_h: float, Ik_a: float) -> np.ndarray:
    # Flux linkage model using tanh saturation
    return (L0_h * Ik_a) * np.tanh(i_m / Ik_a)

# Example "typical" parameters (illustrative)
# Protection CT: higher knee current (later saturation)
L0_prot_h = 20.0
Ik_prot_a = 0.50

# Metering CT: lower knee current (earlier saturation)
L0_met_h = 10.0
Ik_met_a = 0.10

# Current sweep
i_m = np.linspace(-2.0, 2.0, 2001)

lam_prot = lambda_tanh(i_m, L0_prot_h, Ik_prot_a)
lam_met = lambda_tanh(i_m, L0_met_h, Ik_met_a)

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=i_m,
    y=lam_prot,
    mode="lines",
    name="CT protection (tanh)"
))

fig.add_trace(go.Scatter(
    x=i_m,
    y=lam_met,
    mode="lines",
    name="CT metering (tanh)"
))

fig.update_layout(
    title="Typical CT magnetization curve using tanh model",
    xaxis_title="Magnetizing current i_m (A)",
    yaxis_title="Flux linkage lambda (Wb-turn)",
    legend_title="Curve",
)

fig.show()
```
