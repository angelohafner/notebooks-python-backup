```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
def f(t):
    #return np.exp(-1*t) * np.heaviside(t,0)
    return np.heaviside(t+1,0) - np.heaviside(t-1,0)
```

```python
def g(t):
    #return np.exp(-2*t) * np.heaviside(t,0)
    return np.heaviside(t+1,0)
```

```python
dt = 0.05
tt = np.arange(-2,10,dt)
nn = len(tt);
```

$$\int\limits_0^t {f\left( {\tau  - t} \right)g\left( \tau  \right)d\tau } $$

```python
tau = tt
convol = np.zeros(nn)
```

```python
gt = g(tt)
ut = f(tt)

full_convolution = np.convolve(gt, ut)
convol = full_convolution[:nn] * dt

plt.plot(tt, convol)
plt.show()
```
