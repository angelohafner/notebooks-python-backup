$$\phi  = \frac{{{n_1}{i_1} - {n_2}{i_2}}}{\mathcal{R}} = \frac{{{n_1}{i_1} - {n_2}\frac{{ϵ{_2}}}{r}}}{\mathcal{R}}$$


$${ϵ_2} =  - {n_2}\dot \phi $$


$$\frac{{{ϵ_2}}}{{{n_2}}} = \frac{{{n_1}{{\dot i}_1} - {n_2}{{\dot i}_2}}}{\mathcal{R}}$$


$$\mathcal{R}r{i_2} = {n_1}{n_2}{{\dot i}_1} - {n_2}{n_2}{{\dot i}_2}$$


$${{\dot i}_2} =  - \frac{{r\mathcal{R}}}{{n_2^2}}{i_2} + \frac{{{n_1}}}{{{n_2}}}{{\dot i}_1}$$

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Definindo a EDO de primeira ordem: dy/dt = f(t, y) + termo_fonte
# Exemplo: dy/dt = -2y + cos(t), com condição inicial y(0) = 1
def edo(t, y):
    mu = 5000 * 4 * np.pi * 1e-7
    R = 0.01
    r = 10
    n2 = 200
    n1 = 20
    a = r*R/(n2**2)
    b = n1/n2 * np.cos(t)
    return -a*y + b

# Intervalo de tempo para a solução: de 0 a 10
t_span = (0, 6)
# Condições iniciais
y0 = [0]

# Resolvendo a EDO
sol = solve_ivp(edo, t_span, y0, t_eval=np.linspace(t_span[0], t_span[1], 100000))

# Plotando a solução
plt.plot(sol.t, sol.y[0], label='y(t)')
plt.xlabel('t')
plt.ylabel('y(t)')
plt.legend()
plt.show()
```
