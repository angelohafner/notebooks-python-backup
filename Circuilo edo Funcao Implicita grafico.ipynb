```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
xrange = np.linspace(-10, 10, 1000)
yrange = np.linspace(-10, 10, 1000)

xrange2 = np.linspace(-10, 10, 11)
yrange2 = np.linspace(-10, 10, 11)
```

```python
x, y = np.meshgrid(xrange, yrange)
x2, y2 = np.meshgrid(xrange2, yrange2)
```

```python
z2 = -x2 / y2
ang = np.arctan(z2)
uu = np.cos(ang)
vv = np.sin(ang)
```

```python
z = x**2 + y**2
plt.figure(figsize=(4,4), dpi=300)
cs = plt.contour(x, y, z, [3, 9, 25, 50 ])
plt.clabel(cs, cs.levels, inline=True, fmt='%3.0f')
plt.quiver(x2, y2, uu, vv, angles='xy')
```
