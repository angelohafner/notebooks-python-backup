```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
xrange = np.linspace(-3, 3, 1000)
yrange = np.linspace(-3, 3, 1000)

xrange2 = np.linspace(-3, 3, 21)
yrange2 = np.linspace(-3, 3, 21)
```

```python
x, y = np.meshgrid(xrange, yrange)
x2, y2 = np.meshgrid(xrange2, yrange2)
```

```python
z2 = (y2 - x2**2) / (y2**2 - x2)
ang = np.arctan(z2)
uu = np.cos(ang)
vv = np.sin(ang)
```

$${x^3} + {y^3} - 3xy = c$$

$$y' = \frac{{y - {x^2}}}{{{y^2} - x}}$$

```python
z = x**3 + y**3 - 3*x*y
plt.figure(figsize=(4,4), dpi=300)
cs = plt.contour(x, y, z, [-10, -5, 0, 5, 10])
plt.clabel(cs, cs.levels, inline=True, fmt='%3.0f')
plt.quiver(x2, y2, uu, vv, angles='xy')
```
