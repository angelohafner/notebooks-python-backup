```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-2, 2, 11)
y = x
X, Y = np.meshgrid(x, y)
s = plt.scatter(X, Y)
```

```python
v = s.get_offsets().data.T
v
plt.scatter(v[0,:], v[1,:])
```

```python
T = np.array([[2,  0],
              [0,  3],
              [1, -2]])
T
```

```python
w = T @ v
w
```

```python
fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(projection='3d')
ax.scatter(v[0,:], v[1,:], 0*v[0,:]) 
ax.scatter(w[0,:], w[1,:], w[2,:]) 
ax.view_init(10,20)
```
