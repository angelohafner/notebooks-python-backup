```python
import matplotlib.animation as animation
import matplotlib.pyplot as plt
import numpy as np

# Configurações do Matplotlib
plt.rcParams["animation.html"] = "jshtml"  # Saída da animação em HTML e JavaScript
plt.rcParams['figure.dpi'] = 150  # Aumenta a resolução da figura

# Configurações básicas da onda
amplitude = 1
frequencia = 1
velocidade = 1
comprimento_onda = 1
k = 2 * np.pi / comprimento_onda
omega = 2 * np.pi * frequencia

# Configuração do espaço e tempo
x = np.linspace(0, 4 * np.pi, 1000)
tempo = np.linspace(0, 10, 200)

# Configuração da figura
fig, ax = plt.subplots()
linha_e, = ax.plot([], [], lw=2, label='Campo Elétrico (E)')
linha_b, = ax.plot([], [], lw=2, label='Campo Magnético (B)')
ax.set_xlim(0, 4 * np.pi)
ax.set_ylim(-amplitude, amplitude)
ax.set_xlabel('Espaço')
ax.set_ylabel('Amplitude')
ax.set_title('Animação da Onda TEM')
ax.legend()

# Função de inicialização
def init():
    linha_e.set_data([], [])
    linha_b.set_data([], [])
    return linha_e, linha_b

# Função de atualização
def update(t):
    y_e = amplitude * np.sin(k * x - omega * t)
    y_b = amplitude * np.sin(k * x - omega * t + np.pi/2)  # B está em quadratura com E
    linha_e.set_data(x, y_e)
    linha_b.set_data(x, y_b)
    return linha_e, linha_b

# Criação da animação
ani = animation.FuncAnimation(fig, update, frames=tempo, init_func=init, blit=True, interval=50, repeat=True)

ani
```
