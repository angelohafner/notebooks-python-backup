```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

$$v\left( t \right) = {V_p}\sin \left( {\omega t} \right)$$

```python
w = 1
Vp = 1
t_num = np.linspace(0, 2*np.pi)
v_num = Vp*np.sin(w*t_num)
plt.fill_between(t_num, v_num, alpha=0.25)
plt.plot(t_num, v_num)
```

```python
int_sin_num = np.trapz(v_num, t_num)
int_sin_num
```

$v\left( t \right) = {V_p}\sin \left( {\omega t} \right)$

Simbólico

```python
t = sp.symbols("t")
v = Vp*sp.sin(w*t)
v
```

```python
int_simbolica = sp.integrate(v, (t, 0, 2*sp.pi))
int_simbolica
```

```python
int_simbolica = sp.integrate(v**2, (t, 0, 2*sp.pi))
int_simbolica
```

```python
int_sin_num = np.trapz(v_num**2, t_num)
int_sin_num
```

```python
plt.fill_between(t_num, v_num**2, alpha=0.25)
plt.plot(t_num, v_num**2)
```

```python
fig2, ax2 = plt.subplots()
yyy = np.sin(w*t_num)*np.cos(w*t_num)
ax2.fill_between(t_num, yyy, alpha=0.25)
ax2.plot(t_num, yyy)
```

```python
int_sinXcos_num = np.trapz(yyy, t_num)
int_sinXcos_num
```

```python
vvv = sp.sin(w*t)*sp.cos(w*t)
int_sinXcos_simb = sp.integrate(vvv, (t, 0, 2*sp.pi))
int_sinXcos_simb
```

```python
vvv = sp.sin(w*t)*sp.sin(2*w*t)
int_sinXsin_simb = sp.integrate(vvv, (t, 0, 2*sp.pi))
int_sinXsin_simb
```

```python
fig3, ax3 = plt.subplots(dpi=150)
yyy = np.sin(w*t_num)*np.sin(2*w*t_num)
ax3.fill_between(t_num, yyy, alpha=0.25)
ax3.plot(t_num, yyy, label=r'$\sin(\omega t) \; \sin(2\omega t)$')
ax3.plot(t_num, np.sin(w*t_num), label=r'$\sin(\omega t)$')
ax3.plot(t_num, np.sin(2*w*t_num), label=r'$\sin(2\omega t)$')
ax3.legend()
```
