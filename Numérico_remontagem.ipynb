```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
tt = np.linspace(0, 10/60, 60*128*10)
kk = np.arange(5)

w  = kk * 2 * np.pi * 60
v1 = 100*np.sin(w[1]*tt) + 50*np.sin(w[2]*tt) + 20*np.sin(w[3]*tt) + 10*np.sin(w[4]*tt)
v2 = 20 + 120*np.cos(w[1]*tt) + 52*np.cos(w[2]*tt) + 22*np.cos(w[3]*tt) + 12*np.cos(w[4]*tt)
vv  = v1 + v2
```

```python
fig, ax = plt.subplots(figsize = (18,5))
ax.plot(tt, vv)
```

$${b_k} = \frac{{\int\limits_0^T {v\left( t \right)\sin \left( {k\omega t} \right)dt} }}{{\int\limits_0^T {{{\sin }^2}\left( {k\omega t} \right)dt} }}$$

$${a_k} = \frac{{\int\limits_0^T {v\left( t \right)\cos \left( {k\omega t} \right)dt} }}{{\int\limits_0^T {{{\cos }^2}\left( {k\omega t} \right)dt} }}$$




```python
bb = np.zeros(5)
aa = np.zeros(5)
for kk in range(5):
    num = np.trapz(vv*np.sin(w[kk]*tt), tt)
    den = np.trapz((np.sin(w[kk]*tt))**2, tt)
    if den == 0 and num == 0:
        bb[kk] = 0
    else:
        bb[kk] = num/den
    num = np.trapz(vv*np.cos(w[kk]*tt), tt)
    den = np.trapz((np.cos(w[kk]*tt))**2, tt)
    aa[kk] = num/den
```

```python
print("-----------------------")
print("Ondas Coseno")
print("-----------------------")
for kk in range(5):
    print(np.round(aa[kk], 2))
print("-----------------------")
```

```python
print("-----------------------")
print("Ondas Seno")
print("-----------------------")
for kk in range(5):
    print(np.round(bb[kk], 2))
print("-----------------------")
```
