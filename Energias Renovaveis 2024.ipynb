```python
import matplotlib.pyplot as plt

# Dados de capacidade instalada (em GW)
anos = list(range(2012, 2023))
hidreletrica = [84.7, 87.1, 90.1, 93.1, 96.6, 98.6, 100.2, 101.8, 103.5, 105.0, 106.4]
eolica = [2.5, 3.5, 5.4, 8.7, 10.7, 12.8, 14.7, 15.9, 17.2, 19.2, 21.5]
solar = [0.01, 0.02, 0.03, 0.1, 0.2, 0.9, 1.5, 2.4, 4.0, 7.5, 13.0]
termica = [25.0, 27.5, 30.1, 33.2, 35.5, 37.0, 39.0, 40.0, 41.5, 43.0, 45.0]
petroleo = [9.5, 10.0, 10.4, 10.8, 11.2, 11.5, 11.8, 12.1, 12.4, 12.7, 13.0]

# Criando o gráfico
plt.figure(figsize=(12, 8))
plt.plot(anos, hidreletrica, marker='o', linestyle='-', color='b', label='Hidrelétrica')
plt.plot(anos, eolica, marker='o', linestyle='-', color='g', label='Eólica')
plt.plot(anos, solar, marker='o', linestyle='-', color='r', label='Solar')
plt.plot(anos, termica, marker='o', linestyle='-', color='y', label='Térmica')
plt.plot(anos, petroleo, marker='o', linestyle='-', color='k', label='Petróleo')

# Adicionando título e legendas
plt.title('Capacidade Instalada por Fonte de Energia no Brasil (2012-2022)')
plt.xlabel('Ano')
plt.ylabel('Capacidade Instalada (GW)')
plt.legend()
plt.grid(True)
plt.xticks(anos)
plt.yticks(range(0, 110, 10))
plt.tight_layout()

# Exibindo o gráfico
plt.show()
```
