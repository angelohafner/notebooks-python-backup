# Current Transformer Ratio Selection for 7.2 MVAR Capacitor Bank

## Intial calculations

The current transformers come in multi ratios, and the initial step involves choosing the appropriate ratio for the system. The CT ratio must align with the rated current of the capacitor bank.

```python
import numpy as np
```

```python
Q_cap = 7.2e6
V_cap = 11e3
I_cap_phase = Q_cap/(np.sqrt(3)*V_cap)
arredondado = int(100*np.round(I_cap_phase/100, 0))
Phase_CT_Ratio = int(arredondado / 5)
Neutral_CT_Ratio = int(2.5e-2 * Phase_CT_Ratio)
print(f"Therefore, we select here a CT ratio of {int(arredondado)}/5.")
print(f"Phase CT Ratio = {arredondado}/5 = {Phase_CT_Ratio}.")
print(f"Neutral CT Ratio = {int(2.5e-2* arredondado)}/5 = {Neutral_CT_Ratio}.")
```

## Pickup Setting for Phase Element

The capacitor bank is designed to have 20 percent overload, hence we will select the pickup current for phase element at 120 percent.

```python
I_cap_phase_1p2 = 1.2 * I_cap_phase
pickup_current_sec = I_cap_phase_1p2 / Phase_CT_Ratio
print(f"Primary current at 20 percent over load = {I_cap_phase_1p2}")
print(f"Relay pickup current is set at the secondary current = {pickup_current_sec}")
```

## Fault Current

The next step is to calculate the maximum fault current. To do this, you need to obtain the value of the transformer percentage impedance from the nameplate data. While calculating the maximum fault current, the percentage impedance of the transformer is considered at the lowest tap, which is done to calculate the worst-case scenario.

```python
S_transformer = 26e6
Z_perc_transformer = 0.1995
V_transformer = 11e3
S_fault = S_transformer / Z_perc_transformer
I_fault = S_fault / (np.sqrt(3) * V_transformer)
I_fault_sec_TC = I_fault / Phase_CT_Ratio
print(f"Corresponding fault secondary current on CT = {I_fault_sec_TC}")
```

Operating time/Tripping Time chosen =	0.4 sec

```python
Multiple_of_PickUp = I_fault_sec_TC / pickup_current_sec
print(f"Multiple of Pick up = {Multiple_of_PickUp}")
```
