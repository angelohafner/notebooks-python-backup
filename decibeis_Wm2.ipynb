```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
I0 = 1e-12  # Intensidade de referência em W/m²

# Eixo x: Nível em decibéis
dB_values = np.linspace(60, 120, 500)  # Valores de 60 a 120 dB

# Eixo y: Intensidade sonora em W/m²
intensity_values = I0 * 10**(dB_values / 10)

# Adicionando os pontos de destaque
highlight_points = {
    60: "Nível confortável",
    85: "Danos longa exposição",
    100: "Nível perigoso",
    120: "Dor imediata"
}

highlight_x = np.array(list(highlight_points.keys()))
highlight_y = I0 * 10**(highlight_x / 10)
highlight_labels = list(highlight_points.values())

# Plotando o gráfico
plt.figure(figsize=(8, 4))  # Aumentando a largura da figura
plt.plot(dB_values, intensity_values, label=r"$L = 10 \cdot \log_{10}\left(\frac{I}{I_0}\right)$", color='blue')

# Adicionando os pontos de scatter
plt.scatter(highlight_x, highlight_y, color='red', zorder=5, label="Pontos destacados")
for x, y, label in zip(highlight_x, highlight_y, highlight_labels):
    plt.text(x, y, label, fontsize=9, ha='left', va='bottom')

# Configurações do gráfico
plt.xscale("linear")
plt.yscale("log")
plt.title("Relação entre Potência Sonora e Nível em Decibéis")
plt.xlabel("Nível em dB")
plt.ylabel("Intensidade Sonora ($\mathrm{W/m^2}$)")
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Eixo x: Níveis de decibéis em pequena faixa
dB_values = np.linspace(0, 10, 1000)  # Valores de 0 a 10 dB

# Constantes
I0 = 1e-12  # Intensidade de referência em W/m²

# Calculando as potências sonoras em W/m²
intensity_values = I0 * 10**(dB_values / 10)

# Calculando o eixo y: Razão das potências sonoras
reference_intensity = intensity_values[0]  # Primeiro valor como referência
ratio_values = intensity_values / reference_intensity

# Plotando o gráfico
plt.figure(figsize=(8, 4))
plt.plot(dB_values, ratio_values, color='blue', label="Razão da potência sonora")

# Configurações do gráfico
plt.title("Multiplicação da Potência Sonora com o Aumento dos dB")
plt.xlabel("Nível em dB")
plt.ylabel("Razão da Potência Sonora ($I / I_0$)")

# Configurando linhas de grade personalizadas
plt.xticks(np.arange(0, 12, 3))  # Linhas de grade no eixo x de 3 em 3
plt.yticks(np.arange(0, np.ceil(ratio_values.max()) + 2, 2))  # Linhas de grade no eixo y de 2 em 2
plt.grid(True, linestyle="--", linewidth=0.5)

plt.tight_layout()
plt.show()
```
