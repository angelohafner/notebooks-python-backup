```python
import numpy as np
from scipy.integrate import solve_ivp
import plotly.graph_objects as go
width_plot= 3000

# Circuit parameters
R = 10
L = 0.1
C = 100e-6
V0 = 10
f = 60
omega = 2 * np.pi * f

# ODE system
def rlc_system(t, y):
    q = y[0]
    dq_dt = y[1]
    d2q_dt2 = (V0*np.sin(omega*t) - R*dq_dt - q / C) / L
    return [dq_dt, d2q_dt2]

# Solve
y0 = [0, 0]
t_span = (0, 0.2)
t_eval = np.linspace(*t_span, 10000)
sol = solve_ivp(rlc_system, t_span, y0, t_eval=t_eval)

# Results
q_t = sol.y[0]
i_t = sol.y[1]
t = sol.t
v_source = V0 * np.sin(omega * t)
v_R = R * i_t
v_L = L * np.gradient(i_t, t_eval, edge_order=2)
v_C = q_t / C

# Create figure with secondary y-axis
from plotly.subplots import make_subplots

fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add voltages to primary y-axis
fig.add_trace(go.Scatter(x=t, y=v_source, name="v_s(t)"), secondary_y=False)
fig.add_trace(go.Scatter(x=t, y=v_R, name="v_R(t)"), secondary_y=False)
fig.add_trace(go.Scatter(x=t, y=v_L, name="v_L(t)"), secondary_y=False)
fig.add_trace(go.Scatter(x=t, y=v_C, name="v_C(t)"), secondary_y=False)

# Add current to secondary y-axis
fig.add_trace(go.Scatter(x=t, y=i_t, name="i(t)", line=dict(dash='dash', color='black')), secondary_y=True)

# Layout
fig.update_layout(
    title_text="Voltages and Current in RLC Series Circuit",
    width=1000,
    height=500
)

fig.update_xaxes(title_text="Time [s]")
fig.update_yaxes(title_text="Voltage [V]", secondary_y=False)
fig.update_yaxes(title_text="Current [A]", secondary_y=True)
fig.update_layout(
    title_text="Voltages and Current in RLC Series Circuit",
    width=width_plot,
    height=500
)

fig.show()
```

```python
v_C = q_t / C
p_C = v_C * i_t  # Instantaneous power in the capacitor

# Create figure
fig = go.Figure()

# v_C(t) — eixo y à esquerda (yaxis)
fig.add_trace(go.Scatter(
    x=t,
    y=v_C,
    name="v_C(t) [V]",
    line=dict(color="blue"),
    yaxis="y"
))

# i_C(t) — eixo y à direita (yaxis2)
fig.add_trace(go.Scatter(
    x=t,
    y=i_t,
    name="i_C(t) [A]",
    line=dict(color="black", dash="dash"),
    yaxis="y2"
))

# p_C(t) — eixo y à direita, deslocado (yaxis3)
fig.add_trace(go.Scatter(
    x=t,
    y=p_C,
    name="p_C(t) [W]",
    fill="tozeroy",
    mode="lines",
    line=dict(color="red"),
    fillcolor="rgba(255,0,0,0.3)",
    yaxis="y3"
))

# Layout
fig.update_layout(
    title="Capacitor: Voltage, Current, and Instantaneous Power",
    width=width_plot,
    height=500,
    xaxis=dict(title="Time [s]"),

    yaxis=dict(  # y
        title="Voltage [V]",
        titlefont=dict(color="blue"),
        tickfont=dict(color="blue"),
        position=0.0
    ),

    yaxis2=dict(  # y2
        title="Current [A]",
        titlefont=dict(color="black"),
        tickfont=dict(color="black"),
        overlaying="y",
        side="right",
        position=0.93
    ),

    yaxis3=dict(  # y3
        title="Power [W]",
        titlefont=dict(color="red"),
        tickfont=dict(color="red"),
        overlaying="y",
        side="right",
        anchor="free",
        position=1.0  # importante: define a posição deslocada
    ),

    legend=dict(x=0.01, y=0.99)
)
```

```python
import plotly.graph_objects as go
import numpy as np

# Potências instantâneas
v_R = R * i_t
v_C = q_t / C
v_L = L * np.gradient(i_t, t, edge_order=2)
v_s = V0 * np.sin(omega * t)

p_R = v_R * i_t               # ou R * i_t**2
p_C = v_C * i_t
p_L = v_L * i_t
p_s = v_s * i_t

# Figura única
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=t, y=p_R,
    mode="lines",
    name="Resistor",
    line=dict(color="red"),
    fill="tozeroy",
    fillcolor="rgba(255,0,0,0.2)"
))

fig.add_trace(go.Scatter(
    x=t, y=p_L,
    mode="lines",
    name="Indutor",
    line=dict(color="blue"),
    fill="tozeroy",
    fillcolor="rgba(0,0,255,0.2)"
))

fig.add_trace(go.Scatter(
    x=t, y=p_C,
    mode="lines",
    name="Capacitor",
    line=dict(color="green"),
    fill="tozeroy",
    fillcolor="rgba(0,255,0,0.2)"
))

fig.add_trace(go.Scatter(
    x=t, y=p_s,
    mode="lines",
    name="Fonte",
    line=dict(color="black", dash="dot"),
    fill="tozeroy",
    fillcolor="rgba(0,0,0,0.1)"
))

# Layout
fig.update_layout(
    title="Potências Instantâneas: Fonte, Resistor, Indutor, Capacitor",
    xaxis_title="Tempo [s]",
    yaxis_title="Potência Instantânea [W]",
    width=width_plot,
    height=500,
    legend=dict(x=0.01, y=0.99),
    yaxis=dict(zeroline=True)
)

fig.show()
```
