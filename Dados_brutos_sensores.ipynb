```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
```

```python
t = np.linspace(0, 20, 10000)
w = 2 *np.pi*0.5
A = 20
sinal = np.zeros(len(t))
```

```python
for ii in range (1, 23, 1):
    sinal = sinal + A/ii * np.sin(ii*(w*t-np.pi/6))
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=t, y=sinal,
                    mode='lines',
                    name='sinal'))
fig.add_trace(go.Scatter(x=t, y=sinal,
                    mode='markers',
                    name='sinal'))
fig.show()
```
