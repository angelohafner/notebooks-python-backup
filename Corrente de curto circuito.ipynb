```python
import math
import numpy as np

def CalcAB(iPico, tPico, tMetade, amp):
    B = math.log(1.0 - iPico / amp) / tPico
    ebtm = math.exp(B * tMetade)
    A = math.log(iPico / (2 * amp) + ebtm) / tMetade
    tp = math.log(B / A) / (A - B)
    return A, B, tp

def main():
    tPico = 1.2e-6
    tMetade = 50e-6
    iPico = 10000

    # Função I = Amp*(exp(A*t) - exp(B*t))
    i = lambda amp, a, b, t: amp * (math.exp(a * t) - math.exp(b * t))
    i1 = lambda amp, a, t: amp * (math.exp(a * t))
    i2 = lambda amp, b, t: amp * (-math.exp(b * t))

    amp0 = iPico * 1.02

    [A, B, tp] = CalcAB(iPico, tPico, tMetade, amp0)

    tPicoAjust = tPico

    while abs(tp - tPico) / tPico > 1e-4:
        tPicoAjust *= tPico / tp
        [A, B, tp] = CalcAB(iPico, tPicoAjust, tMetade, amp0)

    print(f'Amp = {amp0:.0f}\nA = {A:.0f}\nB = {B:.0f}')

if __name__ == "__main__":
    main()
```
