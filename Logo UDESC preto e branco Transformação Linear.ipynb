```python
import numpy as np
import matplotlib.image as image
import matplotlib.pyplot as plt
from PIL import Image
import pandas as pd

mamute_Feliz_matriz=image.imread('/content/mamute_Feliz.png')
fig, ax = plt.subplots()
ax.imshow(mamute_Feliz_matriz)
ax.axis('off')
fig.show()
```

```python
mamute_Feliz_matriz
```

```python
R = mamute_Feliz_matriz[:,:,0]
G = mamute_Feliz_matriz[:,:,0]
B = mamute_Feliz_matriz[:,:,0]
```

```python
rr = 0*mamute_Feliz_matriz
rr[:,:,0] = R
print(rr.shape)
figr, axr = plt.subplots()
axr.imshow(np.array(rr))
axr.axis('off')
figr.show()
```

```python
gg = 0*mamute_Feliz_matriz
gg[:,:,1] = G
print(gg.shape)
figg, axg = plt.subplots()
axg.imshow(np.array(gg))
axg.axis('off')
figg.show()
```

```python
bb = 0*mamute_Feliz_matriz
bb[:,:,2] = B
print(bb.shape)
figb, axb = plt.subplots()
axb.imshow(np.array(bb))
axb.axis('off')
figb.show()
```

```python
cinza = 0.2126 * R + 0.7152 * G + 0.0722 * B
```

```python
figgray, axgrey = plt.subplots()
axgrey.imshow(cinza, cmap='Greys_r', interpolation='nearest')
figgray.show()
```
