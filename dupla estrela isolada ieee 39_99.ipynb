# Explicação de Cálculos de Desequilíbrio em Bancos de Capacitores em Duplo Wye
## 1. Introdução
Este notebook tem como objetivo explicar as fórmulas utilizadas para calcular desequilíbrios em bancos de capacitores conectados em duplo wye, conforme apresentado na Tabela 7 da norma IEEE Std C37.99-2012. As fórmulas são essenciais para entender como falhas, como fusíveis queimados, afetam o comportamento elétrico do banco e como ajustar sistemas de proteção adequadamente. Aqui, você encontrará explicações detalhadas, exemplos numéricos e um código Python interativo para facilitar a compreensão e aplicação prática.

## 2. Definição das Variáveis
Antes de explorar as fórmulas, vamos definir as variáveis utilizadas:


* $f$:   Número de fusíveis queimados
* $G$:   Fator de aterramento (0 para aterrado, 1 para não aterrado).
* $P$:   Número total de unidades em paralelo por fase.
* $S$:   Número de grupos em série por fase.
* $Cg$:  Capacitância do grupo paralelo afetado.
* $Cs$:  Capacitância da string afetada.
* $Cp$:  Capacitância da fase afetada.
* $Vng$: Tensão neutro-terra.
* $Vln$: Tensão linha-neutro na fase afetada.
* $Vcu$: Tensão na unidade afetada.
* $Ve$:  Tensão nos elementos afetados.
* $Iu$:  Corrente pelos capacitores afetados.
* $Iph$: Corrente na fase afetada.
* $Ig$:  Corrente de desequilíbrio.
* $In$:  Corrente neutro-terra.
* $Id$:  Corrente diferencial.
* $Cu$:  Capacitância residual da unidade afetada (valor típico fornecido como exemplo).
* $Vg$:  Fator de divisão de tensão nos elementos internos (valor típico fornecido como exemplo).



# 1. Capacitância do Grupo Paralelo, $C_g$

**Fórmula:**  
$$C_g = \frac{P-f}{P} + C_u$$

**Comentário:** Capacitância do grupo de capacitores que inclui a unidade afetada. Para todas as unidades do grupo, exceto a afetada, a capacitância por unidade é 1. Para a unidade afetada, é $C_u$.

---

## Derivação:

- Em um grupo paralelo com $P$ unidades, a capacitância total é a soma das capacitâncias individuais.
- Normalmente, cada unidade tem capacitância por unidade de 1. Quando $f$ fusíveis queimam, $f$ unidades são efetivamente removidas, reduzindo o número de unidades saudáveis para $P - f$.
- A capacitância do grupo é então a contribuição das unidades saudáveis, $\frac{P-f}{P}$ (proporção de unidades restantes), mais a capacitância da unidade afetada, $C_u$, que pode ser reduzida dependendo da condição de falha.

---

## Exemplo:

- Para $f = 0$, $C_g = \frac{P-0}{P} + 0 = 1$ (condição normal).
- Para $f = 1$, $C_g = \frac{P-1}{P} + C_u$, onde $C_u$ varia com $f$ e $G$, como mostrado na tabela (e.g., $C_g = 0.9750$ para $f = 1$).

```python
def calculate_C_i(N, f):
    return (N - f) / N
```

```python
def calcular_V_g(Su, N, f):
    num = Su * N
    den = (Su - 1) * (N - f) + N
    V_g = num / den
    return V_g
```

```python
def calcular_C_u(Su, Ci):
    """Capacitância da unidade afetada"""
    num = Su * Ci
    den = Ci * (Su - 1) + 1
    return num / den
```

```python
def calcular_C_g(P, C_u):
    """Capacitância do grupo paralelo com unidades falhas."""
    return (P - 1 + C_u) / P
```

```python
def calcular_C_s(S, C_g):
    """Capacitância da string afetada."""
    num = S * C_g
    den = C_g * (S - 1) + 1
    return num / den
```

```python
def calcular_C_p(C_s, P, Pt):
    """Capacitância da fase afetada"""
    num = (C_s * P) + Pt - P
    return num / Pt
```

```python
def calcular_V_ng(G, C_p):
    """Tensão neutro-terra."""
    if G == 1:  # não aterrado
        return 3 / (2 + C_p) - 1
    else:
        return 0
```

```python
C_i = calculate_C_i(N, f)
V_g = calcular_V_g(Su, N, f)
C_u = calcular_C_u(Su, C_i)
C_g = calcular_C_g(P, C_u)
C_s = calcular_C_s(S, C_g)
C_p = calcular_C_p(C_s, P, Pt)
V_ng = calcular_V_ng(G, C_p)

print(np.round(C_i,4))
print(np.round(V_g,4))
print(np.round(C_u,4))
print(np.round(C_g,4))
print(np.round(C_s,4))
print(np.round(C_p,4))
print(np.round(V_ng,4))
```
