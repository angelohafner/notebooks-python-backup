```python
# Instalação silenciosa do Plotly (caso ainda não esteja instalado)
!pip install -q plotly

import numpy as np
import plotly.graph_objects as go
import plotly.io as pio

# Define o renderer adequado para Google Colab
pio.renderers.default = 'colab'

# Frequência de 1 kHz a 10 MHz
f = np.logspace(4, 5, 1000)
w = 2 * np.pi * f

# Parâmetros do circuito
Cmain = 100e-9
Lstray = 20e-6
L1 = 100e-6
L2 = 148e-6
C2 = 81e-9
R2 = 16.1

# Impedâncias
Z_Cmain = 1 / (1j * w * Cmain)
Z_Lstray = 1j * w * Lstray
Z_L1 = 1j * w * L1
Z_L2 = 1j * w * L2
Z_C2 = 1 / (1j * w * C2)
Z_R2 = R2

Z_C2_R2_parallel = 1 / (1 / Z_C2 + 1 / Z_R2)
Z_branch = Z_L2 + Z_C2_R2_parallel
Z_tuning = 1 / (1 / Z_L1 + 1 / Z_branch)

Z_total = Z_Cmain + Z_Lstray + Z_tuning
Z_mag = np.abs(Z_total)

# Gráfico com altura aumentada
fig = go.Figure()
fig.add_trace(go.Scatter(x=f, y=Z_mag, mode='lines', name='|Z_total|'))

fig.update_layout(
    title='Frequency Response of PLC Coupling Circuit',
    xaxis=dict(title='Frequency [Hz]', type='linear'),
    yaxis=dict(title='|Z_total| [Ohm]', type='log'),
    height=1000,  # AUMENTA a altura da figura aqui
    template='plotly_white'
)

fig.show()
```
