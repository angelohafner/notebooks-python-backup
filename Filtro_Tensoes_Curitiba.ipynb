```python
import cmath
import numpy as np

def calcular_corrente(V, R, L, C, I1, I2, w, t):
    # Calcular a impedância do circuito RLC
    X_L = w * L
    X_C = 1 / (w * C)
    Z = cmath.sqrt(R**2 + (X_L - X_C)**2)
    
    # Calcular a corrente através do circuito RLC
    I_RLC = V / Z

    # Calcular as correntes I1 e I2
    I1 = I1 * cmath.exp(1j * w * t)
    I2 = I2 * cmath.exp(1j * w * t)

    # Calcular a corrente total
    I = I_RLC + I1 + I2

    return I

w_fund = 2 * np.pi * 60  # frequência angular em rad/s (60 Hz)
tt = np.arange(0, 0.1, 0.1e-3)
V = 10 * np.exp(1j*(1*w_fund*tt))

R = 1  # ohms
L = 0  # henries
C = 9e99  # farads
phi_h = -30 * np.pi/180
tt = np.arange(0, 100e-3, 100e-6)
Ih = np.zeros((7,len(tt)))

for hh in range(3, 7, 2):
    Ih = 10/hh * np.exp(1j*(hh*w*t+phi_h))





I = calcular_corrente(V, R, L, C, I1, I2, w, t)
print(f"A corrente total no circuito é {I.real} + {I.imag}j A.")
```
