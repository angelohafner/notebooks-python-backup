```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = 100
b = 30
k = 10
w = 5
```

```python
dt = 1e-6
t = np.arange(0, 50, dt)
n = len(t)
x = np.zeros(n, dtype='float')
v = np.zeros(n, dtype='float')
a = np.zeros(n, dtype='float')
x[0] = 0.
v[0] = 0.
f = 10000 * np.sin(w*t)
```

```python
for k in range(n-1):
    a[k] = 1/m * ( -b*v[k] - k*x[k] + f[k])
    v[k+1] = v[k] +  a[k]*dt
    x[k+1] = x[k] +  v[k]*dt
```

```python
plt.figure(figsize=(25, 5), dpi=200)
plt.plot(t, x, lw=0.5)
```
