```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def quiverLog(Ex,Ey):
    mag = np.sqrt(Ex**2 + Ey**2)
    ExNorm = Ex / mag
    EyNorm = Ey / mag

    log_Ex = np.log(mag/np.min(mag))  * ExNorm;
    log_Ey = np.log(mag/np.min(mag))  * EyNorm;

    return [log_Ex, log_Ey]
```

```python
k = 9e9
Q1 = 1e-6
Q2 = -1e-6
x = np.linspace(-2, 2, 25)
y = np.linspace(-2, 2, 25)
xx, yy = np.meshgrid(x,y, indexing = 'ij')
```

```python
r1 = np.sqrt((xx+1)**2 + yy**2)
V1 = k*Q1/r1

r2 = np.sqrt((xx-1)**2 + yy**2)
V2 = k*Q2/r2
```

```python
# plt.scatter(xx, yy, s=0.1)
plt.scatter(-1, 0, label="Q_1")
plt.scatter( 1, 0, label="Q_2")
plt.contour(xx, yy, V1)
plt.contour(xx, yy, V2)
plt.grid()
```

```python
plt.scatter(-1, 0, label="Q_1")
plt.scatter( 1, 0, label="Q_2")
plt.contour(xx, yy, V1+V2)
plt.grid()
```

```python
Ex,Ey = np.gradient(V1+V2)
Ex = - Ex
Ey = - Ey
log_Ex, log_Ey = quiverLog(Ex,Ey)
```

```python
plt.scatter(-1, 0, label="Q_1")
plt.scatter( 1, 0, label="Q_2")
plt.contour(xx, yy, V1+V2)
plt.quiver(xx.T, yy.T, log_Ex.T, log_Ey.T)
plt.grid()
```
