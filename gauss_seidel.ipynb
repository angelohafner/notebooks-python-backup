```python
import numpy as np

# Definição do sistema
# CUIDAR PARA SEMPRE COLOCAR MATRIZES DIAGONALMENTE DOMINANTES, CASO CONTRÁRIO PODE NÃO CONVERGIR
nn = 3
A = np.array([[ 5, 2, 1],
              [ 1, -5, 1],
              [-1, 0, 9]])
b = np.array([0, 1, 5])
x = np.zeros(nn)  # Inicialização do vetor solução com zeros

# Verificação de elementos diagonais zero
if np.any(np.diag(A) == 0):
    print("Erro: Elemento diagonal zero encontrado. O método não pode ser aplicado.")
else:
    # Parâmetros do método
    tol = 1e-6  # Tolerância para convergência
    max_iter = 1000  # Número máximo de iterações

    # Iterações do método de Gauss-Seidel
    for interacao in range(max_iter):
        x_old = x.copy()  # Guarda a solução da iteração anterior
        for ii in range(nn):
            soma = 0
            for jj in range(nn):
                if ii != jj:
                    soma += A[ii, jj] * x[jj]
            x[ii] = (b[ii] - soma) / A[ii, ii]

        # Verifica convergência
        if np.linalg.norm(x - x_old, ord=np.inf) < tol:
            print(f"Convergência alcançada após {interacao + 1} iterações.")
            break
    else:
        print("Número máximo de iterações atingido sem convergência.")

    # Resultados
    print("Solução aproximada:", x)
    print("Verificação (A @ x):", A @ x)
    print("Solução exata (inv(A) @ b):", np.linalg.inv(A) @ b)
```
