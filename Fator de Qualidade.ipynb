```python
import numpy as np
#P = 8.543e3     # reactor rated active power
P = 3.6e3       # reactor rated active power
Q = 134.5e3     # reactor rated reactive power
I = 102         # reactor rated current
```

```python
R_calc = P / I**2
X_calc = Q / I**2
L_calc = X_calc/(2*np.pi*60)
print(f"Reactor calculated resistance = {R_calc}")
print(f"Reactor calculated reactance  = {X_calc}")
print(f"Reactor calculated inductance = {L_calc}")
```

```python
Q0_reactor = X_calc/R_calc
print(f"Reactor quality factor @  60 Hz: {Q0_reactor}")
print(f"Reactor quality factor @ 300 Hz: {5*Q0_reactor}")
```
