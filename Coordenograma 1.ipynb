```python
import numpy as np
from matplotlib.ticker import EngFormatter
formatter_VA = EngFormatter(unit='VA', places=2)
formatter_V = EngFormatter(unit='V', places=2)
formatter_A = EngFormatter(unit='A', places=2)
```

## Dados da Concessionária

```python
cc3f = 3225
cc2f = 2984
ccft_max = 2328
ccft_min = 193
```

## Dados do Cliente

```python
S_trafos = [500, 750]
S_trafos = 1e3*np.array(S_trafos)
S_trafos

S_trafos_formatted = [formatter_VA.format_data(x) for x in S_trafos]
print(f"Potencia dos transformadores ordem decrescente = {S_trafos_formatted}")
```

## Correntes Nominais

```python
Demanda = 850 * 1e3
D = 1.05 # entre 1.05 e 1.30
B = 0.2 # entre 0.2 e 0.33
FDesl = 0.92
Vff = 13.8 * 1e3;

I_demanda = Demanda / (np.sqrt(3) * Vff * FDesl )
I_trafos = S_trafos / (np.sqrt(3) * Vff )

I_trafos_formatted = [formatter_A.format_data(x) for x in I_trafos]
print(f"Correntes dos trafos = {I_trafos_formatted}")
I_demanda_formatted = formatter_A.format_data(I_demanda)
print(f"Demanda de corrente = {I_demanda_formatted}")

corrente_51F = I_demanda * D
corrente_51F_formatted = formatter_A.format_data(corrente_51F)
print(f"corrente_51F = {corrente_51F_formatted}")

corrente_51FN = corrente_51F * B
corrente_51FN_formatted = formatter_A.format_data(corrente_51FN)
print(f"corrente_51FN = {corrente_51FN_formatted}")
```

## Correntes de Curto Circuito

```python
Z_trafos = [4.25, 5.32]
Z_trafos = np.array(Z_trafos) / 100

I_curto_trafos_linha = I_trafos / Z_trafos
I_curto_trafos_fase = I_curto_trafos_linha / np.sqrt(3)

I_curto_trafos_linha_formatted = [formatter_A.format_data(x) for x in I_curto_trafos_linha]
I_curto_trafos_fase_formatted = [formatter_A.format_data(x) for x in I_curto_trafos_fase]

print(f"I_curto_trafos_linha = {I_curto_trafos_linha_formatted}")
print(f"I_curto_trafos_fase_formatted = {I_curto_trafos_fase_formatted}")
```
