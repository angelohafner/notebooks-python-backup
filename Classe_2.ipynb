```python
import math

class TrianguloRetangulo:
    def __init__(self, cateto1, cateto2):
        """
        Inicializa o triângulo retângulo com os dois catetos.

        :param cateto1: Comprimento do primeiro cateto (deve ser positivo)
        :param cateto2: Comprimento do segundo cateto (deve ser positivo)
        :raises ValueError: Se qualquer cateto for não positivo
        """
        if cateto1 <= 0 or cateto2 <= 0:
            raise ValueError("Os catetos devem ser positivos.")
        self._cateto1 = float(cateto1)
        self._cateto2 = float(cateto2)
        self._hipotenusa = self._calcular_hipotenusa()
        self._angulo1 = self._calcular_angulo1()  # Em radianos, oposto ao cateto1
        self._angulo2 = self._calcular_angulo2()  # Em radianos, oposto ao cateto2

    def _calcular_hipotenusa(self):
        """
        Calcula a hipotenusa usando o teorema de Pitágoras.

        :return: Comprimento da hipotenusa
        """
        return math.sqrt(self._cateto1**2 + self._cateto2**2)

    def _calcular_angulo1(self):
        """
        Calcula o ângulo oposto ao cateto1 em radianos.

        :return: Ângulo em radianos
        """
        return math.atan(self._cateto1 / self._cateto2)

    def _calcular_angulo2(self):
        """
        Calcula o ângulo oposto ao cateto2 em radianos.

        :return: Ângulo em radianos
        """
        return math.atan(self._cateto2 / self._cateto1)

    def calcular_area(self):
        """
        Calcula a área do triângulo retângulo.

        :return: Área do triângulo
        """
        return (self._cateto1 * self._cateto2) / 2

    def calcular_perimetro(self):
        """
        Calcula o perímetro do triângulo retângulo.

        :return: Perímetro do triângulo
        """
        return self._cateto1 + self._cateto2 + self._hipotenusa

    def seno_angulo1(self):
        """
        Calcula o seno do ângulo oposto ao cateto1.

        :return: Seno do ângulo1
        """
        return math.sin(self._angulo1)

    def cosseno_angulo1(self):
        """
        Calcula o cosseno do ângulo oposto ao cateto1.

        :return: Cosseno do ângulo1
        """
        return math.cos(self._angulo1)

    def tangente_angulo1(self):
        """
        Calcula a tangente do ângulo oposto ao cateto1.

        :return: Tangente do ângulo1
        """
        return math.tan(self._angulo1)

    def seno_angulo2(self):
        """
        Calcula o seno do ângulo oposto ao cateto2.

        :return: Seno do ângulo2
        """
        return math.sin(self._angulo2)

    def cosseno_angulo2(self):
        """
        Calcula o cosseno do ângulo oposto ao cateto2.

        :return: Cosseno do ângulo2
        """
        return math.cos(self._angulo2)

    def tangente_angulo2(self):
        """
        Calcula a tangente do ângulo oposto ao cateto2.

        :return: Tangente do ângulo2
        """
        return math.tan(self._angulo2)

    def imprimir_informacoes(self):
        """
        Imprime todas as informações do triângulo, incluindo catetos, hipotenusa,
        área, perímetro, ângulos em graus e funções trigonométricas dos ângulos.
        """
        area = self.calcular_area()
        perimetro = self.calcular_perimetro()
        angulo1_graus = math.degrees(self._angulo1)
        angulo2_graus = math.degrees(self._angulo2)
        print(f"Cateto 1: {self._cateto1}")
        print(f"Cateto 2: {self._cateto2}")
        print(f"Hipotenusa: {self._hipotenusa}")
        print(f"Área: {area}")
        print(f"Perímetro: {perimetro}")
        print(f"Ângulo 1: {angulo1_graus:.2f}°")
        print(f"Ângulo 2: {angulo2_graus:.2f}°")
        print(f"Seno do ângulo 1: {self.seno_angulo1():.4f}")
        print(f"Cosseno do ângulo 1: {self.cosseno_angulo1():.4f}")
        print(f"Tangente do ângulo 1: {self.tangente_angulo1():.4f}")
        print(f"Seno do ângulo 2: {self.seno_angulo2():.4f}")
        print(f"Cosseno do ângulo 2: {self.cosseno_angulo2():.4f}")
        print(f"Tangente do ângulo 2: {self.tangente_angulo2():.4f}")

    def __str__(self):
        """
        Retorna uma representação em string do triângulo.

        :return: String representando o triângulo
        """
        return f"Triângulo Retângulo com catetos {self._cateto1} e {self._cateto2}"

# Exemplo de uso:
if __name__ == "__main__":
    triangulo = TrianguloRetangulo(3, 4)
    triangulo.imprimir_informacoes()
```
