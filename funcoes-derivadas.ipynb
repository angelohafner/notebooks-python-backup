```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1.5, 1.5, 21)
y1 = x**1
y3 = x**3
y5 = x**5

plt.plot(x, y1, label="$x$")
plt.plot(x, y3, label="$x^3$")
plt.plot(x, y5, label="$x^5$")
plt.legend()
plt.grid()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1.5, 1.5, 21)
y3 = x**3+2
d_y3 = 3*x**2

plt.plot(x, y3, label="$x^3$")
plt.plot(x, d_y3, label="$3x^2$")
plt.legend()
plt.grid()
```
