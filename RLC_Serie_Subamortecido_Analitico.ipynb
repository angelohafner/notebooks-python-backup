# 🔹 Circuito RLC Série Subamortecido e sua Analogia com o Movimento Torcional

---

## 1. Circuito RLC Série — Caso Subamortecido

### 1.1 Equação diferencial geral

Para um circuito RLC série em regime transitório (fonte desligada após $t = 0$):

$$
v_R + v_L + v_C = 0
$$

Sabendo que:

$$
v_R = Ri, \quad v_L = L \frac{di}{dt}, \quad v_C = \frac{1}{C} \int i \, dt
$$

Derivando em relação ao tempo e rearranjando, obtemos a **equação diferencial homogênea**:

$$
L \frac{d^2 i}{dt^2} + R \frac{di}{dt} + \frac{i}{C} = 0
$$

Dividindo tudo por $L$:

$$
\frac{d^2 i}{dt^2} + \frac{R}{L} \frac{di}{dt} + \frac{1}{LC} i = 0
$$

---

### 1.2 Forma padrão e parâmetros característicos

$$
\frac{d^2 i}{dt^2} + 2\alpha \frac{di}{dt} + \omega_0^2 i = 0
$$

onde:

- $\alpha = \frac{R}{2L}$ → **fator de amortecimento**
- $\omega_0 = \frac{1}{\sqrt{LC}}$ → **frequência natural não amortecida**

---

### 1.3 Caso Subamortecido ($\alpha < \omega_0$)

Neste regime, o sistema oscila, mas com amplitude decrescente no tempo.  
A solução da equação diferencial é:

$$
i(t) = e^{-\alpha t} \left( A_1 \cos(\omega_d t) + A_2 \sin(\omega_d t) \right)
$$

onde:

$$
\omega_d = \sqrt{\omega_0^2 - \alpha^2}
$$

é a **frequência amortecida** (menor que $\omega_0$).

---

### 1.4 Comportamento físico

- A corrente $i(t)$ oscila com frequência $\omega_d$.
- A envoltória $e^{-\alpha t}$ representa o decaimento exponencial da amplitude.
- O sistema troca energia entre o indutor (campo magnético) e o capacitor (campo elétrico), com perdas resistivas no resistor.


---

## 2. Analogia com o Movimento Torcional de um Eixo

O circuito RLC é **análogo** ao sistema mecânico rotacional massa–mola–amortecedor, conforme o quadro abaixo:

| Sistema Elétrico (RLC Série) | Sistema Mecânico Rotacional (Eixo) |
|-------------------------------|-------------------------------------|
| Corrente $i(t)$ | Velocidade angular $\omega(t)$ |
| Carga $q(t)$ | Deslocamento angular $\theta(t)$ |
| Indutor $L$ | Momento de inércia $J$ |
| Resistor $R$ | Coeficiente de atrito viscoso $b$ |
| Capacitor $C$ | Mola torcional (constante $K$) |
| Tensão $v(t)$ | Torque aplicado $T(t)$ |

---

### 2.1 Equação diferencial do movimento torcional

$$
J \frac{d^2 \theta}{dt^2} + b \frac{d\theta}{dt} + K \theta = 0
$$

Comparando com o circuito RLC:

$$
L \frac{d^2 q}{dt^2} + R \frac{dq}{dt} + \frac{1}{C} q = 0
$$

A correspondência é direta:

$$
L \leftrightarrow J, \quad R \leftrightarrow b, \quad \frac{1}{C} \leftrightarrow K
$$

---

### 2.2 Caso Subamortecido (Sistema Oscilatório Amortecido)

Quando:

$$
\zeta = \frac{b}{2\sqrt{JK}} < 1
$$

o sistema é **subamortecido**, e o deslocamento angular segue:

$$
\theta(t) = e^{-\zeta \omega_n t} (B_1 \cos(\omega_d t) + B_2 \sin(\omega_d t))
$$

onde:

$$
\omega_n = \sqrt{\frac{K}{J}}, \quad \omega_d = \omega_n \sqrt{1 - \zeta^2}
$$

O comportamento é o mesmo da corrente $i(t)$ no circuito RLC.

---

### 2.3 Interpretação física da analogia

| Fenômeno Elétrico | Fenômeno Mecânico |
|--------------------|-------------------|
| Energia armazenada no capacitor → campo elétrico | Energia elástica armazenada na mola torcional |
| Energia armazenada no indutor → campo magnético | Energia cinética rotacional do eixo |
| Dissipação no resistor | Perdas por atrito viscoso |
| Oscilação da corrente amortecida | Oscilação angular amortecida |


---

## 3. Resumo e Interpretação Final

| Conceito | Circuito RLC | Sistema Torcional |
|-----------|---------------|-------------------|
| Equação | $L i'' + R i' + \frac{1}{C} i = 0$ | $J \theta'' + b \theta' + K \theta = 0$ |
| Parâmetro de amortecimento | $\alpha = \frac{R}{2L}$ | $\zeta = \frac{b}{2\sqrt{JK}}$ |
| Frequência natural | $\omega_0 = \frac{1}{\sqrt{LC}}$ | $\omega_n = \sqrt{\frac{K}{J}}$ |
| Caso subamortecido | $\alpha < \omega_0$ | $\zeta < 1$ |
| Solução | Oscilação amortecida de $i(t)$ | Oscilação amortecida de $\theta(t)$ |

---

### 🧭 **Resumo Conceitual**

> O circuito RLC série subamortecido e o sistema torcional com atrito viscoso leve são ambos sistemas **de segunda ordem oscilatórios amortecidos**, em que a energia alterna entre duas formas armazenadas (magnética/elétrica ou cinética/elástica), enquanto parte é dissipada (resistor ou atrito).

---


````python
# Interactive analytic solution for underdamped series RLC step response (Plotly)
# User-defined initial conditions (q0, i0) or vc0 -> q0 = C*vc0
# English comments; SI prefixes on axes using d3 "~s" tick format.

import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import plotly.io as pio

# Ensure figures display inline in Google Colab / Jupyter
# Options: "colab", "notebook_connected", "notebook", "iframe_connected", etc.
pio.renderers.default = "colab"

def rlc_step_analytic_ic_plotly(
    E=1.0, R=10.0, L=100e-3, C=10e-6,
    q0=None, i0=None, vc0=None,
    n_periods=6, n_points=20000,
    out_dir=".", base_name="rlc_plotly",
    show_combined=True, show_separate=False,
    save_png=False, display_inline=True
):
    """
    Interactive Plotly visualization for the analytic step response of an underdamped series RLC.

    Inputs:
      - E [V]: DC step amplitude at t = 0+
      - R [ohm], L [H], C [F]
      - q0 [C], i0 [A]: initial conditions at t = 0+
        (alternatively provide vc0 [V] => q0 = C * vc0)
      - n_periods: number of damped periods to simulate
      - n_points: number of time samples
      - out_dir: directory to save HTML/PNG
      - base_name: base file name for combined figure
      - show_combined: create 3 stacked subplots (q, i, di/dt) in one HTML
      - show_separate: also create 3 separate HTML figures
      - save_png: export PNGs (requires 'kaleido')
      - display_inline: call .show() to display figures inline

    Returns:
      dict with created figures (keys: "combined", "q_t", "i_t", "di_dt" as applicable)
    """

    # ---- Natural parameters ----
    alpha  = R / (2.0 * L)            # damping factor [1/s]
    omega0 = 1.0 / np.sqrt(L * C)     # undamped natural frequency [rad/s]
    Delta  = (R / L)**2 - 4.0 / (L * C)  # discriminant

    print(f"omega0 = {omega0:.6f} rad/s")
    print(f"alpha  = {alpha:.6f} 1/s")
    print(f"Delta  = {Delta:.6f}")

    # Only underdamped case
    if Delta >= 0.0:
        print("O circuito NÃO é subamortecido (Delta >= 0). Ajuste R, L e C.")
        return {}

    omegad = np.sqrt(omega0**2 - alpha**2)
    Td = 2.0 * np.pi / omegad
    print(f"omegad = {omegad:.6f} rad/s, Td = {Td:.6e} s")

    # ---- Time axis ----
    t_max = n_periods * Td
    t = np.linspace(0.0, t_max, int(n_points))

    # ---- Particular (DC step) ----
    q_ss = C * E

    # ---- Initial conditions ----
    # Priority: (q0, i0) if provided; else use vc0; default rest (0, 0)
    if q0 is None and vc0 is not None:
        q0 = C * vc0
    if q0 is None:
        q0 = 0.0
    if i0 is None:
        i0 = 0.0

    print(f"Initial conditions: q0 = {q0:.6e} C, i0 = {i0:.6e} A (vc0 = {q0/C:.6e} V)")

    # ---- Homogeneous solution for x = q - q_ss ----
    # x(t) = e^{-alpha t} [ A cos(omegad t) + B sin(omegad t) ]
    # with:
    #   A = x(0) = q0 - q_ss
    #   x'(0) = i0 = -alpha*A + omegad*B  =>  B = (i0 + alpha*A)/omegad
    A = q0 - q_ss
    B = (i0 + alpha * A) / omegad

    exp_term = np.exp(-alpha * t)
    cos_term = np.cos(omegad * t)
    sin_term = np.sin(omegad * t)

    # Charge q(t)
    q = q_ss + exp_term * (A * cos_term + B * sin_term)

    # Current i(t) = dq/dt (analytic derivative)
    i = exp_term * ((-alpha * A + omegad * B) * cos_term
                    - (alpha * B + omegad * A) * sin_term)

    # di/dt via circuit equation: L di/dt + R i + q/C = E
    di_dt = (E - R * i - q / C) / L

    figs = {}

    # ---- Combined figure (3 stacked subplots) ----
    if show_combined:
        fig = make_subplots(
            rows=3, cols=1, shared_xaxes=True, vertical_spacing=0.08,
            subplot_titles=("Charge q(t)", "Current i(t)", "Current derivative di/dt")
        )

        fig.add_trace(go.Scatter(x=t, y=q, name="q(t)", mode="lines"), row=1, col=1)
        fig.add_trace(go.Scatter(x=t, y=i, name="i(t)", mode="lines"), row=2, col=1)
        fig.add_trace(go.Scatter(x=t, y=di_dt, name="di/dt", mode="lines"), row=3, col=1)

        fig.update_layout(
            height=900, width=1000, template="plotly_white",
            title=f"Series RLC Step Response (Underdamped) — R={R} Ω, L={L} H, C={C} F, E={E} V",
            margin=dict(l=60, r=40, t=60, b=40)
        )

        # Axis titles and SI-prefix ticks
        fig.update_xaxes(title_text="Time (s)", tickformat="~s", row=3, col=1)
        fig.update_yaxes(title_text="Charge (C)", tickformat="~s", row=1, col=1)
        fig.update_yaxes(title_text="Current (A)", tickformat="~s", row=2, col=1)
        fig.update_yaxes(title_text="di/dt (A/s)", tickformat="~s", row=3, col=1)

        # Save and show
        combined_html = f"{out_dir}/{base_name}_combined.html"
        fig.write_html(combined_html)
        print(f"Arquivo salvo: {combined_html}")

        if display_inline:
            fig.show()

        if save_png:
            try:
                fig.write_image(f"{out_dir}/{base_name}_combined.png", scale=2)
                print(f"Arquivo salvo: {out_dir}/{base_name}_combined.png")
            except Exception as e:
                print("PNG não salvo (kaleido ausente?). Execute: pip install -U kaleido")
                print(f"Detalhes: {e}")

        figs["combined"] = fig

    # ---- Separate figures (optional) ----
    if show_separate:
        series = {
            "q_t":   (q,     "Charge on the capacitor q(t)",  "Charge (C)"),
            "i_t":   (i,     "Loop current i(t)",             "Current (A)"),
            "di_dt": (di_dt, "Current derivative di/dt",      "di/dt (A/s)")
        }
        for key, (y, title, ylab) in series.items():
            f = go.Figure()
            f.add_trace(go.Scatter(x=t, y=y, name=key, mode="lines"))
            f.update_layout(
                height=400, width=1000, template="plotly_white",
                title=title + f" — R={R} Ω, L={L} H, C={C} F, E={E} V",
                xaxis_title="Time (s)", yaxis_title=ylab,
                margin=dict(l=60, r=40, t=60, b=40)
            )
            f.update_xaxes(tickformat="~s")
            f.update_yaxes(tickformat="~s")

            out_html = f"{out_dir}/{key}.html"
            f.write_html(out_html)
            print(f"Arquivo salvo: {out_html}")

            if display_inline:
                f.show()

            if save_png:
                try:
                    f.write_image(f"{out_dir}/{key}.png", scale=2)
                    print(f"Arquivo salvo: {out_dir}/{key}.png")
                except Exception as e:
                    print("PNG não salvo (kaleido ausente?). Execute: pip install -U kaleido")
                    print(f"Detalhes: {e}")

            figs[key] = f

    print("Simulação analítica (Plotly) concluída.")
    return figs


# ============================
# Exemplo de uso (Colab/Jupyter)
# ============================
# Caso clássico (repouso): q0 = 0, i0 = 0
figs = rlc_step_analytic_ic_plotly(
    E=1.0, R=10.0, L=100e-3, C=10e-6,
    q0=0.0, i0=0.0,       # ou vc0=0.0  (equivalente a q0 = C*vc0)
    n_periods=8, n_points=30000,
    out_dir=".", base_name="rlc_plotly",
    show_combined=True, show_separate=False,
    save_png=False,       # mude para True se tiver 'kaleido' instalado
    display_inline=True   # garante exibição inline no Colab
)
````

# ⚡ Prompt Markdown — Simulação Analítica Interativa de um Circuito RLC Série Subamortecido

Gere um **código Python completo e executável** no **Google Colab** que resolva **analiticamente** a resposta ao degrau de um **circuito RLC série subamortecido**, exibindo **gráficos interativos em Plotly**.

## 🎯 Objetivo

Desenvolver um programa que calcule e visualize as variáveis:

- 🔹 **Carga no capacitor:** $q(t)$  
- 🔹 **Corrente na malha:** $i(t)$  
- 🔹 **Derivada da corrente:** $\dfrac{di}{dt}$  

para um circuito **RLC série** excitado por um **degrau DC de 1 V**, no **regime subamortecido**.

---

## ⚙️ Especificações Técnicas

### 🔸 Equação diferencial

O circuito RLC série é descrito por:

$$
L \frac{d^2 i}{dt^2} + R \frac{di}{dt} + \frac{i}{C} = 0
$$

ou, na forma padrão:

$$
\frac{d^2 i}{dt^2} + 2\alpha \frac{di}{dt} + \omega_0^2 i = 0
$$

onde:

- $\alpha = \dfrac{R}{2L}$  → fator de amortecimento  
- $\omega_0 = \dfrac{1}{\sqrt{LC}}$  → frequência natural não amortecida  

O caso **subamortecido** ocorre se:

$$
\Delta = \left(\frac{R}{L}\right)^2 - \frac{4}{LC} < 0
$$

---

## 🧮 Requisitos do Código

1. **Entrada:**  
   - Parâmetros elétricos: `E`, `R`, `L`, `C`  
   - Condições iniciais ajustáveis:
     - `q0` → carga inicial (C)  
     - `i0` → corrente inicial (A)  
     - `vc0` → tensão inicial no capacitor (V), com $q_0 = C \, v_{c0}$

2. **Cálculo analítico:**  
   - Resolver para:
     $$
     q(t) = q_{ss} + e^{-\alpha t}\left[A \cos(\omega_d t) + B \sin(\omega_d t)\right]
     $$
     com
     $$
     A = q_0 - q_{ss}, \quad B = \frac{i_0 + \alpha A}{\omega_d}, \quad \omega_d = \sqrt{\omega_0^2 - \alpha^2}
     $$

   - Derivar analiticamente $i(t) = \dfrac{dq}{dt}$  
   - Calcular $\dfrac{di}{dt}$ via equação de malha:

     $$
     L \frac{di}{dt} + R i + \frac{q}{C} = E
     $$

3. **Verificação de regime:**  
   - Se $\Delta \ge 0$, exibir mensagem:  
     > “O circuito **não é subamortecido**. Ajuste R, L e C.”
