```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = 60
w = 2*np.pi*f
t = np.linspace(0, 6/f, 6*128)
```

```python
fig, ax = plt.subplots(figsize=(10,5))
v = np.sin(w*t)
ax.plot(60*t, v, color='black', lw=1.5)
p = np.ones(len(v))
ax.plot(60*t, p, color='black', lw=0.5, ls=":")
ax.plot(60*t, -p, color='black', lw=0.5, ls=":")
z = np.zeros(len(v))
ax.plot(60*t, z, color='black', lw=0.5, ls=":")

ax.set_xlabel("Tempo em ciclos de 60 Hz")
ax.set_ylabel("Amplitude [pu]")

fig.show()
fig.savefig("Corrente Simérica.png",  dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(10,5))
v = 1 + np.sin(w*t)
ax.plot(60*t, v, color='black', lw=1.5)
p = 1 + np.ones(len(v))
ax.plot(60*t, p, color='black', lw=0.5, ls=":")
p = np.zeros(len(v))
ax.plot(60*t, p, color='black', lw=0.5, ls=":")
z = 1 + np.zeros(len(v))
ax.plot(60*t, z, color='black', lw=0.5, ls=":")

ax.set_xlabel("Tempo em ciclos de 60 Hz")
ax.set_ylabel("Amplitude [pu]")

fig.show()
fig.savefig("Totally offset ac wave.png",  dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(10,5))
v = np.sin(w*t) * np.exp(-t/0.05)
ax.plot(60*t, v, color='black', lw=1.5)
p = np.exp(-t/0.05)*np.ones(len(v))
ax.plot(60*t, p, color='black', lw=0.5, ls=":")
ax.plot(60*t, -p, color='black', lw=0.5, ls=":")
z = np.zeros(len(v))
ax.plot(60*t, z, color='black', lw=0.5, ls=":")
ax.set_xlabel("Tempo em ciclos de 60 Hz")
ax.set_ylabel("Amplitude [pu]")
fig.show()
fig.savefig("Decaying symmetrical ac wave.png",  dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(10,5))
k = 1
tau = 0.02
v = k*np.exp(-t/tau)
ax.plot(60*t, v, color='black', lw=1.5)
p = np.ones(len(v))
ax.plot(60*t, p, color='black', lw=0.5, ls=":")
ax.plot(60*t, -p, color='black', lw=0.5, ls=":")
z = np.zeros(len(v))
ax.plot(60*t, z, color='black', lw=0.5, ls=":")
ax.set_xlabel("Tempo em ciclos de 60 Hz")
ax.set_ylabel("Amplitude [pu]")
fig.show()
fig.savefig("Decaying dc wave.png",  dpi=300, bbox_inches='tight')
```

```python
fig, ax = plt.subplots(figsize=(10,5))
k = 0.4
tau = 0.02
v = np.sin(w*t-np.arcsin(k)) + k * np.exp(-t/tau)
ax.plot(60*t, v, color='black', lw=1.5)
p = np.ones(len(v))
ax.plot(60*t, p, color='black', lw=0.5, ls=":")
ax.plot(60*t, -p, color='black', lw=0.5, ls=":")
z = np.zeros(len(v))
ax.plot(60*t, z, color='black', lw=0.5, ls=":")
ax.set_xlabel("Tempo em ciclos de 60 Hz")
ax.set_ylabel("Amplitude [pu]")
fig.show()
fig.savefig("Asymmetrical fault current ac wave.png",  dpi=300, bbox_inches='tight')
```
