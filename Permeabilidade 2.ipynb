```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
nCurvas = 5
nPontos = 11

YY1 = np.zeros((nCurvas, nPontos))
YY2 = np.zeros((nCurvas, nPontos))
ss = np.arange(0, nCurvas, 1)
XX = np.zeros((nCurvas, nPontos +1))
XX.shape
```

```python
plt.figure(dpi=75, figsize=(12,4))
for ii in range(nCurvas):
    XX[ii,0:nPontos] = np.linspace(ii, ii+1, nPontos)
    XX[ii,nPontos] = ss[ii]
    YY1[ii,:] = np.exp(-5*(XX[ii,0:nPontos]-ss[ii]))
    plt.plot(XX[ii,0:nPontos], YY1[ii,:], label='saln'+str(ss[ii]))

for ii in range(nCurvas):
    XX[ii,0:nPontos] = np.linspace(ii, ii+1, nPontos)
    XX[ii,nPontos] = ss[ii]
    YY2[ii,:] = np.exp(1/2*(XX[ii,0:nPontos]-ss[ii])) - 1
    plt.plot(XX[ii,0:nPontos], YY2[ii,:], label='saln'+str(ss[ii]), ls=":")    
plt.legend()
plt.show()
```
