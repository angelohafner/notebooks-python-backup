### 2. Importando o TensorFlow

```python
import tensorflow as tf
```

### 3. Tensores no TensorFlow

```python
# Tensor constante
t_const = tf.constant([[1, 2], [3, 4]])

# Tensor variável (permite mudanças de valor)
t_var = tf.Variable([[1, 2], [3, 4]])

print("Tensor constante:", t_const)
print("Tensor variável:", t_var)
```

### 4. Operações com Tensores

```python
# Soma de tensores
t_soma = tf.add(t_const, t_var)

# Multiplicação de tensores
t_mult = tf.multiply(t_const, t_var)

print("Soma dos tensores:\n", t_soma)
print("Multiplicação dos tensores:\n", t_mult)
```

### 5. Gradientes Automáticos

```python
# Definindo uma função simples: f(x) = x^2
def func(x):
    return x ** 2

# Variável inicial
x = tf.Variable(3.0)

# Abrindo o Gradient Tape para calcular o gradiente
with tf.GradientTape() as tape:
    y = func(x)

# Calculando o gradiente de y em relação a x
grad = tape.gradient(y, x)
print("Gradiente de f(x) = x^2 no ponto x = 3:", grad)
```

### 6. Construindo uma Rede Neural Simples

```python
# Definindo o modelo
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=128, activation='relu', input_shape=(784,)),
    tf.keras.layers.Dense(units=10, activation='softmax')
])

# Compilando o modelo
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Exibindo a estrutura do modelo
model.summary()
```

### 7. Treinando a Rede Neural

```python
# Carregando os dados MNIST
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Pré-processando os dados
x_train, x_test = x_train / 255.0, x_test / 255.0

# Definindo o modelo com uma camada Flatten
model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),  # Achatando as imagens 28x28 em vetores de 784 elementos
    tf.keras.layers.Dense(units=128, activation='relu'),
    tf.keras.layers.Dense(units=10, activation='softmax')
])

# Compilando o modelo
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Treinando o modelo
model.fit(x_train, y_train, epochs=5)

# Avaliando o modelo
test_loss, test_acc = model.evaluate(x_test, y_test)
print('Acurácia no teste:', test_acc)
```

### 8. Previsões com o Modelo Treinado

```python
# Fazendo previsões
predicoes = model.predict(x_test)

# Exibindo a previsão para o primeiro exemplo de teste
print("Predição para a primeira imagem:", tf.argmax(predicoes[0]))
print("Valor real:", y_test[0])
```

### 9. Salvando e Carregando o Modelo

```python
# Salvando o modelo
model.save('meu_modelo.h5')

# Carregando o modelo salvo
novo_modelo = tf.keras.models.load_model('meu_modelo.h5')
```
