```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def T(theta, vetor):
    M = np.array([[np.cos(theta), -np.sin(theta)],
                 [np.sin(theta), np.cos(theta)]])
    w = M@vetor
    P, D = np.linalg.eig(M)
    return P, D, w
```

```python
xx = np.linspace(-2, 2, 21)
yy = xx
nn = len(xx)
X, Y = np.meshgrid(xx, yy)
xx
```

```python
ss = plt.scatter(X,Y)
v = ss.get_offsets()
vv = v.data.T
vv.shape
```

```python
import matplotlib.animation
from matplotlib import rc

plt.rcParams["animation.html"] = "html5"
plt.rcParams['figure.dpi'] = 100  
plt.ioff()
fig, ax = plt.subplots()
fig = plt.figure(dpi = 100, figsize=(10,8))

deg = np.pi/180
nn = 72
d_ang = 360 / nn

def animate(k):
    
    plt.cla()
    ww = T(k*10*deg,vv)[2]
    
    ax = plt.axes()
    ax.set_xlim(-1, 1)
    ax.set_ylim(-1, 1)
    
    #ax.scatter(vv[0,:], vv[1,:], label='v', color='red')
    ax.scatter(ww[0,:], ww[1,:], label='w', color='green')


anim = matplotlib.animation.FuncAnimation(fig, animate, frames=nn, interval=100)
anim
```
