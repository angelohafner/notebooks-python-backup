```python
# Em seguida, importe as bibliotecas necessárias
import ezdxf
import xml.etree.ElementTree as ET
from google.colab import files

# Carregar o arquivo DXF
uploaded = files.upload()

# Supondo que o nome do arquivo seja 'Drawing.dxf'
dxf_file_path = list(uploaded.keys())[0]
dwg = ezdxf.readfile(dxf_file_path)

# Criar o elemento raiz do XML
root = ET.Element("Drawing")

# Adicionar camadas ao XML
layers = ET.SubElement(root, "Layers")
for layer in dwg.layers:
    layer_element = ET.SubElement(layers, "Layer", name=layer.dxf.name)

# Adicionar entidades ao XML
entities = ET.SubElement(root, "Entities")
for entity in dwg.entities:
    entity_element = ET.SubElement(entities, entity.dxftype())
    for key, value in entity.dxf.all_existing_dxf_attribs().items():
        entity_element.set(key, str(value))

# Converter a árvore XML para uma string e salvar em um arquivo
tree = ET.ElementTree(root)
xml_file_path = 'Drawing.xml'
tree.write(xml_file_path)

# Baixar o arquivo XML gerado
files.download(xml_file_path)

print(f"DXF file converted to XML and saved to {xml_file_path}")
```
