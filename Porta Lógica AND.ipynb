```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import tensorflow as tf
from keras.callbacks import LambdaCallback
tf.random.set_seed(42)
from matplotlib import rcParams
rcParams['font.family'] = 'Palatino Linotype'
from keras.callbacks import LambdaCallback
```

```python
# Entrada
X = [[0, 0], [0,1], [1,0], [1,1]]
# Saída
Y = [[0],[0],[0],[1]]
```

```python
# Gerar o Modelo
model = tf.keras.Sequential([tf.keras.layers.Dense(1, activation='sigmoid', input_shape=(2,))])

model.compile(optimizer= tf.keras.optimizers.SGD(lr=0.01), 
              loss="mse", 
              metrics=['mse','mae'])
```

```python
model.save_weights('ttt.txt')
```

```python
# Treinar o Modelo
history = model.fit(X, Y, batch_size=4, epochs=100, verbose=0)
print("treinada...")
```

```python
mpl.rcParams['font.size'] = 10
fig, ax = plt.subplots(figsize=(10/2.54,6/2.54))
ax.loglog(history.history["mse"], label=r"$\sqrt{\frac{\sum{\varepsilon _{i}^{2}}}{n}}$", lw=0.5)
ax.loglog(history.history["mae"], label=r"$\frac{\sum{\left| {{\varepsilon }_{i}} \right|}}{n}$", ls="--")
ax.set_xlabel("Época #")
ax.set_ylabel("Erro")
ax.legend(frameon=False)
fig.show()
```

```python
X_pred = X
Y_pred = model.predict(X_pred)
np.round(Y_pred,0)
```
