```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
t = np.linspace(-2*np.pi, 2*np.pi, 256)
v =     10*np.sin(1*t) +5*np.sin(2*t) + 3*np.sin(3*t) + 2*np.cos(2*t) + 1*np.cos(5*t)
```

```python
plt.plot(t,v)
```

Termos do cosseno

$${{a}_{k}}=\frac{\int_{{}}^{{}}{f\left( x \right)\cos \left( kx \right)dx}}{\int_{{}}^{{}}{\cos \left( kx \right)\cos \left( kx \right)dx}}$$

```python
k = 5
num = np.trapz(v * np.cos(k*t), t)
den = np.trapz(np.cos(k*t)*np.cos(k*t), t)
aa = num/den
aa
```

```python
nn = 11
a = np.zeros(nn)
b = np.zeros(nn)
for k in range(nn):
    # cosseno
    num = np.trapz(v * np.cos(k*t), t)
    den = np.trapz(np.cos(k*t)*np.cos(k*t), t)
    if den != 0:
        a[k] = num/den
    # seno
    num = np.trapz(v * np.sin(k*t), t)
    den = np.trapz(np.sin(k*t)*np.sin(k*t), t)
    if den != 0:
        b[k] = num/den
```

```python
freqs = np.arange(nn)
plt.plot(freqs, a, label='cosseno')
plt.plot(freqs, b, label='seno')
plt.scatter(freqs, a, label='cosseno')
plt.scatter(freqs, b, label='seno')
plt.legend()
plt.show()
```
