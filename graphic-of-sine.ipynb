```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
omega = 2 * np.pi * 60
omega
```

```python
tempo = np.linspace(0, 1/60, 16)
tempo
```

```python
tensao = 10 * np.sin(omega*tempo)
tensao
```

```python
fig, ax = plt.subplots()
ax.plot(1e3*tempo, tensao)
ax.set_xlabel(r"$t \, {\rm[ms]}$")
ax.set_ylabel(r"$v(t) \, {\rm[V]}$")
ax.grid()
fig.set_size_inches(5, 10)
fig.show()
```
