```python
import math

class TrianguloRetangulo:
    def __init__(self, cateto1, cateto2):
        if cateto1 <= 0 or cateto2 <= 0:
            raise ValueError("Os catetos devem ser positivos.")
        self.cateto1 = cateto1
        self.cateto2 = cateto2

    def calcular_hipotenusa(self):
        return math.sqrt(self.cateto1**2 + self.cateto2**2)

    def calcular_area(self):
        return (self.cateto1 * self.cateto2) / 2

    def calcular_perimetro(self):
        hipotenusa = self.calcular_hipotenusa()
        return self.cateto1 + self.cateto2 + hipotenusa

    def calcular_angulos(self):
        angulo1 = math.degrees(math.atan(self.cateto1 / self.cateto2))
        angulo2 = math.degrees(math.atan(self.cateto2 / self.cateto1))
        return angulo1, angulo2

    def imprimir_informacoes(self):
        hipotenusa = self.calcular_hipotenusa()
        area = self.calcular_area()
        perimetro = self.calcular_perimetro()
        angulo1, angulo2 = self.calcular_angulos()
        print(f"Cateto 1: {self.cateto1}")
        print(f"Cateto 2: {self.cateto2}")
        print(f"Hipotenusa: {hipotenusa}")
        print(f"Área: {area}")
        print(f"Perímetro: {perimetro}")
        print(f"Ângulos agudos: {angulo1}° e {angulo2}°")

    def __str__(self):
        return f"Triângulo Retângulo com catetos {self.cateto1} e {self.cateto2}"

# Exemplo de uso:
if __name__ == "__main__":
    triangulo = TrianguloRetangulo(3, 4)
    triangulo.imprimir_informacoes()
```
