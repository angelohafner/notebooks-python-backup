Intersecção entre 
$${x^2} + {y^2} + {z^2} = 1$$ e
$$x + y + z = 0$$

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
theta = np.linspace(0, 2 * np.pi, 50)
phi = np.linspace(0, np.pi, 50)
r = 1
xe = r * np.outer(np.cos(theta), np.sin(phi))
ye = r * np.outer(np.sin(theta), np.sin(phi))
ze = r * np.outer(np.ones(np.size(theta)), np.cos(phi))
```

```python
x = np.linspace(-1.2, 1.2, 50)
y = x
xp, yp = np.meshgrid(x, y)
zp = -xp -yp
zp [zp>1.2] = np.nan
zp [zp<-1.2] = np.nan
```

```python
fig = plt.figure(dpi=100,figsize=(6,6))
ax = plt.axes()
zpc = -xe -ye
cs = ax.contour(xe, ye, zpc-ze, [0])
#ax.clabel(cs, cs.levels, inline=True)
p = cs.collections[0].get_paths()[0]
v = p.vertices
xc = v[:,0]
yc = v[:,1]
zc = -xc -yc
xc
```

```python
plt.plot(xc, yc)
```

```python
fig = plt.figure(dpi=200,figsize=(5,5))
ax = ax = plt.axes(projection='3d')
ax.plot_surface(xe, ye, ze, alpha=0.50, linewidth=0, edgecolor='none')
ax.plot_surface(xp, yp, zp, alpha=0.50, linewidth=0, edgecolor='none')
ax.plot(xc, yc, zc, linewidth=5)
#ax.plot(xc, yc, -1.2*zc/zc, linewidth=5)
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 1.2)
ax.set_zlim(-1.2, 1.2)
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel('$z$')
ax.view_init(10,-30)
plt.show()
```
