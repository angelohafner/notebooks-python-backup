```python
import numpy as np
import matplotlib.pyplot as plt

# Constantes
R = 8.314  # Constante dos gases em J/(mol K)
deltaH = 280  # entalpia de mistura em J/mol

# Faixa de temperaturas
temperaturas = np.linspace(299, 301, 2)  # de 300K a 1500K
frações_molares = np.linspace(0, 1, 100)  # frações molares de 0 a 1

# Função para calcular a energia livre de Gibbs de mistura
def gibbs_mix(x1, T):
    x2 = 1 - x1
    if x1 == 0 or x2 == 0:
        return 0
    entropia_mix = R * (x1 * np.log(x1) + x2 * np.log(x2))
    gibbs = deltaH * x1 * x2 + T * entropia_mix
    return gibbs

# Matriz para armazenar os valores de energia livre
gibbs_values = np.zeros((len(frações_molares), len(temperaturas)))

for i, x1 in enumerate(frações_molares):
    for j, T in enumerate(temperaturas):
        gibbs_values[i, j] = gibbs_mix(x1, T)

# Plot do diagrama de fase aproximado
plt.figure(figsize=(10, 6))
plt.contour(temperaturas, frações_molares, gibbs_values, levels=50, cmap='viridis')
plt.colorbar(label='Energia Livre de Gibbs (J/mol)')
plt.xlabel('Temperatura (K)')
plt.ylabel('Fração Molar de Ouro (X1)')
plt.title('Diagrama de Fase para Liga Au-Cu (Solução Regular)')
plt.show()
```
