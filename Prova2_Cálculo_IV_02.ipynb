```python
#Bernardo Skovronski
#Bibliotecas:
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Dados:
t = sp.symbols('t')
x = sp.Function('x')
pp0= 4 #densidade liquído
p=2 #densidade cubo
g= 9.81 #gravidade  
l= 3 #lado do cubo
```

```python
#Cis:
cis = {x(0): 2, x(t).diff(t).subs(t, 0):3 }
cis
```

```python
#EDOs:
w=(pp0*g/(p*l))
a= 0.25 #atrito 1"
b= 0.50 #atrito 2"
c=1 #atrito 3"
d=2 #atrito 4"
e=4 #atrito 5"
edo= sp.Eq(x(t).diff(t,t)+w*x(t),0) #"sem atrito"
edo1= sp.Eq(x(t).diff(t,t)+x(t).diff(t)*a+w*x(t),0) #"com atrito 1"
edo2= sp.Eq(x(t).diff(t,t)+x(t).diff(t)*b+w*x(t),0) #"com atrito 2"
edo3= sp.Eq(x(t).diff(t,t)+x(t).diff(t)*c+w*x(t),0) #"com atrito 3"
edo4= sp.Eq(x(t).diff(t,t)+x(t).diff(t)*d+w*x(t),0) #"com atrito 4"
edo5= sp.Eq(x(t).diff(t,t)+x(t).diff(t)*e+w*x(t),0) #"com atrito 5"
edo
edo1
edo2
edo3
edo4
edo5
```

```python
#Soluções EDOs:
Sol= sp.dsolve(edo, ics=cis)
Sol

Sol1= sp.dsolve(edo1, ics=cis)
Sol1

Sol2= sp.dsolve(edo2, ics=cis)
Sol2

Sol3= sp.dsolve(edo3, ics=cis)
Sol3

Sol4= sp.dsolve(edo4, ics=cis)
Sol4

Sol5= sp.dsolve(edo5, ics=cis)
Sol5
```

```python
#Soluções Direitas:
Soldireita=Sol.args[1]
Soldireita

Soldireita1=Sol1.args[1]
Soldireita1

Soldireita2=Sol2.args[1]
Soldireita2

Soldireita3=Sol3.args[1]
Soldireita3

Soldireita4=Sol4.args[1]
Soldireita4

Soldireita5=Sol5.args[1]
Soldireita5
```

```python
numsol=sp.lambdify(t,Soldireita)  

numsol1=sp.lambdify(t,Soldireita1) 

numsol2=sp.lambdify(t,Soldireita2)

numsol3=sp.lambdify(t,Soldireita3)

numsol4=sp.lambdify(t,Soldireita4)

numsol5=sp.lambdify(t,Soldireita5)
```

```python
#Gráfico:
fig = plt.figure(dpi=160, figsize=(15,5))
ax1 = plt.axes()
ax1.set_xlabel("Tempo (s)")
ax1.set_ylabel("Deslocamento Vertical")
tt = np.linspace(0,10,1000)
ax1.plot(tt,numsol(tt), color='blue',  lw=2, label='Oscilação sem atrito')
ax1.plot(tt,numsol1(tt), color='purple',  lw=2, label='Oscilação com atrito de 0.25')
ax1.plot(tt,numsol2(tt), color='red',  lw=1.5, label='Oscilação com atrito de 0.50')
ax1.plot(tt,numsol3(tt), color='orange',  lw=1.5, label='Oscilação com atrito de 1')
ax1.plot(tt,numsol4(tt), color='yellow',  lw=1.5, label='Oscilação com atrito de 2')
ax1.plot(tt,numsol5(tt), color='black',  lw=1, label='Oscilação com atrito de 4')
ax1.grid(alpha=0.5)
plt.legend()
plt.show()
```
