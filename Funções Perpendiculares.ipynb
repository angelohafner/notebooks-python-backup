```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0.1, 4*np.pi, 45)
plt.figure(figsize=(15,5), dpi=100)

plt.plot(x-0.03, np.sin(x), color='red')
markerline, stemlines, baseline = plt.stem(x-0.03, np.sin(x), markerfmt='o', label='$\\sin(x)$', use_line_collection=True)
plt.setp(markerline,color='red')
plt.setp(stemlines, color='red')


plt.plot(x+0.03, np.cos(x), color='green')
markerline, stemlines, baseline = plt.stem(x+0.03, np.cos(x), markerfmt='o', label='$\\cos(x)$', use_line_collection=True)
plt.setp(markerline,color='green')
plt.setp(stemlines, color='green')

plt.fill_between(x, np.cos(x)*np.sin(x), color='blue', alpha=0.2)
plt.plot(x, np.cos(x)*np.sin(x), color='blue')
markerline, stemlines, baseline = plt.stem(x, np.cos(x)*np.sin(x), markerfmt='o', label='$\\cos(x) \\, \\sin(x)$', use_line_collection=True)
plt.setp(markerline,color='blue')
plt.setp(stemlines, color='blue')

plt.grid(alpha = 0.5)
gradeX = np.linspace(0,4*np.pi,9)
plt.xticks(gradeX)

plt.xlabel('$x$')
plt.ylabel('$y$')

plt.legend()
plt.show()
```
