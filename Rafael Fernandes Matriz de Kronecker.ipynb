```python
import numpy as np

# Definindo as matrizes A e B
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[0, 5],
              [6, 7]])

# Calculando o produto de Kronecker de A e B
K = np.kron(A, B)

print("Matriz de Kronecker K:")
print(K)

# Definindo o vetor x
x = np.array([1, 2, 3, 4])

# Multiplicando a matriz de Kronecker K pelo vetor x
result = np.dot(K, x)

print("Resultado de K * x:")
print(result)
```
