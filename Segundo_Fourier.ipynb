```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
t = np.linspace(-np.pi, np.pi, 256)
v =  t**3 + 2
```

```python
plt.plot(t,v)
plt.show()
```

Termos do cosseno

$${{a}_{k}}=\frac{\int_{{}}^{{}}{f\left( x \right)\cos \left( kx \right)dx}}{\int_{{}}^{{}}{\cos \left( kx \right)\cos \left( kx \right)dx}}$$

```python
k = 5
num = np.trapz(v * np.cos(k*t), t)
den = np.trapz(np.cos(k*t)*np.cos(k*t), t)
aa = num/den
aa
```

```python
nn = 50
a = np.zeros(nn)
b = np.zeros(nn)
for k in range(nn):
    # cosseno
    num = np.trapz(v * np.cos(k*t), t)
    den = np.trapz(np.cos(k*t)*np.cos(k*t), t)
    if den != 0:
        a[k] = num/den
    # seno
    num = np.trapz(v * np.sin(k*t), t)
    den = np.trapz(np.sin(k*t)*np.sin(k*t), t)
    if den != 0:
        b[k] = num/den

a = np.round(a, 3)
b = np.round(b, 3)
```

```python
freqs = np.arange(nn)
plt.plot(freqs, a, label='cosseno')
plt.plot(freqs, b, label='seno')
plt.scatter(freqs, a, label='cosseno')
plt.scatter(freqs, b, label='seno')
plt.legend()
plt.show()
```

```python
tt = np.linspace(-10, 10, 1000)
rec = np.zeros(len(tt))
fig, ax = plt.subplots()
for ii in range(nn):
    y = a[ii]*np.cos(ii*tt) + b[ii]*np.sin(ii*tt)
    # ax.plot(tt, y, label=str(ii))
    rec = rec + y
ax.plot(tt, rec, label='reconstruido')
ax.scatter(t, t**3+2, label='original', alpha=0.5, s=1, color='black')
ax.legend()
fig.dpi = 100
fig.show()
```
