```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

x = np.linspace(-3, 3, 1000)
y = np.linspace(-3, 3, 1000)
X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

# Flatten the arrays to export as columns
data = {
    'X': X.ravel(),
    'Y': Y.ravel(),
    'Z': Z.ravel()
}


# Optional: plot
fig, ax = plt.subplots()
contours= ax.contour(X, Y, Z)
ax.clabel(contours, inline=True, fontsize=8)  # Add labels on contour lines
plt.show()
```

```python

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z)
plt.show()
```
