```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
t = np.linspace(0, 10/28, 256*1000)
w_fund = 2*np.pi*28
```

```python
v = np.zeros(len(t))
for k in range(31):
    if k !=0:
        v = v + 10/k * np.sin(k*w_fund*t) + 15/k * np.cos(k*w_fund*t)
```

```python
plt.figure(figsize=(15,5))
plt.plot(t,v)
plt.show()
```

Termos do cosseno

$${{a}_{k}}=\frac{\int_{{}}^{{}}{f\left( x \right)\cos \left( kx \right)dx}}{\int_{{}}^{{}}{\cos \left( kx \right)\cos \left( kx \right)dx}}$$

```python
values = np.array([t, v]).T
df = pd.DataFrame(values, columns=['tempo', 'vibração'])
df
```
