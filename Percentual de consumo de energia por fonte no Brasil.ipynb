```python
import pandas as pd
import matplotlib.pyplot as plt


# Com base nas informações disponíveis, vamos criar um gráfico do percentual de cada fonte de energia consumida no Brasil em 2023.
# Os dados são fictícios e servem apenas como exemplo.

dados_consumo_energia = {
    "Fonte de Energia": ["Hidrelétrica", "Eólica", "Solar", "Biomassa", "Fóssil"],
    "Percentual": [60, 15, 10, 5, 10]  # Valores percentuais fictícios
}

# Criando DataFrame
df_consumo_energia = pd.DataFrame(dados_consumo_energia)

# Plotando o gráfico de pizza
plt.figure(figsize=(8, 8), dpi=600)
plt.pie(df_consumo_energia["Percentual"], labels=df_consumo_energia["Fonte de Energia"], autopct='%1.1f%%', startangle=140)
%plt.title("Percentual de Consumo de Energia por Fonte no Brasil em 2023")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.

plt.show()
```
