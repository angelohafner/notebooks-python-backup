# Massa - Mola - Amortecedor


```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np
```

```python
t = sp.symbols('t')
t
```

```python
x = sp.Function('x')
x(t)
```

```python
m1 = sp.Rational(10,1)
b1 = sp.Rational(2,1)
k1 = sp.Rational(5,1)
```

Condições iniciais

```python
cis= {x(1): 2, x(t).diff(t).subs(t, 1): 1}
cis
```

Definindo a EDO

$$m x{\left(t \right)} + b \frac{d}{d t} x{\left(t \right)} + k \frac{d^{2}}{d t^{2}} x{\left(t \right)} = 0 $$

```python
edo = sp.Eq( (t**2-3*t)*x(t).diff(t,t) + t*x(t).diff(t) - (t+3)*x(t) , 0 )
edo
```

```python
cis
```

Encontrando a solução da EDO

```python
xSol = sp.dsolve(edo)
# xSol = sp.dsolve(edo)
xSol.rhs
```

```python
from sympy.plotting import plot
plot(t**2)
```

```python
xSolDireita = xSol.args[1]
xSolDireita
```

Concertendo a expressão numa forma que se possa calucular para fazer o gráfico.

```python
solNum = sp.lambdify(t, xSolDireita)
```

```python
fig = plt.figure(dpi=100, figsize=(16/2.54,10/2.54))
ax1 = plt.axes()
ax1.set_xlabel("$t$")
ax1.set_ylabel("$x$")
tt = np.linspace(0,50,1000)
ax1.plot(tt,solNum(tt))
ax1.grid()
```
