```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Definição da função de plot
def grafico_hiperbole(a, b):
    c = np.sqrt(a**2 + b**2)
    x = np.linspace(-4.0*a, 4.0*a, 40)
    y = np.linspace(-4.0*b, 4.0*b, 40)
    x, y = np.meshgrid(x, y)

    z = x**2/a**2 - y**2/b**2 - 1
    plt.figure()
    plt.contour(x, y, z, levels=[0], colors='b')
    plt.scatter(-c, 0, color='r')
    plt.scatter(+c, 0, color='r')
    plt.scatter(-a, 0, color='b')
    plt.scatter(+a, 0, color='b')
    plt.scatter(0, -b, color='g')
    plt.scatter(0, +b, color='g')
    plt.axis('equal')
    plt.xlabel('$x$')
    plt.ylabel('$y$')
    plt.xlim(-3*a, 3*a)
    plt.ylim(-3*b, 3*b)
    plt.title(r'$\dfrac{x^2}{a^2} - \dfrac{y^2}{b^2} = 1 $')
    plt.grid(ls=':')
    plt.show()


a_slider = FloatSlider(min=0.1, max=2.0, step=0.1, value=1.0, description='a')
b_slider = FloatSlider(min=0.1, max=2.0, step=0.1, value=1.0, description='b')

interact(grafico_hiperbole, a=a_slider, b=b_slider)
```
