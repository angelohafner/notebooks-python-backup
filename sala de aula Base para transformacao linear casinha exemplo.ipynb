```python
import matplotlib.pyplot as plt
import numpy as np

# Coordenadas fornecidas
x_externo = [1.0, 4.0, 4.0, 4.2, 2.5, 0.8, 1.0, 1.0]
y_externo = [1.0, 1.0, 3.0, 2.7, 5.0, 2.7, 3.0, 1.0]

x_janela = [1.5, 1.5, 2.5, 2.5, 1.5]
y_janela = [1.5, 2.5, 2.5, 1.5, 1.5]

# Criação do gráfico
plt.figure(figsize=(10, 10))
plt.scatter(x_externo, y_externo, color='red', label='Pe')
plt.scatter(x_janela, y_janela, color='blue', label='Pj')
plt.plot(x_externo, y_externo, color='red', linestyle='-', linewidth=1.5, label='Le')
plt.plot(x_janela, y_janela, color='blue', label='Lj')
plt.legend()
plt.xlim(-6 , 6)
plt.ylim(-6 , 6)
plt.grid()
plt.show()
```

```python
vetor_externo = np.array([x_externo, y_externo])
vetor_externo
```

```python
matriz_transformacao = np.array([[-1, 0],
                                 [0, 1]])
matriz_transformacao
```

```python
vetor_externo_trasnformado = matriz_transformacao @ vetor_externo
vetor_externo_trasnformado
```

```python
plt.figure(figsize=(10, 10))
plt.scatter(vetor_externo_trasnformado[0, :],
            vetor_externo_trasnformado[1, :], color='red', label='Pe')
plt.plot(vetor_externo_trasnformado[0, :],
            vetor_externo_trasnformado[1, :], color='red', label='Le')


plt.xlim(-6 , 6)
plt.ylim(-6 , 6)
```
