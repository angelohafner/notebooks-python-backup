```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-6, 7, 0.5)
y = np.sin(x)

fig, ax = plt.subplots()
fig.set_size_inches(12, 4)
ax.plot(x, y)
ax.scatter(x, y)
```

```python
delta_x = np.diff(x)
delta_y = np.diff(y)
derivada = delta_y / delta_x
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(12, 4)
ax.plot(x, y, label = '$\\sin x$')
ax.scatter(x, np.cos(x), label = '$\\cos x$')
ax.plot(x[0:25], derivada, label='$\\frac{d[\\sin x]}{dx}$')
ax.legend()
```

```python
derivada2 = np.gradient(y, x)
fig, ax = plt.subplots()
fig.set_size_inches(12, 4)
ax.plot(x, y, label = '$\\sin x$')
ax.scatter(x, np.cos(x), label = '$\\cos x$')
ax.plot(x[0:25], derivada , label='diff')
ax.plot(x,       derivada2, label='gradient')
ax.legend()
```
