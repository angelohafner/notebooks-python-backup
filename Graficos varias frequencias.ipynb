```python
import numpy as np
import matplotlib.pyplot as plt

def plot_sine_waves(amplitudes, frequencies, time_period=0.034, sample_rate=50000):
    """
    Plota ondas senoidais com as amplitudes e frequências fornecidas.

    Parâmetros:
    amplitudes (lista de floats): Amplitudes das ondas senoidais.
    frequencies (lista de floats): Frequências das ondas senoidais em Hz.
    time_period (float): Duração do eixo do tempo em segundos.
    sample_rate (int): Número de amostras por segundo.
    """
    t = np.linspace(0, time_period, int(sample_rate * time_period), endpoint=False)
    plt.figure(figsize=(30, 5))

    for amplitude, frequency in zip(amplitudes, frequencies):
        y = amplitude * np.sin(2 * np.pi * frequency * t)
        if frequency >= 500:
            plt.plot(t, y, label=f'Amplitude: {amplitude}, Frequência: {frequency} Hz', linewidth=0.2)
        else:
            plt.plot(t, y, label=f'Amplitude: {amplitude}, Frequência: {frequency} Hz')

    plt.title('Gráficos de Ondas Senoidais')
    plt.xlabel('Tempo [s]')
    plt.ylabel('Amplitude')
    plt.legend(loc='upper right')
    plt.grid(True)
    plt.show()
```

```python
# Exemplo de uso
amplitudes = [120, 120, 120]
frequencies = [15, 1000, 60]
plot_sine_waves(amplitudes, frequencies)
```
