```python
import numpy as np

nnn = 5
mat_original = np.random.randint(-5, 6, (nnn, nnn))
mat_identidade = np.eye(nnn)
mat_expandida = np.hstack((mat_original, mat_identidade))
nr_linhas = mat_original.shape[0]
nr_colunas = mat_original.shape[1]
print(mat_expandida)
```

Fazendo com que todos os elementos da diagonal principal sejam diferentes de zero

```python
print(mat_expandida)
print("===")
for linha in range(nr_linhas-1):
    # preciso consultar todas as linhas além da imediatamente abaixo, vai que a imediatamente abaixo tenha elemento nulo também
    ii = linha
    while mat_expandida[linha, linha] == 0:
        linha_com_zero = mat_expandida[linha, :].copy()
        mat_expandida[linha, :] = mat_expandida[linha + ii, :]
        mat_expandida[linha+ii, :] = linha_com_zero
        ii = ii + 1

        if ii >= nr_linhas:
            print("Matriz não tem inversa")
            break

    if ii >= nr_linhas:
        print("Matriz não tem inversa")
        break

print(mat_expandida)
print("===")
```

Diagonal inferior

```python
for linha in range(0, nr_linhas-1, 1):
    ii = linha + 1
    while ii < nr_linhas:
        pivo1 = mat_expandida[linha, linha].copy()
        pivo2 = mat_expandida[ii, linha].copy()
        mat_expandida[ii, :] = (pivo1 * mat_expandida[ii, :] -
                                pivo2 * mat_expandida[linha, :])
        ii = ii + 1

print(mat_expandida[:, 0:nr_colunas])
```

```python
for linha in range(nr_linhas-1, 0, -1):
    ii = linha - 1
    while ii >= 0:
        pivo1 = mat_expandida[linha, linha].copy()
        pivo2 = mat_expandida[ii, linha].copy()
        mat_expandida[ii, :] = (pivo1 * mat_expandida[ii, :] -
                                pivo2 * mat_expandida[linha, :])
        ii = ii - 1

print(mat_expandida[:, 0:nr_colunas])
```

```python
for linha in range(nr_linhas):
    pivo = mat_expandida[linha, linha]
    mat_expandida[linha, :] = mat_expandida[linha, :] / pivo

print("===")
print(mat_expandida)
print("===")
```

```python
mat_inversa = mat_expandida[:, nr_colunas:2*nr_colunas]
print("===")
print(np.round(mat_inversa, 4))
print("===")
```

```python
conferindo = mat_original @ mat_inversa
print(np.round(conferindo, 4))
```

Diagonal Superior

```python
mat_inversa_numpy = np.linalg.inv(mat_original)
print(np.round(mat_inversa_numpy, 4))
```
