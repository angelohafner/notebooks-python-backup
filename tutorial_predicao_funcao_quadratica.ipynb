# Predição de uma Função Quadrática usando TensorFlow

Neste exemplo, vamos usar o TensorFlow para criar um modelo simples que preveja o valor da função $y = x^2$. O modelo será treinado com valores conhecidos de x e y, e depois vamos usá-lo para prever novos valores de y dado um valor de x.

### 1. Importando o TensorFlow

```python
import tensorflow as tf
import numpy as np
```

### 2. Criando os dados para treino

```python
# Gerando os dados de treino (x e y)
x_train = np.arange(-2, 2.1, 0.1)
y_train = x_train**2

print("Dados de treino (x):", x_train)
print("Dados de treino (y):", y_train)
```

### 3. Criando o modelo

```python
# Definindo o modelo
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])  # Camada com um neurônio
])

# Compilando o modelo
model.compile(optimizer='sgd', loss='mean_squared_error')

# Exibindo a estrutura do modelo
model.summary()
```

### 4. Treinando o modelo

```python
# Treinando o modelo com os dados de treino
model.fit(x_train, y_train, epochs=500, verbose=1)

print("Treinamento concluído!")
```

### 5. Fazendo predições

```python
# Fazendo predições com o modelo treinado
x_test = np.array([1.0, -1.0], dtype=float)  # Valores de x para teste
predictions = model.predict(x_test)

print("Valores previstos para x_test:")
for i, pred in enumerate(predictions):
    print(f"x = {x_test[i]}: y previsto = {pred[0]}")
```

### Conclusão

Neste exemplo, treinamos um modelo simples para aprender a relação entre x e y para a função $y = x^2$. Após o treinamento, usamos o modelo para prever valores de y para novos valores de x.
