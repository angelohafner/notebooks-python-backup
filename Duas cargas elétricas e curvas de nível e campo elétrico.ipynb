```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-20, 20, 100)
y = np.linspace(-20, 20, 100)
X, Y = np.meshgrid(x,y)
```

```python
x_q1 = -10
x_q2 = 10
y_q1 = 0
y_q2 = 0
```

```python
k = 9e9
q1 = 1e-6
q2 = -1e-6
D1 = np.sqrt((X-x_q1)**2 + (Y-y_q1)**2)
D2 = np.sqrt((X-x_q2)**2 + (Y-y_q2)**2)
Vq1 = k * q1 / D1
Vq2 = k * q2 / D2
Vtotal = Vq1 + Vq2
```

```python
fig, ax = plt.subplots()
contours= ax.contour(X, Y, Vtotal)
ax.clabel(contours, inline=True, fontsize=8)  # Add labels on contour lines
plt.show()
```
