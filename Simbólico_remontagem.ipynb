```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
```

```python
t, k = sp.symbols("t, k")
w = 2*sp.pi*60
v1 = 100*sp.sin(1*w*t) + 50*sp.sin(2*w*t) + 20*sp.sin(3*w*t) + 10*sp.sin(4*w*t)
v2 = 120*sp.cos(1*w*t) + 52*sp.cos(2*w*t) + 22*sp.cos(3*w*t) + 12*sp.cos(4*w*t)
vv  = v1 + v2
vv
```

```python
num = sp.integrate(vv * sp.sin(1*w*t), (t,0,2*sp.pi))
den = sp.integrate((sp.sin(1*w*t))**2, (t,0,2*sp.pi))
a1 = num/den
a1.evalf()
```

```python
num = sp.integrate(sp.simplify(vv * sp.cos(3*w*t)), (t,0,2*sp.pi))
den = sp.integrate(sp.simplify((sp.cos(3*w*t))**2), (t,0,2*sp.pi))
b3 = num/den
b3
```

```python
b3 = num.simplify()/den.simplify()
b3
```
