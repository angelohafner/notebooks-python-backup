# Dedução da relação entre tensão e fluxo

Partimos da equação fundamental:

$$
-e(t) = \frac{d}{dt}\lambda(t)
$$

## 1. Integração no tempo

Integramos ambos os lados de 0 até t:

$$
-\int_{0}^{t} e(\tau)\,d\tau = \int_{\lambda(0)}^{\lambda(t)} d\lambda
$$

O lado direito resulta em:

$$
\lambda(t) - \lambda(0)
$$

Logo:

$$
\lambda(t) - \lambda(0) = -\int_{0}^{t} e(\tau)\,d\tau
$$

---

## 2. Substituindo a tensão

Dada:

$$
e(t) = 10\sin(10t)
$$

Substituindo:

$$
\lambda(t) - \lambda(0) = \int_{0}^{t} -10\sin(10\tau)\,d\tau
$$

---

## 3. Resolvendo a integral

Sabemos que:

$$
\int \sin(10\tau)\,d\tau = -\frac{1}{10}\cos(10\tau)
$$

Então:

$$
\int -10\sin(10\tau)\,d\tau = \cos(10\tau)
$$

Aplicando limites:

$$
\lambda(t) - \lambda(0) = \left[\cos(10\tau)\right]_0^t
$$

---

## 4. Aplicando condição inicial

Assumindo:

$$
\lambda(0) = 0
$$

Então:

$$
\lambda(t) = \cos(10t) - \cos(0)
$$

Sabendo que:

$$
\cos(0) = 1
$$

---

## 5. Resultado final

$$
\lambda(t) = \cos(10t) - 1
$$

---

## ✔ Conclusão

- A tensão é a derivada negativa do fluxo  
- O fluxo é a integral negativa da tensão  
- A constante vem da condição inicial $\lambda(0)=0$

```python
import numpy as np
import matplotlib.pyplot as plt

# Time vector
t = np.linspace(0, 3*2*np.pi/10, 1000)

# Tensão induzida e(t)
e = 10 * np.sin(10 * t)

# Fluxo concatenado λ(t)
lambda_t = np.cos(10 * t) - 1

# Plot
plt.figure()

plt.plot(t, e, label=r"$e(t)$", lw=1)
plt.plot(t, lambda_t, label=r"$\lambda(t)$", lw=2)

plt.xlabel("Tempo (s)")
plt.ylabel("Volts e Webers")
plt.title("Tensão Induzida e Fluxo Concatenado")
plt.legend()
plt.grid(ls=":")

plt.show()
```
