```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-12*np.pi, 12*np.pi, 800)
y = np.sin(x) / x
```

```python
plt.figure(dpi=600, figsize=(20,5))
plt.grid()
plt.xlabel('Eixo $x$')
plt.ylabel('Eixo $y$')
plt.title(r'Titulo para $y = \dfrac{\sin(x)}{x}$')
plt.plot(x, y)
plt.show()
```
