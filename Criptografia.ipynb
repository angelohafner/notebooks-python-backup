```python
%%manim -qm -v WARNING displayEquations
class displayEquations(Scene):
    def construct(self):
        # Create Tex objects
        first_line =     Text('M	A	M	U	T	E	_	F	E	L	I	Z	!	_	_', font_size=34)
        second_line =    Text('13	1	13	21	20	5	0	6	5	12	9	26	27	0	0', font_size=30)
        equation =       Text('13	1	13	21	20	5	0	6	5	12	9	26	27	0	0', font_size=30)
        terceira_linha = Text('M	A	M	U	T	E	_	F	E	L	I	Z	!	_	_', font_size=34)

        # Position second line underneath first line
        second_line.next_to(first_line, DOWN)
        terceira_linha.next_to(second_line, DOWN)
        
        # Displaying text and equation
        self.play(Write(first_line))
        self.wait(1)
        self.play(Write(second_line))
        self.wait(1)
        self.play(ReplacementTransform(first_line, equation), FadeOut(second_line))
        self.wait(1)
        self.play(Write(terceira_linha))
        self.wait(3)
        self.play(FadeOut(terceira_linha), FadeOut(equation))
```
