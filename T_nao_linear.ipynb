```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-4, 4, 61)
y = x
X, Y = np.meshgrid(x, y)
s = plt.scatter(X, Y)
```

```python
v = s.get_offsets().data.T
v
plt.scatter(v[0,:], v[1,:])
```

$$T\left( {x,y} \right) = \left( {{x^2},3x - 2y} \right)$$

```python
np.shape(v)
```

```python
w = np.zeros(np.shape(v))
w
```

```python
w[0,:] = v[0,:] * v[0,:]
w[1,:] = 3* v[0,:] -2* v[1,:]
w
```

```python
plt.figure(figsize=(10,10))
plt.scatter(v[0,:], v[1,:])
plt.scatter(w[0,:], w[1,:])
```
