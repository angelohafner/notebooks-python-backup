```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import EngFormatter
```

```python




vl = 2.688e3*np.exp(1j*(179.72)*deg)
vc = 22.607e3*np.exp(1j*(-0.28)*deg)
vr = 97.462*np.exp(1j*(89.72)*deg)
ifiltro = 35.147*np.exp(1j*(89.72)*deg)
vt = vl+vc+vr


fig, ax = plt.subplots()

yy = 500*ifiltro
ax.quiver(0, 0, yy.real, yy.imag, angles='xy', scale_units='xy', scale=1, color='green', label='$I_{filtro}$', width=0.005, headwidth=3., headlength=4.)
# ax.quiver(0, 0, vl.real, vl.imag, angles='xy', scale_units='xy', scale=1, color='red', label='$V_L$', width=0.005, headwidth=3., headlength=4.)
ax.quiver(0, 0, vc.real, vc.imag, angles='xy', scale_units='xy', scale=1, color='blue', label='$V_C$', lw=2, width=0.008, headwidth=3., headlength=4.)
x0 = vc.real
y0 = vc.imag
ax.quiver(x0, y0, vl.real, vl.imag, angles='xy', scale_units='xy', scale=1, color='red', label='$V_L$', width=0.005, headwidth=3., headlength=4.)
ax.quiver(0, 0, vt.real, vt.imag, angles='xy', scale_units='xy', scale=1, color='purple', label='$V_{\\phi} = V_R+V_C+V_L$', width=0.005, headwidth=5., headlength=6.)
ax.axis('equal')

intervalo = np.arange(0, 30001, 5000)  # Círculos de 0 a 30,000 com intervalos de 5,000

# Desenhar os círculos de grade
for r in intervalo:
    circulo = plt.Circle((0, 0), r, color='gray', ls=":", lw=0.5, fill=False)
    ax.add_artist(circulo)



ax.legend()


ax.set_xlim(-10e3, 30e3)
ax.set_ylim(-10e3, 30e3)

formatter = EngFormatter()
ax.xaxis.set_major_formatter(formatter)
ax.yaxis.set_major_formatter(formatter)


plt.show()
```
