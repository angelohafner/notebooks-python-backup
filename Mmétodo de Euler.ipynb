```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def ax(t, vx):
    return 0
```

```python
def ay(t, vy):
    return -9.8
```

```python
nn = 100
tt = np.linspace(0, 3, nn)
dt = tt[1] - tt[0]
vx = np.zeros(nn)
vx[0] = 12
vy = np.zeros(nn)
vy[0] = 15
```

```python
for k in range(0, nn-1, 1):
    vx[k+1] = vx[k] + dt * ax(tt[k], vx[k]) 
```

```python
for k in range(0, nn-1, 1):
    vy[k+1] = vy[k] + dt * ay(tt[k], vy[k]) 
```

```python
x = np.zeros(nn)
for k in range(0, nn-1, 1):
    x[k+1] = x[k] + dt * vx[k]
```

```python
y = np.zeros(nn)
for k in range(0, nn-1, 1):
    y[k+1] = y[k] + dt * vy[k]
```

```python
plt.plot(tt, x,tt, y)
plt.show()
```

```python
plt.plot(x, y)
plt.show()
```
