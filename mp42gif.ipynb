```python
from google.colab import files
import subprocess

# Upload video
uploaded = files.upload()

input_file = list(uploaded.keys())[0]
palette = "palette.png"
output_gif = "output.gif"

# Parameter
n = 4

# Get video duration (in seconds)
result = subprocess.run(
    ['ffprobe', '-v', 'error', '-show_entries', 'format=duration',
     '-of', 'default=noprint_wrappers=1:nokey=1', input_file],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT
)

duration = float(result.stdout)
partial_duration = duration / n

# Step 1: generate palette (only first 1/n of video)
!ffmpeg -y -t {partial_duration} -i "$input_file" -vf "fps=30,scale=800:-1:flags=lanczos,palettegen" $palette

# Step 2: generate GIF using palette (only first 1/n)
!ffmpeg -y -t {partial_duration} -i "$input_file" -i $palette -lavfi "fps=30,scale=800:-1:flags=lanczos[x];[x][1:v]paletteuse" $output_gif

# Download GIF
files.download(output_gif)
```
