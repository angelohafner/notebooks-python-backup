```python
def hanoi(n, origem, auxiliar, destino):
    """ Solves the Tower of Hanoi problem using recursion. """
    if n == 1:
        print(f"Mova o disco 1 de {origem} para {destino}")
        return
    hanoi(n - 1, origem, destino, auxiliar)
    print(f"Mova o disco {n} de {origem} para {destino}")
    hanoi(n - 1, auxiliar, origem, destino)

```

```python
# Testando com 3 discos
print("Resolvendo Torre de Hanoi com 3 discos:")
hanoi(3, 'A', 'B', 'C')

```

```python
# Testando com 4 discos
print("\nResolvendo Torre de Hanoi com 4 discos:")
hanoi(4, 'A', 'B', 'C')
```
