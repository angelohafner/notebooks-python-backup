```python
import numpy as np
import matplotlib.pyplot as plt
```

Gráfico de $y=x^2$

```python
x = np.linspace(-10, 10, 11)
fig, ax = plt.subplots()
fig.set_size_inches(15, 3)
ax.scatter(x, x**2)
ax.scatter(x, 0*x)
fig.show()
```

Gráfico de $z=x^2+y^2$

```python
x = np.linspace(-10, 10, 11)
y = np.linspace(-10, 10, 11)
X, Y = np.meshgrid(x, y)
fig, ax = plt.subplots()
fig.set_size_inches(15, 3)
ax.scatter(X, Y)
ax.scatter(x, y)
fig.show()
```

```python
Z = X**2 + Y**2
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, alpha=0.5)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
fig.show()
```

```python
raio = np.linspace(-10, 10, 11)
theta = np.linspace(0, 2*np.pi, 11)
RAIO, THETA = np.meshgrid(raio, theta)
X = RAIO*np.cos(THETA)
Y = RAIO*np.sin(THETA)
Z = RAIO**2

r = 5  # raio do cilindro
Z_cilindro = np.linspace(-10, 180, 111)
THETA_cilindro, Z_cilindro = np.meshgrid(theta, Z_cilindro)
X_cilindro = r * np.cos(THETA_cilindro)
Y_cilindro = r * np.sin(THETA_cilindro)

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, alpha=1.0)  # Superfície original
ax.plot_surface(X_cilindro, Y_cilindro, Z_cilindro, color='r', alpha=1.0)  # Cilindro
ax.view_init(elev=45, azim=75, roll=0)
```
