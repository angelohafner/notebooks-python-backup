```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
x,n,L,lbd = sp.symbols('x n L lambda', real=True, positive=True)
An = sp.integrate((sp.sin(n*lbd*x))**2,(x,0,L))
A = An.factor()
A
```

```python
sp.latex(A)
```

```python
x,n,L,lbd = sp.symbols('x n L lambda', real=True, positive=True)
Bn = sp.integrate((sp.cos(n*lbd*x))**2,(x,0,L))
B = Bn.factor()
B
```

```python
sp.latex(B)
```
