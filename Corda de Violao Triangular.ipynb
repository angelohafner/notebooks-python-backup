```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import time
```

```python
L = 0.5
c = 0.5
T = 5
x = np.linspace(0, L, 100)
t = np.linspace(0, T, 10)
u0 = 8e-3*x * ( np.heaviside(x,1) - np.heaviside(x-L/2,1) ) + (-8e-3*x+4e-3)*( np.heaviside(x-L/2, 1)-np.heaviside(x-L, 1) )
```

```python
nh = 100
b = np.zeros(nh, dtype=float)
l = np.zeros(nh, dtype=float)

for k in range(1,nh):
    l[k] = k*np.pi/L
    b[k] = np.trapz(u0*np.sin(l[k]*x), x) / np.trapz(np.sin(l[k]*x)*np.sin(l[k]*x), x)
```

```python
fig = plt.figure(dpi = 100)
ax = plt.axes()

u = np.zeros((len(t),len(x)), dtype=float)
for i in range(len(t)):
    for k in range(1,nh):
        u[i] = u[i] + b[k] * np.sin(l[k]*x) * np.cos(l[k]*c*t[i])
    
    time.sleep(0.1)
    ax.plot(x, u[i], label=t[i])
    plt.show()
```
