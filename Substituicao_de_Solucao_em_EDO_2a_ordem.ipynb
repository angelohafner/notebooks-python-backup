```python
import sympy as sp
t = sp.symbols('t')
y1 = sp.Pow(t,sp.Rational(1,2))
y2 = sp.Pow(t,-1)
```

```python
D1 = sp.diff(y1,t)
D1
```

```python
D2 = sp.diff(y1,t,t)
D2
```

```python
edo1 = 2 * t**2 * D2  +  3*t*D1  -  y1
edo1
```

```python
D1 = sp.diff(y2,t)
D1
```

```python
D2 = sp.diff(y2,t,t)
D2
```

```python
edo2 = 2 * t**2 * D2  +  3*t*D1  -  y2
edo2
```
