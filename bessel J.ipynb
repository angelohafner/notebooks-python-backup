```python
import math as mt
import numpy as np
import scipy.special as sp
import matplotlib.pyplot as plt


def termo(x, m, n):
    num1 = (-1)**m
    den1 = mt.factorial(m) * mt.factorial(n+m)
    fra2 = ( x / 2 ) ** (2*m + n)
    return num1/den1 * fra2
```

```python

x_lista = np.arange(0, 30.1, 0.1)
fig, ax = plt.subplots()
for n in range(4):
    y_lista = []
    for x in x_lista:
        s = 0
        for m in range(100):
            t = termo(x, m, n)
            s = s + t
        y_lista.append(s)
        # exato = sp.jv(0, x)

    ax.plot(x_lista, y_lista, label=f"n = {n}")

fig.set_size_inches(20, 4)
ax.set_xlabel("x")
ax.set_ylabel("y")
fig.legend()




```
