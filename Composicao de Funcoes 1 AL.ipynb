```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-np.pi, np.pi, 1000)
y = x**3
```

```python
nn = 100
b = np.zeros(nn)
a = np.zeros(nn)

for k in range(nn):
    num = np.trapz(y*np.sin(k*x), x)
    den = np.trapz(np.sin(k*x)*np.sin(k*x), x)
    b[k] = num/den

    num = np.trapz(y*np.cos(k*x), x)
    den = np.trapz(np.cos(k*x)*np.cos(k*x), x)
    a[k] = num/den

b[0] = 0
```

```python
rec = np.zeros(len(x))
for k in range(nn):
    rec = rec + b[k]*np.sin(k*x) + a[k]*np.cos(k*x)
```

```python
plt.plot(x, y, label="$x^2$")
plt.plot(x, rec, label="recuperado")
plt.legend()
```
