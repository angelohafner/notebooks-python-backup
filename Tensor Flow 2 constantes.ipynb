```python
import tensorflow as tf
import tensorflow.compat.v1 as tf
tf.disable_v2_behavior()
```

```python
valor1 = tf.constant(2)
valor2 = tf.constant(3)
```

```python
print(valor1)
```

```python
soma = valor1 + valor2
```

```python
type(soma)
```

```python
print(soma)
```

Quando se cria uma operação fora de uma sessão criam-se os grafos.
A operação somente é processada dentro da Session.

```python
with tf.Session() as sess:
    s = sess.run(soma)
```

```python
print(s)
```
