```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as si
```

$$\begin{align}
  & \left( {{t}^{2}}-3t \right){{y}'}'+t{y}'-\left( t+3 \right)y=0 \\ 
 & y\left( 1 \right)=2 \\ 
 & {y}'\left( 1 \right)=1 \\ 
\end{align}$$

```python
def dydt(y, t):
    dydt_1 = y[1];
    dydt_2 = ( -t*y[1] + (t+3)*y[0] ) / (t**2-3*t)
    return [dydt_1, dydt_2]
```

```python
y0 = [1, 2]
t1 = np.linspace(-2.0, -0.1, 200)
t2 = np.linspace( 0.1,  2.9, 200)
t3 = np.linspace( 3.1,  5.0, 200)
```

```python
y0 = [2, 1]
y1 = si.odeint(dydt, y0, t1)
y0 = [2, 1]
y2 = si.odeint(dydt, y0, t2)
y0 = [2, 1]
y3 = si.odeint(dydt, y0, t3)
```

```python
plt.figure(figsize=(15,5), dpi=300)
plt.plot(t1, y1[:,0], label='$t < 0$')
plt.plot(t2, y2[:,0], label='$0<t<3$', lw=3)
plt.plot(t3, y3[:,0], label='$t>3$')
plt.plot(t1, y1[:,1], color='k', lw=0.3)
plt.plot(t2, y2[:,1], color='k', lw=0.3)
plt.plot(t3, y3[:,1], color='k', lw=0.3)
plt.grid(ls='dashed', lw=0.1)
plt.xlabel('$t$')
plt.ylabel('$y(t)$')
plt.ylim(-5, 10)
plt.legend()
plt.show()
```
