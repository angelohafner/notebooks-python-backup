```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

primeira linha defino o espaço de onde até vai minha função botando a variável t_num, depois defino minha função cosseno e boto ela para aparecer(plotar).


```python

t_num = np.linspace(0, 4*np.pi)
w_num = np.cos(t_num)
plt.plot(t_num,w_num )
```

Aqui integro a função numérica de toda área que está dentro dos limites do gráfico com a função np.trapz.

```python
int_sin_num = np.trapz(w_num, t_num)
int_sin_num
```

Aqui faço a mesma integral, só que com a função cosseno ao quadrado

```python
int_sin_num = np.trapz(w_num**2, t_num)
int_sin_num
```

Aqui eu mostro o gráfico da função cosseno agora ao quadrado

```python
plt.plot(t_num, w_num**2)
```

Simbólico

Aqui defino a variável que antes estava no meio numérico para o meio simbólico usando a biblioteca sympy.

```python
t = sp.symbols("t")
w = sp.cos(t)
w
```

Agora integro a variável novamente, mas agora usando a biblioteca sympy, sp.integrate para declarar que a variável é simbólica.

```python
int_simbolica = sp.integrate(w, (t, 0, 4*sp.pi))
int_simbolica
```

Integro a mesma função simbólica, so que agora com a variável ao quadrado.

```python
int_simbolica = sp.integrate(w**2, (t, 0, 4*sp.pi))
int_simbolica
```
