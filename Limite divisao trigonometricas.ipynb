```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def num(x):
    return 1 - np.cos(3*xx)
```

```python
def den(x):
    y = (np.cos(5*x))**2 -1
    return y
```

```python
def funcao(x):
    return num(x)/den(x)
```

```python
xx = np.linspace(-0.45, 0.45, 100)
```

```python
plt.figure(figsize=(8,4), dpi=100)
plt.plot(xx, num(xx), label='num$ = 1-\\cos (3x)$', ls='--')
plt.plot(xx, den(xx), label='den$ = \\cos^2 (5x) - 1$', ls='--')
plt.plot(xx, funcao(xx), lw = 3, label=r'$\dfrac{{{\rm{num}}}}{{{\rm{den}}}} = \dfrac{{1 - \cos \left( {3x} \right)}}{{{{\cos }^2}\left( {5x} \right) - 1}}$')
plt.plot(xx, xx/xx*-9/50, label='$-\\dfrac{9}{50}$', lw=0.5, ls='--')
plt.grid()
plt.legend()
plt.show()
```
