```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

```

```python
a = 1*np.exp(1j*2*np.pi/3)
```

```python
Ia = 100 * np.exp( 1j*np.pi/2/10)
Ib = 120 * np.exp(-1j*2*np.pi/3.3)
Ic = 200 * np.exp( 1j*2*np.pi/2.7)
Iabc = [[Ia],[Ib],[Ic]]
Iabc = np.array(Iabc)
Iabc
```

```python
F = [[1, 1   ,    1],
     [1, a**2,    a],
     [1, a   , a**2]]
F = np.array(F)
F
```

```python
I012 = np.linalg.inv(F) @ Iabc
I012
```

```python
I0a = I012[0]
I0b = I0a
I0c = I0a

I1a = I012[1]
I1b = I1a * a**2
I1c = I1a * a

I2a = I012[2]
I2b = I2a * a
I2c = I2a * a**2
```

```python
max_escala = np.max([np.max(np.real(I012)), np.max(np.imag(I012))])
pd = 1
```

```python
fig, (ax0, ax1, ax2) = plt.subplots(1, 3)
fig.set_size_inches(15, 5)
fig.dpi = 150

ax0.set_xlim(-max_escala,max_escala)
ax0.set_ylim(-max_escala,max_escala)
ax0.quiver(  0,   0, np.real(I0a), np.imag(I0a), angles='xy', scale_units='xy', scale=1, color='red',   label='$a$')
ax0.quiver(-pd, -pd, np.real(I0b), np.imag(I0b), angles='xy', scale_units='xy', scale=1, color='green', label='$b$')
ax0.quiver(+pd, +pd, np.real(I0c), np.imag(I0c), angles='xy', scale_units='xy', scale=1, color='blue',  label='$c$')
ax0.legend()
ax0.set_title('Sequência Zero')
ax0.set_xlabel('Real')
ax0.set_ylabel('Imaginário')
ax0.grid()

ax1.set_xlim(-max_escala,max_escala)
ax1.set_ylim(-max_escala,max_escala)
ax1.quiver(  0,   0, np.real(I1a), np.imag(I1a), angles='xy', scale_units='xy', scale=1, color='red',   label='$a$')
ax1.quiver(  0,   0, np.real(I1b), np.imag(I1b), angles='xy', scale_units='xy', scale=1, color='green', label='$b$')
ax1.quiver(  0,   0, np.real(I1c), np.imag(I1c), angles='xy', scale_units='xy', scale=1, color='blue',  label='$c$')
ax1.legend()
ax1.set_title('Sequência Positiva')
ax1.set_xlabel('Real')
ax1.set_ylabel('Imaginário')
ax1.grid()

ax2.set_xlim(-max_escala,max_escala)
ax2.set_ylim(-max_escala,max_escala)
ax2.quiver(  0,   0, np.real(I2a), np.imag(I2a), angles='xy', scale_units='xy', scale=1, color='red',   label='$a$')
ax2.quiver(  0,   0, np.real(I2b), np.imag(I2b), angles='xy', scale_units='xy', scale=1, color='green', label='$b$')
ax2.quiver(  0,   0, np.real(I2c), np.imag(I2c), angles='xy', scale_units='xy', scale=1, color='blue',  label='$c$')
ax2.legend()
ax2.set_title('Sequência Negativa')
ax2.set_xlabel('Real')
ax2.set_ylabel('Imaginário')
ax2.grid()
```
