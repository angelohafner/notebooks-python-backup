## 1. Parâmetros Iniciais

- **Raio interno das bobinas ($ r_{\text{int}} $)**: 10 cm = 0,10 m
- **Raio externo das bobinas ($ r_{\text{ext}} $)**: 11 cm = 0,11 m
- **Raio médio ($ r_{\text{médio}} $)**:

  $$
  r_{\text{médio}} = \frac{r_{\text{int}} + r_{\text{ext}}}{2} = \frac{0{,}10 + 0{,}11}{2} = 0{,}105 \, \text{m}
  $$

- **Largura da bobina ($ w $)**:

  $$
  w = r_{\text{ext}} - r_{\text{int}} = 0{,}11 - 0{,}10 = 0{,}01 \, \text{m}
  $$

- **Número de espiras da bobina transmissora ($ N_t $)**: 10
- **Número de espiras da bobina receptora ($ N_r $)**: 10
- **Frequência de operação ($ f $)**: 5 kHz
- **Frequência angular ($ \omega $)**:

  $$
  \omega = 2\pi f = 2\pi \times 5{,}000 = 31{,}415{,}93 \, \text{rad/s}
  $$

- **Corrente na bobina transmissora ($ I_0 $)**: 1 A
- **Distância entre as bobinas ($ d $)**: 5 cm = 0,05 m

<a name="2"></a>
## 2. Cálculo das Indutâncias das Bobinas

Utilizaremos a **Fórmula de Wheeler** para calcular a indutância de bobinas planas (espirais):

$$
L = \frac{N^2 r^2}{8r + 11w}
$$

Onde:

- $ L $ é a indutância em **microhenrys ($ \mu H $)**
- $ N $ é o número de espiras
- $ r $ é o raio médio em **centímetros**
- $ w $ é a largura da bobina em **centímetros**

### Conversão das Unidades

- Raio médio em centímetros: $ r_{\text{médio}} = 0{,}105 \, \text{m} = 10{,}5 \, \text{cm} $
- Largura da bobina em centímetros: $ w = 0{,}01 \, \text{m} = 1 \, \text{cm} $

### Cálculo da Indutância da Bobina Transmissora ($ L_t $)

$$
L_t = \frac{N_t^2 \times r_{\text{médio}}^2}{8r_{\text{médio}} + 11w} = \frac{10^2 \times (10{,}5)^2}{8 \times 10{,}5 + 11 \times 1} = \frac{100 \times 110{,}25}{84 + 11} = \frac{11{,}025}{95} \approx 116{,}05 \, \mu H
$$

Convertendo para henrys:

$$
L_t = 116{,}05 \times 10^{-6} \, H = 116{,}05 \, \mu H
$$

### Cálculo da Indutância da Bobina Receptora ($ L_r $)

Como $ N_r = N_t $, temos:

$$
L_r = L_t = 116{,}05 \, \mu H
$$

<a name="3"></a>
## 3. Cálculo dos Capacitores de Ressonância

A capacitância necessária para ressonância é dada por:

$$
C = \frac{1}{(2\pi f)^2 L}
$$

### Cálculo do Capacitor de Ressonância da Bobina Transmissora ($ C_t $)

$$
C_t = \frac{1}{(2\pi f)^2 L_t} = \frac{1}{(2\pi \times 5{,}000)^2 \times 116{,}05 \times 10^{-6}} = \frac{1}{(31{,}415{,}93)^2 \times 116{,}05 \times 10^{-6}}
$$

Calculando:

$$
C_t = \frac{1}{(986{,}96 \times 10^{6}) \times 116{,}05 \times 10^{-6}} = \frac{1}{114{,}701} = 8{,}72 \times 10^{-6} \, F = 8{,}72 \, \mu F
$$

### Cálculo do Capacitor de Ressonância da Bobina Receptora ($ C_r $)

Como $ L_r = L_t $, temos:

$$
C_r = C_t = 8{,}72 \, \mu F
$$

---

<a name="4"></a>
## 4. Cálculo do Coeficiente de Acoplamento ($ k $)

O coeficiente de acoplamento entre duas bobinas circulares coaxiais é dado por:

$$
k = \frac{r_{\text{médio}}^2}{r_{\text{médio}}^2 + d^2}
$$

Substituindo os valores:

$$
k = \frac{(0{,}105)^2}{(0{,}105)^2 + (0{,}05)^2} = \frac{0{,}011025}{0{,}011025 + 0{,}0025} = \frac{0{,}011025}{0{,}013525} \approx 0{,}815
$$

---

<a name="5"></a>
## 5. Cálculo da Indutância Mútua ($ M $)

A indutância mútua é dada por:

$$
M = k \sqrt{L_t L_r}
$$

Substituindo os valores (em henrys):

$$
M = 0{,}815 \times \sqrt{116{,}05 \times 10^{-6} \times 116{,}05 \times 10^{-6}} = 0{,}815 \times 116{,}05 \times 10^{-6} = 94{,}61 \times 10^{-6} \, H = 94{,}61 \, \mu H
$$

---

<a name="6"></a>
## 6. Cálculo do Fator de Qualidade ($ Q $)

Assumindo uma resistência equivalente $ R = 1 \, \Omega $ para ambas as bobinas:

### Fator de Qualidade da Bobina Transmissora ($ Q_t $)

$$
Q_t = \frac{\omega L_t}{R_t} = \frac{31{,}415{,}93 \times 116{,}05 \times 10^{-6}}{1} = 3{,}645
$$

### Fator de Qualidade da Bobina Receptora ($ Q_r $)

$$
Q_r = Q_t = 3{,}645
$$

---

<a name="7"></a>
## 7. Cálculo da Tensão Induzida na Bobina Receptora

### Corrente em Ressonância na Bobina Transmissora ($ I_{t_{\text{res}}} $)

A corrente em ressonância é amplificada pelo fator de qualidade:

$$
I_{t_{\text{res}}} = I_0 \times Q_t = 1 \times 3{,}645 = 3{,}645 \, A
$$

### Tensão Induzida Máxima na Bobina Receptora ($ V_{\text{ind}} $)

$$
V_{\text{ind}} = \omega M I_{t_{\text{res}}} Q_r = 31{,}415{,}93 \times 94{,}61 \times 10^{-6} \times 3{,}645 \times 3{,}645
$$

Calculando:


$$
V_{\text{ind}} = 31{,}415{,}93 \times 94{,}61 \times 10^{-6} \times 13{,}288 \approx 39{,}55 \, V
$$

# Cálculo da tensão induzida na bobina receptora com ressonância

import math

# Parâmetros iniciais
r_int = 0.10  # Raio interno em metros
r_ext = 0.11  # Raio externo em metros
r_medio = (r_int + r_ext) / 2  # Raio médio em metros
w = r_ext - r_int  # Largura da bobina em metros

N_t = 10  # Número de espiras da bobina transmissora
N_r = 10  # Número de espiras da bobina receptora

f = 5e3  # Frequência em Hz
omega = 2 * math.pi * f  # Frequência angular em rad/s

I0 = 1.0  # Amplitude da corrente na bobina transmissora em A
d = 0.05  # Distância entre as bobinas em metros

# Cálculo das indutâncias usando a Fórmula de Wheeler
# Converter para centímetros
r_medio_cm = r_medio * 100
w_cm = w * 100

# Indutância da bobina transmissora (em µH)
L_t_uH = (N_t ** 2 * r_medio_cm ** 2) / (8 * r_medio_cm + 11 * w_cm)
L_t = L_t_uH * 1e-6  # Convertendo para Henries

# Indutância da bobina receptora (em µH)
L_r_uH = (N_r ** 2 * r_medio_cm ** 2) / (8 * r_medio_cm + 11 * w_cm)
L_r = L_r_uH * 1e-6  # Convertendo para Henries

# Cálculo dos capacitores de ressonância
C_t = 1 / ((2 * math.pi * f) ** 2 * L_t)
C_r = 1 / ((2 * math.pi * f) ** 2 * L_r)

# Cálculo do coeficiente de acoplamento (k)
k = (r_medio ** 2) / (r_medio ** 2 + d ** 2)

# Cálculo da indutância mútua (M)
M = k * math.sqrt(L_t * L_r)

# Assumindo fator de qualidade alto (Q)
# Cálculo do fator de qualidade (Q_t e Q_r) - estimativa
R_t = 1  # Resistência equivalente da bobina transmissora (Ohms)
R_r = 1  # Resistência equivalente da bobina receptora (Ohms)
Q_t = omega * L_t / R_t
Q_r = omega * L_r / R_r

# Corrente amplificada pela ressonância na bobina transmissora
I_t_res = I0 * Q_t

# Tensão induzida na bobina receptora em ressonância
Vind_max = omega * M * I_t_res * Q_r

# Conversão de unidades para exibição
L_t_mH = L_t * 1e3
L_r_mH = L_r * 1e3
C_t_uF = C_t * 1e6
C_r_uF = C_r * 1e6
M_uH = M * 1e6  # Indutância mútua em µH

# Exibir os resultados
print(f"=== Resultados dos Cálculos com Ressonância ===")
print(f"Raio médio da bobina (r_medio): {r_medio:.3f} m")

print(f"Frequência angular (omega): {omega:.2f} rad/s")

print("\nBobina Transmissora:")
print(f"Número de espiras (N_t): {N_t}")
print(f"Indutância (L_t): {L_t_mH:.6f} mH")
print(f"Capacitância de ressonância (C_t): {C_t_uF:.6f} µF")
print(f"Fator de qualidade (Q_t): {Q_t:.2f}")
print(f"Corrente em ressonância (I_t_res): {I_t_res:.2f} A")

print("\nBobina Receptora:")
print(f"Número de espiras (N_r): {N_r}")
print(f"Indutância (L_r): {L_r_mH:.6f} mH")
print(f"Capacitância de ressonância (C_r): {C_r_uF:.6f} µF")
print(f"Fator de qualidade (Q_r): {Q_r:.2f}")

print(f"\nCoeficiente de acoplamento (k): {k:.4f}")
print(f"Indutância mútua (M): {M_uH:.6f} µH")
print(f"Tensão induzida máxima na bobina receptora (Vind_max): {Vind_max:.3f} V")


````python
# Cálculo da tensão induzida na bobina receptora com ressonância

import math

# Parâmetros iniciais
r_int = 0.10  # Raio interno em metros
r_ext = 0.11  # Raio externo em metros
r_medio = (r_int + r_ext) / 2  # Raio médio em metros
w = r_ext - r_int  # Largura da bobina em metros

N_t = 10  # Número de espiras da bobina transmissora
N_r = 10  # Número de espiras da bobina receptora

f = 5e3  # Frequência em Hz
omega = 2 * math.pi * f  # Frequência angular em rad/s

I0 = 1.0  # Amplitude da corrente na bobina transmissora em A
d = 0.05  # Distância entre as bobinas em metros

# Cálculo das indutâncias usando a Fórmula de Wheeler
# Converter para centímetros
r_medio_cm = r_medio * 100
w_cm = w * 100

# Indutância da bobina transmissora (em µH)
L_t_uH = (N_t ** 2 * r_medio_cm ** 2) / (8 * r_medio_cm + 11 * w_cm)
L_t = L_t_uH * 1e-6  # Convertendo para Henries

# Indutância da bobina receptora (em µH)
L_r_uH = (N_r ** 2 * r_medio_cm ** 2) / (8 * r_medio_cm + 11 * w_cm)
L_r = L_r_uH * 1e-6  # Convertendo para Henries

# Cálculo dos capacitores de ressonância
C_t = 1 / ((2 * math.pi * f) ** 2 * L_t)
C_r = 1 / ((2 * math.pi * f) ** 2 * L_r)

# Cálculo do coeficiente de acoplamento (k)
k = (r_medio ** 2) / (r_medio ** 2 + d ** 2)

# Cálculo da indutância mútua (M)
M = k * math.sqrt(L_t * L_r)

# Assumindo fator de qualidade alto (Q)
# Cálculo do fator de qualidade (Q_t e Q_r) - estimativa
R_t = 1  # Resistência equivalente da bobina transmissora (Ohms)
R_r = 1  # Resistência equivalente da bobina receptora (Ohms)
Q_t = omega * L_t / R_t
Q_r = omega * L_r / R_r

# Tensão na bobina transmissora em ressonância
V_t = I0 * R_t  # Tensão aplicada na bobina transmissora

# Corrente amplificada pela ressonância na bobina transmissora
I_t_res = I0 * Q_t

# Tensão amplificada na bobina transmissora
V_t_res = I_t_res * R_t

# Tensão induzida na bobina receptora em ressonância
Vind_max = omega * M * I_t_res * Q_r

# Conversão de unidades para exibição
L_t_mH = L_t * 1e3
L_r_mH = L_r * 1e3
C_t_uF = C_t * 1e6
C_r_uF = C_r * 1e6
M_uH = M * 1e6  # Indutância mútua em µH

# Exibir os resultados
print(f"=== Resultados dos Cálculos com Ressonância ===")
print(f"Raio médio da bobina (r_medio): {r_medio:.3f} m")
print(f"Largura da bobina (w): {w:.3f} m")
print(f"Frequência angular (omega): {omega:.2f} rad/s")

print("\nBobina Transmissora:")
print(f"Número de espiras (N_t): {N_t}")
print(f"Indutância (L_t): {L_t_mH:.6f} mH")
print(f"Capacitância de ressonância (C_t): {C_t_uF:.6f} µF")
print(f"Fator de qualidade (Q_t): {Q_t:.2f}")
print(f"Corrente em ressonância (I_t_res): {I_t_res:.2f} A")

print("\nBobina Receptora:")
print(f"Número de espiras (N_r): {N_r}")
print(f"Indutância (L_r): {L_r_mH:.6f} mH")
print(f"Capacitância de ressonância (C_r): {C_r_uF:.6f} µF")
print(f"Fator de qualidade (Q_r): {Q_r:.2f}")

print(f"\nCoeficiente de acoplamento (k): {k:.4f}")
print(f"Indutância mútua (M): {M_uH:.6f} µH")
print(f"Tensão induzida máxima na bobina receptora (Vind_max): {Vind_max:.3f} V")
````
