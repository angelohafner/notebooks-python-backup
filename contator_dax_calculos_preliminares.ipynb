```python
import math

class Contactor:
    MU0 = 4 * math.pi * 1e-7  # H/m

    def __init__(self, n_espiras: int, corrente: float, area: float, entreferro: float):
        self.n = n_espiras
        self.i = corrente
        self.area = area
        self.g = entreferro

    def densidade_fluxo(self) -> float:
        # B = μ₀·N·I / g
        return (Contactor.MU0 * self.n * self.i) / self.g

    def forca_por_fluxo(self) -> float:
        # F = B²·A / (2·μ₀)
        B = self.densidade_fluxo()
        return (B**2 * self.area) / (2 * Contactor.MU0)

    def forca_por_corrente(self) -> float:
        # F = μ₀·(N·I)²·A / (2·g²)
        return Contactor.MU0 * (self.n * self.i)**2 * self.area / (2 * self.g**2)
```

```python
# parametros
N = 200         # nr de espiras da bobina
I = 10          # corrente da bobina em A
A = (50e-2)**2  # área da seção transversal do entreferro em m²
g = 2e-2        # gap de ar antes do fechamento das bobinas em metros

cont = Contactor(N, I, A, g)
print("Densidade de fluxo B:", cont.densidade_fluxo(), "T")
print("Forca (via fluxo):", cont.forca_por_fluxo(), "N")
print("Forca (via corrente):", cont.forca_por_corrente(), "N")
```
