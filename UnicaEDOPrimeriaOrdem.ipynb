```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
t, a = sp.symbols('t a')
x = sp.Function('x')(t)
```

```python
edo = sp.Eq(x.diff(t), 3*x)
```

```python
xSol = sp.dsolve(edo, func=x, ics={x.subs(t, 0): 2})
xSol.rhs
```

```python
fx = sp.lambdify(t, xSol.rhs)
```

```python
t_num = np.linspace(0, 1, 2000)
x_num = fx(t_num)
```

```python
plt.plot(t_num, x_num)
```
