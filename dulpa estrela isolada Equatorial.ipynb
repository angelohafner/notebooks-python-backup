```python
import numpy as np

def polar_form(value):
    magnitude = abs(value)
    angle = np.angle(value, deg=True)
    return f"{magnitude:.4f} ∠ {angle:.2f}°"

# Definição das tensões de fase
V3f = 34.5e3  # Tensão de linha em V
Van = V3f / np.sqrt(3) * np.exp(-1j * np.radians(30))
Vbn = V3f / np.sqrt(3) * np.exp(-1j * np.radians(150))
Vcn = V3f / np.sqrt(3) * np.exp(1j * np.radians(90))

# Impedância dos ramos
z = -1j * 106
Za1, Za2 = 10**10 * z, z
Za = 1 / (1/Za1 + 1/Za2)

Zb1, Zb2 = z, z
Zb = 1 / (1/Zb1 + 1/Zb2)

Zc1, Zc2 = z, z
Zc = 1 / (1/Zc1 + 1/Zc2)

# Montagem da matriz de impedâncias
Z_matrix = np.array([[Za + Zc, -Zc],
                     [-Zc, Zb + Zc]])

V_vector = np.array([V3f * np.exp(1j * np.radians(120)),
                     V3f * np.exp(-1j * np.radians(120))])

# Cálculo das correntes Iαβ
I_alpha_beta = np.linalg.inv(Z_matrix) @ V_vector

# Cálculo das correntes e tensões em cada ramo
Ia = -I_alpha_beta[0]
Ib = I_alpha_beta[1]
Ic = I_alpha_beta[0] - I_alpha_beta[1]

Vao = Ia * Za
Vbo = Ib * Zb
Vco = Ic * Zc

# Cálculo da tensão de neutro
Vneutro = (Vao + Vbo + Vco) / 3

# Cálculo das correntes nos sub-ramos
Ia1 = Vao / Za1
Ia2 = Vao / Za2

Ib1 = Vbo / Zb1
Ib2 = Vbo / Zb2

Ic1 = Vco / Zc1
Ic2 = Vco / Zc2

# Exibição dos resultados
print(f"I_alpha = {polar_form(I_alpha_beta[0])} A")
print(f"I_beta = {polar_form(I_alpha_beta[1])} A")
print(f"Ia = {polar_form(Ia)} A")
print(f"Ib = {polar_form(Ib)} A")
print(f"Ic = {polar_form(Ic)} A")
print(f"Ia+Ib+Ic = {polar_form(Ia+Ib+Ic)} A")
print(f"Vao = {polar_form(Vao)} V")
print(f"Vbo = {polar_form(Vbo)} V")
print(f"Vco = {polar_form(Vco)} V")
print(f"Tensão de neutro = {polar_form(Vneutro)} V")
print(f"Ia1 = {polar_form(Ia1)} A, Ia2 = {polar_form(Ia2)} A")
print(f"Ib1 = {polar_form(Ib1)} A, Ib2 = {polar_form(Ib2)} A")
print(f"Ic1 = {polar_form(Ic1)} A, Ic2 = {polar_form(Ic2)} A")
```
