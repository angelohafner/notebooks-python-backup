```python
# Verificar se o pint já está instalado; instalar apenas se necessário
try:
    from pint import UnitRegistry
    print("pint já está instalado.")
except ImportError:
    print("Instalando pint...")
    !pip install pint
    from pint import UnitRegistry

# Importar outras bibliotecas
import math

# Criar um registro de unidades
ureg = UnitRegistry()

# Definir constantes com unidades
omega_1 = 2 * math.pi * 60 * ureg.radian / ureg.second  # Frequência angular
S_trafo = (1.5 * 10**6 / 2) * ureg.VA  # Potência aparente
V_trafo = 380 * ureg.volt  # Tensão nominal
I_trafo = S_trafo / (math.sqrt(3) * V_trafo)  # Corrente nominal

# Banco de Capacitores Base
Q_capacitores = 59 * 10**3 * ureg.VA  # Potência reativa (usando VA como proxy para VAr)
V_capacitores = 380 * ureg.volt  # Tensão do capacitor
X_capacitores = (V_capacitores**2) / Q_capacitores  # Reatância capacitiva
C_capacitores = 1 / (omega_1 * X_capacitores)  # Capacitância
I_capacitores = (V_capacitores / (math.sqrt(3)) / X_capacitores)  # Corrente nominal do capacitor

# Banco de Capacitores Utilizado no Filtro
V_nom_cap = 1500 * ureg.volt  # Tensão nominal do capacitor
Q_nom_cap = V_nom_cap**2 / X_capacitores # Potência reativa nominal do capacitor
I_nom_cap = Q_nom_cap / (math.sqrt(3) * V_nom_cap)  # Corrente nominal do capacitor
I_5 = 0.25 * I_trafo  # Corrente da 5ª harmônica
I_7 = 0.09 * I_trafo  # Corrente da 7ª harmônica
I_capacTotal = (I_nom_cap**2 + I_5**2)**(1/2)  # Corrente total do capacitor
razao_corrente = I_capacTotal / I_nom_cap  # Razão I_capacTotal / I_nome_cap

# Indutores para Harmônicos 5 e 7
L_5 = 1 / ((5 * omega_1)**2 * C_capacitores)  # Indutância para 5ª harmônica
L_7 = 1 / ((7 * omega_1)**2 * C_capacitores)  # Indutância para 7ª harmônica

# Exibição dos Resultados com Prefixos
print("\n=== Dados do Sistema ===")
print(f"ω₁ = {omega_1:.3f}")
print(f"S_trafo = {S_trafo:.1f}")
print(f"V_trafo = {V_trafo}")
print(f"I_trafo = {I_trafo.to(ureg.ampere):.1f} (≈ {I_trafo.to(ureg.kiloampere):.1f})")

print("\n=== Banco de Capacitores Base ===")
print(f"Q_capacitores = {Q_capacitores:.1f} (interpretado como VAr)")
print(f"V_capacitores = {V_capacitores}")
print(f"X_capacitores = {X_capacitores:.2f}")
print(f"I_capacitores = {I_capacitores:.2f}")
print(f"C_capacitores = {C_capacitores.to(ureg.farad):.6f} (≈ {C_capacitores.to(ureg.millifarad):.2f})")

print("\n=== Banco de Capacitores Utilizado no Filtro ===")
print(f"V_max_cap = {V_nom_cap.to(ureg.volt)}")
print(f"I_nom_cap = {I_nom_cap.to(ureg.ampere):.1f}")
print(f"I_5 = {I_5.to(ureg.ampere):.1f}")
print(f"I_7 = {I_7.to(ureg.ampere):.1f}")
print(f"I_capacTotal = {I_capacTotal.to(ureg.ampere):.1f}")
print(f"I_capacTotal / I_nome_cap = {razao_corrente:.2f~P}")

print("\n=== Indutores para Harmônicos 5 e 7 ===")
print(f"L_5 = {L_5:.7f} (≈ {L_5.to(ureg.microhenry):.1f})")
print(f"L_7 = {L_7:.7f} (≈ {L_7.to(ureg.microhenry):.1f})")
```
