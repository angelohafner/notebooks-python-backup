```python
import numpy as np
import matplotlib.pyplot as plt
import numpy as np
```

```python
def binaryStep(x):
    ''' It returns '0' is the input is less then zero otherwise it returns one '''
    return np.heaviside(x,1)
```

```python
x = np.linspace(-10, 10)
plt.plot(x, binaryStep(x))
plt.axis('tight')
plt.title('Função de Ativação: Degrau Unitário')
plt.xlabel('entrada')
plt.ylabel('saída')
plt.show()
```
