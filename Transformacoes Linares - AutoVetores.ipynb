```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive,interact
from scipy.linalg import null_space
```

```python
npts = 11
limte = 6
```

```python
x = np.linspace(-limte,limte,npts)
y = x
v = np.zeros((2,npts*npts))
# aqui são os vetores do domínio
for ii in range(npts):
    for jj in range(npts):
        v[0,jj+ii*npts] = x[ii]
        v[1,jj+ii*npts] = y[jj]
```

```python
def transfLinear(a11=-0.5, a12=1, a21=1, a22=1):
    plt.figure(dpi=200, figsize=(5,5))

    plt.scatter(v[0,:],v[1,:], s=10, alpha=0.5, label='domínio', lw=0)

    T = np.array( [ [a11, a12],
                    [a21, a22] ] )

    w = T @ v

    plt.scatter(w[0,:],w[1,:], s=10, alpha=1.0, label='imagem', lw=0)
    plt.xlim(-2*limte,2*limte)
    plt.ylim(-2*limte,2*limte)

    aVals, aVets = np.linalg.eig(T)
    

    print("aVals =\n", aVals)
    print("aVets =\n", aVets)

    aVets = limte/2 * aVets
    aVetTransf = T @ aVets 

    plt.quiver(aVets[0,0], aVets[1,0], angles='xy', scale_units='xy', scale=1, label='av1', width = 8e-3, alpha = 0.5, color='red')
    plt.quiver(aVetTransf[0,0], aVetTransf[1,0], angles='xy', scale_units='xy', scale=1, label='Tav1', width = 4e-3, alpha = 1.0, color='red')

    plt.quiver(aVets[0,1], aVets[1,1], angles='xy', scale_units='xy', scale=1, label='av2', width = 8e-3, alpha = 0.5, color='blue')
    plt.quiver(aVetTransf[0,1], aVetTransf[1,1], angles='xy', scale_units='xy', scale=1, label='Tav2', width = 4e-3, alpha = 1.0, color='blue')

    nucleo = null_space(T)
    print("nucleo = \n", nucleo)
    nucleo = nucleo * limte/2
    #plt.quiver(nucleo[0,0], nucleo[1,0], angles='xy', scale_units='xy', scale=1, label='Tav2', width = 8e-3, alpha = 1.0, color='green')

    plt.legend(fontsize=4)
```

```python
interactive( transfLinear , a11 = (-2, 2,  0.1),  a12 = (-2, 2,  0.1),
                            a21 = (-2, 2,  0.1),  a22 = (-2, 2,  0.1) )
```
