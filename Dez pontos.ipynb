```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
v0 = np.array([0,0,1])
v1 = np.array([3,6,2])
v2 = np.array([5,7,3])
vetorc = np.array([v0, v1, v2])
vetorc = vetorc.T
vetor = vetorc[0:2,:]
vetor
```

```python
cores = vetorc[-1,:]
cores
```

```python
M = np.array([[0.87, 0.50],
                      [-0.50,0.87]])
M

```

```python
ww = M@vetor
ww
```

```python
plt.scatter(vetor[0,:], vetor[1,:], label='v', c=cores)
plt.scatter(ww[0,:], ww[1,:], label='w', c=cores)
plt.legend()
plt.show()
```
