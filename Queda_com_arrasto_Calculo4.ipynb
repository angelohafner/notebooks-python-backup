```python
import sympy as sp # biblioteca simbólica
import matplotlib.pyplot as plt # gráficos
import numpy as np # numérica
t = sp.symbols('t')
v = sp.Function('v')
```

**negrito**

```python
ode = sp.Eq(sp.Derivative(v(t),t,1) , ( 49-v(t) ) / 5)
ode
```

```python
sol = sp.dsolve(ode,v(t),ics={v(0):40})
sol
```

```python
v_sym = sol.args[1]
```

```python
sol
```

```python
v_num = sp.lambdify(t,v_sym) 
```

```python
tt = np.linspace(0,30,100)
plt.plot(tt,v_num(tt),linewidth=3)
plt.grid('on')
plt.xlabel('Tempo [s]')
plt.ylabel('Velocidade [m/s]')
plt.show()
```

```python
ttt = np.linspace(0,30,10)
vvv = np.linspace(40,60,10)
ttt,vvv = np.meshgrid(ttt,vvv)
deriv = ( 49-vvv ) / 5
aa = np.arctan(  deriv )
u = np.cos(aa)
v = np.sin(aa)
plt.quiver(ttt,vvv,u,v)
plt.grid('on')
plt.xlabel('Tempo [s]')
plt.ylabel('Velocidade [m/s]')

plt.show()
```
