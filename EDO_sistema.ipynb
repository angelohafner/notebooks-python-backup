```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def dX_dt(X,t,b,k):
    v = X[0]
    x = X[1]
    
    dx_dt = v
    dv_dt = -b*v -k*x + 3
    
    return [dv_dt,dx_dt]
```

```python
b = 3
k = 2
T = np.matrix([[-b,-k],[1,0]])
t = np.linspace(0,10,1000)
y0 = [0,0]
from scipy.integrate import odeint
sol = odeint(dX_dt, y0, t,args=(b, k))
```

```python
plt.style.use(['science','notebook','grid'])
plt.plot(t,sol[:,0],label='$v$')
plt.plot(t,sol[:,1],label='$x$')
plt.xlabel('$t$ [s]')
plt.ylabel('$v$ [m/s]')
#plt.title('Velocidade [m/s]')
plt.legend()
plt.show()
```

```python
y0 = [0,0]
P,D = np.linalg.eig(T)
```
